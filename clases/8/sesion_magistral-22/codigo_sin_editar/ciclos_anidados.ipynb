### Ejemplo 1

Hacer un programa que dada `N`, imprimima la serie:

```
1, 2, 2, 3, 3, 3, 4, 4, 4, 4, ...
```

````python
N = 4
for i in range(1,N + 1):
    # print("i = " + str(i))
    for j in range(i):
        # print("   j = " + str(j))
        print(i, end = " ")
````

### Ejemplo 2

Hacer un rectangulo que imprima `NxM` caritas felices.


````python
carita = "^_^"

N = 2
M = 3

for i in range(N):
    # print(i, end = ": ")
    for j in range(M):
        # print(f"i = {i}, j = {j}")
        print(carita, end = "   ")
    print()
# print(total)
````

````python
carita = "^_^"

N = 2
M = 3

i = 0 
j = 0
total = 0
while i < N:
    # print(i, end = ": ")
    while j < M:
        # print(f"i = {i}, j = {j}")
        print(carita, end = "   ")
        j += 1
        total += 1
    i += 1 
    j = 0 # No olvidar inicia
    print()
# print(total)
````

### Ejemplo 4

Hacer un programa que muestre la escalera de Supermario

````python
N = 10
espacios = N - 1
numerales = 1

# Filas (i)
for i in range(N):
    # Espacios
    for j in range(espacios):
        print(" ", end = "")
    # Numerales
    for j in range(numerales):
        print("#", end = "")
    print() # Nueva linea
    # Actualizacion
    espacios -= 1
    numerales += 1
````
