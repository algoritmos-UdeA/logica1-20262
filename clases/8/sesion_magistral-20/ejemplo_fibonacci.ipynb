# Ejemplo con ciclos

En esta ejemplo vamos a ver el uso de variables temporales.

## Enunciado

Hacer un programa que genere los N primeros terminos de una serie Fibonacci.


## Planteamiento

### Definicion de variables



*   **N**: Numero de terminos de la serie Fibonacci a mostrar
*   **term0**: Termino antepenultimo.
*   **term1**: Termino penultimo.
*   **term2**: Termino actual (mostrar de la serie).
*   **i**: Control del ciclo



```python
# Inicializacion
term0 = 0
term1 = 1
term2 = 0
i = 0

# Entradas
N = int(input("Digite la cantidad de terminos a visualizar: "))

# Proceso y salidas
if(N <= 0):
  print("No hay numeros para mostrar")
elif(N == 1):
  # Solo un termino
  print(f"-> {term0}")
elif (N == 2):
  # Solo dos terminos
  print(f"-> {term0}")
  print(f"-> {term1}")
else:
  # Tres o mas terminos
  # Primeros dos terminos (fuera del ciclo)
  print(f"-> {term0}")
  print(f"-> {term1}")
  for i in range(2,N):
    # print(i)
    # Demas terminos
    term2 = term1 + term0
    print(f"-> {term2}")
    term0 = term1
    term1 = term2
```
