## Parte 1 — El problema y un caso calculado a mano

### Enunciado

> Se desea un programa para jugar una **partida** de piedra, papel o tijera entre el usuario y la máquina:
>
> * Al empezar, el programa pide el nombre del usuario.
> * En cada **ronda**, el usuario elige su jugada (1 piedra, 2 papel, 3 tijera; si escribe otro número, se le vuelve a pedir) y la máquina elige la suya al azar.
> * Piedra le gana a tijera, tijera le gana a papel y papel le gana a piedra. Si las dos jugadas son iguales, es empate.
> * Quien gana la ronda suma un punto; en un empate, nadie suma.
> * Después de cada ronda se muestran las jugadas, quién ganó y el marcador.
> * **Gana la partida quien primero gane 3 rondas.**

### Un caso para verificar a mano

Un buen enunciado permite calcular un ejemplo **antes** de programar. Ese ejemplo se usará después para probar el programa completo. Ana juega contra la máquina, y estas son las jugadas de cada ronda:

| Ronda | Ana | Máquina | ¿Quién gana la ronda? | Marcador |
|---|---|---|---|---|
| 1 | piedra | papel | Máquina (papel le gana a piedra) | Ana 0 - 1 Máquina |
| 2 | piedra | tijera | Ana (piedra le gana a tijera) | Ana 1 - 1 Máquina |
| 3 | tijera | piedra | Máquina | Ana 1 - 2 Máquina |
| 4 | papel | papel | Empate: nadie suma | Ana 1 - 2 Máquina |
| 5 | papel | piedra | Ana | Ana 2 - 2 Máquina |
| 6 | tijera | papel | Ana | Ana 3 - 2 Máquina |

Ana llega a 3 rondas ganadas en la ronda 6: **gana la partida Ana**. La ronda 4 se jugó, pero no cambió el marcador.

## Parte 2 — El diseño: de los datos y las tareas a las funciones

Antes de escribir código se responden dos preguntas: **¿qué datos hay que guardar?** y **¿qué tareas hay que hacer?** Las respuestas salen del enunciado: los sustantivos señalan los datos y los verbos señalan las tareas.

### 1. Los datos

| En el enunciado | ¿Cómo se guarda? | Tipo | ¿Cambia durante la partida? |
|---|---|---|---|
| El nombre del usuario | Variable `nombre_usuario` | `str` | No |
| El nombre de la máquina | Variable `nombre_maquina` | `str` | No |
| Los puntos de cada uno | Variables `puntos_usuario` y `puntos_maquina` | `int` | Sí: en cada ronda que alguien gana |
| Una jugada | Un número: 1, 2 o 3 | `int` | Es distinta en cada ronda |
| El resultado de una ronda | Un número: 0, 1 o 2 | `int` | Es distinto en cada ronda |
| Cuántas rondas hay que ganar | Constante `RONDAS_PARA_GANAR` | `int` | No |

Los números de las jugadas y de los resultados se guardan en **constantes con nombre** (`PIEDRA`, `GANA_USUARIO`…), para no tener que recordar que `2` es papel o que `0` es empate.

Hay que fijarse en algo que más adelante va a pesar: **un jugador no existe como un solo dato**. Su nombre va en una variable y sus puntos en otra, y lo único que las relaciona es cómo se llaman.

### 2. Las tareas

Cada tarea se vuelve una función. Para cada una se decide **qué recibe** y **qué retorna**.

| Tarea del enunciado | Función | Recibe | Retorna |
|---|---|---|---|
| Revisar que la jugada sea 1, 2 o 3 | `es_jugada_valida` | Una jugada | `True` o `False` |
| Mostrar una jugada con palabras | `nombre_jugada` | Una jugada | `"piedra"`, `"papel"` o `"tijera"` |
| Decidir quién gana la ronda | `definir_ganador` | Las dos jugadas | `EMPATE`, `GANA_USUARIO` o `GANA_MAQUINA` |
| Que la máquina elija al azar | `jugar_maquina` | Nada | Una jugada |
| Sumar un punto a quien gana | `sumar_si_gana` | Unos puntos, el resultado y cuál resultado suma | Los puntos nuevos |
| Saber si la partida terminó | `partida_terminada` | Los puntos de los dos y la meta | `True` o `False` |
| Mostrar las jugadas y quién ganó | `texto_ronda` | Los dos nombres, las dos jugadas y el resultado | Un texto |
| Mostrar el marcador | `texto_marcador` | Los dos nombres y los dos puntajes | Un texto |
| Decir quién ganó la partida | `nombre_ganador` | Los dos nombres y los dos puntajes | Un nombre |

### 3. Tres decisiones de diseño

1. **Las funciones no usan `input` ni `print`.** Reciben datos y retornan resultados. Por eso las que "muestran" algo en realidad **retornan un texto**, y quien las llama decide imprimirlo. Así cada función se puede probar con `assert`.
2. **La jugada de la máquina llega como parámetro.** `definir_ganador` no elige al azar: recibe las dos jugadas. El azar queda en una sola función, `jugar_maquina`, y todo lo demás se puede probar sin depender de la suerte.
3. **Las reglas van aparte de la partida.** Qué le gana a qué no depende de quién juega ni de cómo va el marcador. Son dos grupos de funciones.

## Parte 3 — Paso 1: las reglas del juego

### Las constantes

Se definen una sola vez, al comienzo, y no se cambian.

````python
import random

# Jugadas
PIEDRA = 1
PAPEL = 2
TIJERA = 3

# Resultados de una ronda
EMPATE = 0
GANA_USUARIO = 1
GANA_MAQUINA = 2
````

### `es_jugada_valida`

````python
def es_jugada_valida(jugada):
    """
    Supone: jugada es un entero.
    Garantiza: retorna True si jugada es PIEDRA, PAPEL o TIJERA, y False
    si no.
    """
    return PIEDRA <= jugada <= TIJERA
````

`PIEDRA <= jugada <= TIJERA` es una comparación encadenada: equivale a `jugada >= PIEDRA and jugada <= TIJERA`. La función retorna directamente el resultado de la comparación, que ya es `True` o `False`.

Se prueban los dos **límites** del rango y un valor **fuera** por cada lado.

````python
# es_jugada_valida
assert es_jugada_valida(PIEDRA) == True         # límite inferior
assert es_jugada_valida(TIJERA) == True         # límite superior
assert es_jugada_valida(0) == False             # fuera del rango
assert es_jugada_valida(4) == False             # fuera del rango
print("es_jugada_valida: pruebas correctas")
````

### `nombre_jugada`

````python
def nombre_jugada(jugada):
    """
    Supone: jugada es PIEDRA, PAPEL o TIJERA.
    Garantiza: retorna "piedra", "papel" o "tijera".
    """
    if jugada == PIEDRA:
        nombre = "piedra"
    elif jugada == PAPEL:
        nombre = "papel"
    else:
        nombre = "tijera"
    return nombre
````

La especificación dice *Supone: jugada es PIEDRA, PAPEL o TIJERA*. La función **confía** en eso: si la jugada no es piedra ni papel, responde `"tijera"` sin revisar más. Revisar que la jugada sea válida es tarea de `es_jugada_valida`, y se hace **antes**.

````python
# nombre_jugada
assert nombre_jugada(PIEDRA) == "piedra"
assert nombre_jugada(PAPEL) == "papel"
assert nombre_jugada(TIJERA) == "tijera"
print("nombre_jugada: pruebas correctas")
````

### `definir_ganador`

Es la regla central del juego. Con tres jugadas posibles para cada uno hay **nueve casos**:

| Usuario ↓ / Máquina → | piedra | papel | tijera |
|---|---|---|---|
| **piedra** | empate | gana la máquina | gana el usuario |
| **papel** | gana el usuario | empate | gana la máquina |
| **tijera** | gana la máquina | gana el usuario | empate |

En vez de escribir nueve condiciones, se agrupan: primero el empate (jugadas iguales), después los tres casos en que gana el usuario, y **todo lo demás** es que gana la máquina.

````python
def definir_ganador(jugada_usuario, jugada_maquina):
    """
    Decide quién gana una ronda: piedra le gana a tijera, tijera le gana
    a papel y papel le gana a piedra.

    Supone: las dos jugadas son PIEDRA, PAPEL o TIJERA.
    Garantiza: retorna EMPATE, GANA_USUARIO o GANA_MAQUINA.
    """
    if jugada_usuario == jugada_maquina:
        resultado = EMPATE
    elif ((jugada_usuario == PIEDRA and jugada_maquina == TIJERA)
          or (jugada_usuario == TIJERA and jugada_maquina == PAPEL)
          or (jugada_usuario == PAPEL and jugada_maquina == PIEDRA)):
        resultado = GANA_USUARIO
    else:
        resultado = GANA_MAQUINA
    return resultado
````

> 🔮 **Prediga.** ¿Qué número imprime cada línea? Las constantes valen `EMPATE = 0`, `GANA_USUARIO = 1`, `GANA_MAQUINA = 2`.

````python
print(definir_ganador(PIEDRA, TIJERA))
print(definir_ganador(PIEDRA, PAPEL))
print(definir_ganador(PAPEL, PAPEL))
print(definir_ganador(PIEDRA, TIJERA) == GANA_USUARIO)
````

Como solo hay nueve casos, se prueban **todos**.

### `sumar_si_gana`

````python
def sumar_si_gana(puntos, resultado, resultado_que_suma):
    """
    Supone: puntos >= 0; resultado es EMPATE, GANA_USUARIO o GANA_MAQUINA.
    Garantiza: retorna puntos + 1 si resultado es resultado_que_suma, y
    puntos si no.
    """
    if resultado == resultado_que_suma:
        puntos = puntos + 1
    return puntos
````

### `partida_terminada`

````python
def partida_terminada(puntos_usuario, puntos_maquina, rondas_para_ganar):
    """
    Garantiza: retorna True si alguno de los dos ya ganó rondas_para_ganar
    rondas, y False si no.
    """
    return (puntos_usuario == rondas_para_ganar
            or puntos_maquina == rondas_para_ganar)
````

Se prueba al comienzo, **a una ronda de terminar** (el caso límite) y con cada uno de los dos ganando.

````python
# partida_terminada (a 3 rondas)
assert partida_terminada(0, 0, 3) == False      # recién empieza
assert partida_terminada(2, 2, 3) == False      # a una ronda de terminar
assert partida_terminada(3, 1, 3) == True       # ganó el usuario
assert partida_terminada(0, 3, 3) == True       # ganó la máquina
print("partida_terminada: pruebas correctas")
````

### `nombre_ganador`

````python
def nombre_ganador(nombre_usuario, puntos_usuario, nombre_maquina, puntos_maquina):
    """
    Supone: la partida terminó (los puntos no son iguales).
    Garantiza: retorna el nombre de quien tiene más puntos.
    """
    if puntos_usuario > puntos_maquina:
        nombre = nombre_usuario
    else:
        nombre = nombre_maquina
    return nombre
````

Para decir un nombre, la función necesita **cuatro datos**: los dos nombres y los dos puntajes. *Supone* que la partida terminó, así que los puntos no pueden ser iguales.

````python
# nombre_ganador
assert nombre_ganador("Ana", 3, "Máquina", 1) == "Ana"
assert nombre_ganador("Ana", 2, "Máquina", 3) == "Máquina"
print("nombre_ganador: pruebas correctas")
````

### `texto_ronda`

````python
def texto_ronda(nombre_usuario, jugada_usuario, nombre_maquina, jugada_maquina, resultado):
    """
    Supone: las jugadas son válidas y resultado es el que les corresponde.
    Garantiza: retorna una línea que describe la ronda, por ejemplo
    "Ana: piedra | Máquina: tijera | gana Ana".
    """
    texto = (nombre_usuario + ": " + nombre_jugada(jugada_usuario) + " | "
             + nombre_maquina + ": " + nombre_jugada(jugada_maquina) + " | ")
    if resultado == EMPATE:
        texto = texto + "empate"
    elif resultado == GANA_USUARIO:
        texto = texto + "gana " + nombre_usuario
    else:
        texto = texto + "gana " + nombre_maquina
    return texto
````

Esta función recibe **cinco parámetros** y usa una función de las reglas, `nombre_jugada`. No imprime: arma el texto y lo retorna.

## Parte 5 — La integración: el programa principal

![Mapa: paso 3, el programa principal](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/mapa_estructurada_paso3.png)

Todas las funciones están probadas. El programa principal las reúne, y tiene una responsabilidad que ninguna función puede asumir: **guardar los datos de la partida** mientras dura.

Para que el notebook siempre dé la misma salida, aquí las jugadas no se piden por teclado ni se eligen al azar: están escritas en dos arreglos, con el caso calculado a mano en la parte 1.

````python
RONDAS_PARA_GANAR = 3


def main():
    # Las jugadas del caso calculado a mano
    jugadas_usuario = [PIEDRA, PIEDRA, TIJERA, PAPEL, PAPEL, TIJERA]
    jugadas_maquina = [PAPEL, TIJERA, PIEDRA, PAPEL, PIEDRA, PAPEL]

    # Los datos de la partida: cuatro variables
    nombre_usuario = "Ana"
    nombre_maquina = "Máquina"
    puntos_usuario = 0
    puntos_maquina = 0

    i = 0
    while not partida_terminada(puntos_usuario, puntos_maquina, RONDAS_PARA_GANAR):
        jugada_usuario = jugadas_usuario[i]
        jugada_maquina = jugadas_maquina[i]
        resultado = definir_ganador(jugada_usuario, jugada_maquina)
        puntos_usuario = sumar_si_gana(puntos_usuario, resultado, GANA_USUARIO)
        puntos_maquina = sumar_si_gana(puntos_maquina, resultado, GANA_MAQUINA)
        print(texto_ronda(nombre_usuario, jugada_usuario,
                          nombre_maquina, jugada_maquina, resultado))
        print("Marcador:", texto_marcador(nombre_usuario, puntos_usuario,
                                          nombre_maquina, puntos_maquina))
        i = i + 1

    print()
    print("Ganador de la partida:", nombre_ganador(nombre_usuario, puntos_usuario,
                                                   nombre_maquina, puntos_maquina))


main()
````

### Una ronda, de punta a punta

Esto es lo que hace el programa principal en **cada vuelta** del ciclo. Los datos salen de sus variables, pasan por una función y vuelven a sus variables.

| # | El programa principal llama a | Le entrega | Recibe | Y lo guarda en |
|---|---|---|---|---|
| 1 | `partida_terminada` | Los dos puntajes y la meta | `False` | No se guarda: decide si el ciclo sigue |
| 2 | `definir_ganador` | Las dos jugadas | El resultado | `resultado` |
| 3 | `sumar_si_gana` | Los puntos del usuario y el resultado | Los puntos nuevos | `puntos_usuario` |
| 4 | `sumar_si_gana` | Los puntos de la máquina y el resultado | Los puntos nuevos | `puntos_maquina` |

| 6 | `texto_marcador` | Dos nombres y dos puntajes | Un texto | No se guarda: se imprime |

**Todo pasa por el programa principal.** Las funciones no se conocen entre sí (salvo `texto_ronda`, que usa `nombre_jugada`): es el programa principal el que lleva cada dato de una a otra.

### El programa de verdad: con teclado y con azar

El programa real cambia solo de dónde salen las jugadas: la del usuario se pide con `input`, repitiendo mientras no sea válida, y la de la máquina la da `jugar_maquina()`.

def leer_jugada():

### Resumen

| Qué pasa | Qué significa | Cómo se corrige |
|---|---|---|
| Un punto "se pierde" y no hay mensaje | Se llamó a la función sin guardar lo que retorna | `puntos = sumar_si_gana(puntos, ...)` |
| El marcador sale al revés y no hay mensaje | Los argumentos se pasaron en otro orden | Revisar el orden contra la definición de la función |
| Un resultado sin sentido y no hay mensaje | No se cumplió lo que la función supone | Validar antes de llamar |
| Se toma el camino equivocado y no hay mensaje | Se comparó con un número y no con la constante | Usar `EMPATE`, `GANA_USUARIO`, `GANA_MAQUINA` |
| `TypeError: ... missing 1 required positional argument` | Faltó un argumento | Contar los parámetros en la definición |

Cuatro de los cinco **no dan mensaje**. Las pruebas con `assert` existen para atraparlos: un `assert` compara el resultado con lo calculado a mano y se detiene si no coinciden.
