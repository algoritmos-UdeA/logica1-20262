## Parte 1 — El problema y un caso calculado a mano

### Enunciado

> Se desea un programa para jugar una **partida** de piedra, papel o tijera entre el usuario y la máquina:
>
> * Al empezar, el programa pide el nombre del usuario.
> * En cada **ronda**, el usuario elige su jugada (1 piedra, 2 papel, 3 tijera; si escribe otro número, se le vuelve a pedir) y la máquina elige la suya al azar.
> * Piedra le gana a tijera, tijera le gana a papel y papel le gana a piedra. Si las dos jugadas son iguales, es empate.

> * Después de cada ronda se muestran las jugadas, quién ganó y el marcador.

| 2 | piedra | tijera | Ana (piedra le gana a tijera) | Ana 1 - 1 Máquina |
| 3 | tijera | piedra | Máquina | Ana 1 - 2 Máquina |

## Parte 2 — El diseño: del enunciado a las clases y sus relaciones

El enunciado no dice qué clases escribir. Para descubrirlas se **subrayan los sustantivos y los verbos**:

* Los **sustantivos** (cosas) son candidatos a **clases** o a **atributos**.
* Los **verbos** (acciones) son candidatos a **métodos**.

### 1. Marcar el enunciado

En **negrita**, los sustantivos; en *cursiva*, los verbos.

> Se desea un programa para *jugar* una **partida** de piedra, papel o tijera entre el **usuario** y la **máquina**:
>
> * Al empezar, el programa *pide* el **nombre** del usuario.
> * En cada **ronda**, el usuario *elige* su **jugada** y la máquina *elige* la suya al azar.
> * Piedra *le gana* a tijera, tijera le gana a papel y papel le gana a piedra. Si las dos jugadas son iguales, es **empate**.
> * Quien gana la ronda *suma* un **punto**; en un empate, nadie suma.
> * Después de cada ronda *se muestran* las jugadas, quién ganó y el **marcador**.
> * *Gana* la partida quien primero gane 3 rondas.

### 2. Decidir qué es cada sustantivo

Para cada sustantivo se hacen tres preguntas, en orden:

1. **¿Tiene varios datos propios y algo que sabe hacer?** Entonces es una **clase**.
2. **¿Es un dato sencillo que describe a otra cosa?** Entonces es un **atributo** de esa cosa.
3. **¿Se puede calcular a partir de otros datos?** Entonces no se guarda: es un **método**.

| Sustantivo | Decisión | Por qué |
|---|---|---|
| **usuario**, **máquina** | **Clase `Jugador`** (dos objetos de la misma clase) | Cada uno tiene dos datos (nombre y puntos) y sabe sumar un punto. La máquina no necesita otra clase: es otro jugador |
| nombre | Atributo de `Jugador` (`str`) | Un dato sencillo que describe al jugador |
| punto | Atributo `puntos` de `Jugador` (`int`) | Un dato sencillo que describe al jugador |
| **ronda** | **Clase `Ronda`** | Tiene varios datos (quién jugó, qué jugó cada uno, el resultado) y sabe describirse |
| jugada | Atributos de `Ronda` (`int`): `jugada_usuario`, `jugada_maquina` | Un número entre 1 y 3: un dato sencillo |
| empate, quién ganó | Atributo `resultado` de `Ronda` (`int`) | Un número: `EMPATE`, `GANA_USUARIO` o `GANA_MAQUINA` |
| **partida** | **Clase `Partida`** | Tiene a los dos jugadores, la meta de rondas y la última ronda, y sabe jugar una ronda y decir si terminó |
| marcador | **Método** de `Partida` (`__str__`) | No se guarda: se arma con los nombres y los puntos de los jugadores |
| programa | Se descarta | Es el programa mismo |

Quedan **tres clases**: `Jugador`, `Ronda` y `Partida`.

La pista de que hay una **relación entre clases** está en los atributos cuyo tipo es otra clase: el *usuario* de una ronda es un `Jugador`; la *última ronda* de una partida es una `Ronda`.

### 3. Decidir de quién es cada verbo

Un método va en **la clase que tiene los datos necesarios para hacerlo**.

| Verbo del enunciado | ¿Dónde va? | Por qué ahí |
|---|---|---|
| sumar un punto | Método `sumar_punto()` de `Jugador` | El jugador tiene los puntos |
| jugar una ronda | Método `jugar_ronda(jugada_usuario, jugada_maquina)` de `Partida` | La partida conoce a los dos jugadores y sabe a quién sumarle |
| ganar la partida | Métodos `terminada()` y `get_ganador()` de `Partida` | La partida conoce los puntos de los dos y la meta |
| mostrar la ronda | Método `__str__()` de `Ronda` | La ronda tiene las jugadas, el resultado y los nombres |
| mostrar el marcador | Método `__str__()` de `Partida` | La partida conoce a los dos jugadores |
| **le gana a** | **Función** `definir_ganador(jugada_usuario, jugada_maquina)` | No necesita datos guardados: recibe dos jugadas y responde |
| elegir al azar | **Función** `jugar_maquina()` | No necesita datos guardados |
| pedir el nombre, elegir la jugada | El programa principal | Es la conversación con el usuario (`input`), que no va dentro de las clases |

> **¿Por qué `definir_ganador` no es un método?** Porque no tiene a quién pertenecer. No usa el nombre ni los puntos de nadie: recibe dos números y retorna otro. Una clase tiene sentido cuando hay **datos que viajan juntos y cosas que se hacen con ellos**. Una regla que solo transforma unos valores en otros se escribe bien como función. La parte 3 vuelve sobre esto.

### 4. Decidir las relaciones

Cada vez que una clase tiene un atributo cuyo tipo es otra clase, hay una relación. Para cada una se responden **tres preguntas**:

1. **¿Quién conoce a quién?** La clase que tiene el atributo conoce a la otra, y no al revés.
2. **¿Quién crea el objeto?** Esto decide el tipo de relación.
3. **¿Quién le pide qué a quién?** Los métodos que una clase usa de la otra.

| Relación | Símbolo | Cuándo |
|---|---|---|
| **Composición** | Rombo negro | La clase **crea** el objeto que guarda |
| **Agregación** | Rombo blanco | La clase **recibe** el objeto ya creado y lo guarda como parte suya |
| **Asociación** | Flecha | La clase **recibe** el objeto ya creado y lo conoce para pedirle cosas |

El rombo se dibuja del lado de la clase que guarda al objeto.

| Relación | ¿Quién conoce a quién? | ¿Quién crea el objeto? | Tipo | ¿Qué le pide? |
|---|---|---|---|---|
| `Ronda` — `Jugador` | La ronda conoce a sus dos jugadores | Nadie dentro de la ronda: los recibe | Asociación | `get_nombre()`, para describirse |
| `Partida` — `Jugador` | La partida conoce a sus dos jugadores | Los recibe ya creados | Agregación | `get_puntos()`, `sumar_punto()`, `get_nombre()` |
| `Partida` — `Ronda` | La partida conoce su última ronda | La partida, dentro de `jugar_ronda` | Composición | `get_resultado()` |

Hay que fijarse en la dirección: la partida conoce a sus jugadores, pero **un jugador no sabe en qué partida está**; la ronda conoce a los jugadores, pero **un jugador no conoce sus rondas**. Por eso `Jugador` es la clase más independiente.

## Parte 3 — Lo que no cambia: las reglas siguen siendo funciones

Las reglas del juego son **las mismas funciones** de la versión estructurada, sin tocar una letra. Las jugadas y los resultados son números con nombre.

````python
import random

# Jugadas
PIEDRA = 1
PAPEL = 2
TIJERA = 3

# Resultados de una ronda
EMPATE = 0
GANA_USUARIO = 1
GANA_MAQUINA = 2
````

````python
def es_jugada_valida(jugada):
    """
    Supone: jugada es un entero.
    Garantiza: retorna True si jugada es PIEDRA, PAPEL o TIJERA, y False
    si no.
    """
    return PIEDRA <= jugada <= TIJERA


def nombre_jugada(jugada):
    """
    Supone: jugada es PIEDRA, PAPEL o TIJERA.
    Garantiza: retorna "piedra", "papel" o "tijera".
    """
    if jugada == PIEDRA:
        nombre = "piedra"
    elif jugada == PAPEL:
        nombre = "papel"
    else:
        nombre = "tijera"
    return nombre
````

`definir_ganador` decide una ronda. Hay nueve casos:

| Usuario ↓ / Máquina → | piedra | papel | tijera |
|---|---|---|---|
| **piedra** | empate | gana la máquina | gana el usuario |
| **papel** | gana el usuario | empate | gana la máquina |
| **tijera** | gana la máquina | gana el usuario | empate |

````python
def definir_ganador(jugada_usuario, jugada_maquina):
    """
    Decide quién gana una ronda: piedra le gana a tijera, tijera le gana
    a papel y papel le gana a piedra.

    Supone: las dos jugadas son PIEDRA, PAPEL o TIJERA.
    Garantiza: retorna EMPATE, GANA_USUARIO o GANA_MAQUINA.
    """
    if jugada_usuario == jugada_maquina:
        resultado = EMPATE
    elif ((jugada_usuario == PIEDRA and jugada_maquina == TIJERA)
          or (jugada_usuario == TIJERA and jugada_maquina == PAPEL)
          or (jugada_usuario == PAPEL and jugada_maquina == PIEDRA)):
        resultado = GANA_USUARIO
    else:
        resultado = GANA_MAQUINA
    return resultado


def jugar_maquina():
    """Garantiza: retorna una jugada al azar: PIEDRA, PAPEL o TIJERA."""
    return random.randint(PIEDRA, TIJERA)
````

Las pruebas también son las mismas, e incluyen los nueve casos.

````python
# es_jugada_valida
assert es_jugada_valida(PIEDRA) == True         # límite inferior
assert es_jugada_valida(TIJERA) == True         # límite superior
assert es_jugada_valida(0) == False             # fuera del rango
assert es_jugada_valida(4) == False             # fuera del rango

# nombre_jugada
assert nombre_jugada(PIEDRA) == "piedra"
assert nombre_jugada(PAPEL) == "papel"
assert nombre_jugada(TIJERA) == "tijera"

# definir_ganador: los 9 casos posibles
assert definir_ganador(PIEDRA, PIEDRA) == EMPATE
assert definir_ganador(PAPEL, PAPEL) == EMPATE
assert definir_ganador(TIJERA, TIJERA) == EMPATE
assert definir_ganador(PIEDRA, TIJERA) == GANA_USUARIO
assert definir_ganador(TIJERA, PAPEL) == GANA_USUARIO
assert definir_ganador(PAPEL, PIEDRA) == GANA_USUARIO
assert definir_ganador(TIJERA, PIEDRA) == GANA_MAQUINA
assert definir_ganador(PAPEL, TIJERA) == GANA_MAQUINA
assert definir_ganador(PIEDRA, PAPEL) == GANA_MAQUINA

# jugar_maquina: el resultado es al azar, se prueba el rango
for i in range(1000):
    assert es_jugada_valida(jugar_maquina())

print("Pruebas de reglas: todas pasaron")
````

## Parte 4 — Paso 1: la clase `Jugador`

![Mapa de clases: paso 1, Jugador](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/mapa_poo_paso1.png)

`Jugador` no conoce a ninguna otra clase. Se escribe y se prueba sola. Resuelve la primera incomodidad: **el nombre y los puntos quedan juntos**.

### Enunciado

> Diseñe una clase llamada `Jugador`. La clase contiene:
>
> * Dos atributos privados: `nombre` (`str`) y `puntos` (`int`), que son las rondas que ha ganado.
> * Un constructor que recibe el nombre y crea al jugador con 0 puntos.
> * Métodos para consultar el nombre y los puntos.
> * Un método `sumar_punto`, que aumenta los puntos en 1.
> * Un método que devuelve al jugador en texto, por ejemplo `Ana: 2`.

### Diagrama UML

![UML de Jugador](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/uml_jugador.png)

### La clase

````python
class Jugador:
    def __init__(self, nombre):
        """Garantiza: crea un jugador con ese nombre y 0 puntos."""
        self.__nombre = nombre
        self.__puntos = 0

    def get_nombre(self):
        return self.__nombre

    def get_puntos(self):
        return self.__puntos

    def sumar_punto(self):
        """Garantiza: los puntos aumentan en 1. No retorna nada."""
        self.__puntos = self.__puntos + 1

    def __str__(self):
        """Garantiza: retorna 'nombre: puntos', por ejemplo 'Ana: 2'."""
        return self.__nombre + ": " + str(self.__puntos)
````

Hay una decisión de diseño que se nota por lo que **falta**: no existe un método `set_puntos`. Los puntos solo cambian con `sumar_punto()`, de uno en uno. Nadie puede, desde afuera, restarle puntos a un jugador ni ponerle 100. Eso es el **encapsulamiento** usado para proteger una regla del juego.

### Uso

````python
ana = Jugador("Ana")
maquina = Jugador("Máquina")
print(ana)

ana.sumar_punto()
ana.sumar_punto()
print(ana)
print(maquina)
print(ana.get_nombre(), "tiene", ana.get_puntos(), "puntos")
````

## Parte 5 — Paso 2: la clase `Ronda`, que conoce a dos jugadores

![Mapa de clases: paso 2, Ronda](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/mapa_poo_paso2.png)

Aquí aparece la primera relación: un objeto que **guarda referencias a otros objetos que ya existían**.

### Enunciado

> Diseñe una clase llamada `Ronda`. La clase contiene:
>
> * Cinco atributos privados: `usuario` y `maquina` (objetos `Jugador`), `jugada_usuario` y `jugada_maquina` (`int`), y `resultado` (`int`).
> * Un constructor que recibe los dos jugadores y sus dos jugadas, y calcula el resultado de la ronda.
> * Métodos para consultar las dos jugadas y el resultado.
> * Un método que devuelve la ronda en texto, por ejemplo `Ana: piedra | Máquina: tijera | gana Ana`.
>
> La clase **supone** que las jugadas son válidas. La ronda no cambia los puntos de los jugadores.

### Diagrama UML

![UML de Ronda](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/uml_ronda.png)

### Las tres preguntas

| Pregunta | Respuesta |
|---|---|
| ¿Quién conoce a quién? | La ronda conoce a sus dos jugadores. Un jugador no conoce sus rondas |
| ¿Quién crea el objeto? | **Nadie dentro de la ronda**: los jugadores llegan ya creados como parámetros del constructor. Es **asociación** |
| ¿Quién le pide qué a quién? | La ronda les pide a los jugadores su nombre (`get_nombre()`), para describirse. No les pide nada más |

### La clase

````python
class Ronda:
    def __init__(self, usuario, jugada_usuario, maquina, jugada_maquina):
        """
        Supone: usuario y maquina son objetos Jugador; las jugadas son
        PIEDRA, PAPEL o TIJERA.
        Garantiza: crea la ronda y calcula su resultado con definir_ganador.
        """
        self.__usuario = usuario
        self.__jugada_usuario = jugada_usuario
        self.__maquina = maquina
        self.__jugada_maquina = jugada_maquina
        self.__resultado = definir_ganador(jugada_usuario, jugada_maquina)

    def get_jugada_usuario(self):
        return self.__jugada_usuario

    def get_jugada_maquina(self):
        return self.__jugada_maquina

    def get_resultado(self):
        """Garantiza: retorna EMPATE, GANA_USUARIO o GANA_MAQUINA."""
        return self.__resultado

    def __str__(self):
        """
        Garantiza: retorna una línea que describe la ronda, por ejemplo
        "Ana: piedra | Máquina: tijera | gana Ana".
        """
        texto = (self.__usuario.get_nombre() + ": "
                 + nombre_jugada(self.__jugada_usuario) + " | "
                 + self.__maquina.get_nombre() + ": "
                 + nombre_jugada(self.__jugada_maquina) + " | ")
        if self.__resultado == EMPATE:
            texto = texto + "empate"
        elif self.__resultado == GANA_USUARIO:
            texto = texto + "gana " + self.__usuario.get_nombre()
        else:
            texto = texto + "gana " + self.__maquina.get_nombre()
        return texto
````

Tres cosas para leer con cuidado:

| Línea | Qué pasa |
|---|---|
| `self.__usuario = usuario` | **No se crea nada**: se guarda la referencia al jugador que llegó. La ronda y quien la creó señalan al mismo objeto |
| `self.__resultado = definir_ganador(jugada_usuario, jugada_maquina)` | La clase **usa una función** de las reglas. Clases y funciones conviven sin problema |
| `self.__usuario.get_nombre()` | Para describirse, la ronda le **pide** el nombre al jugador. No puede leer `__nombre` directamente: es privado en `Jugador` |

Y una que no está: la ronda **no llama a `sumar_punto()`**. Sabe quién ganó, pero sumar el punto no es su responsabilidad. De eso se encargará la partida.

### Uso

Primero se crean los jugadores. Después se crea la ronda y se le **entregan**.

````python
ana = Jugador("Ana")
maquina = Jugador("Máquina")

ronda = Ronda(ana, PIEDRA, maquina, TIJERA)
print(ronda)
print(ronda.get_resultado() == GANA_USUARIO)
````

### Pruebas

La prueba de `Ronda` ya es una pequeña **integración**: usa también a `Jugador` y a las reglas, que ya están probados. Los nueve casos de `definir_ganador` no se repiten aquí: basta **uno por cada resultado**.

| Caso | Resultado esperado |
|---|---|
| piedra contra tijera | `GANA_USUARIO`; texto `... gana Ana` |
| piedra contra papel | `GANA_MAQUINA`; texto `... gana Máquina` |
| tijera contra tijera | `EMPATE`; texto `... empate` |
| Después de crear las tres rondas | Los dos jugadores siguen con 0 puntos |

## Parte 6 — Paso 3: la clase `Partida`, que recibe jugadores y crea rondas

![Mapa de clases: paso 3, Partida](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/mapa_poo_paso3.png)

`Partida` es la clase que lleva el juego. En ella aparecen **dos tipos de relación a la vez**.

### Enunciado

> Diseñe una clase llamada `Partida`. La clase contiene:
>
> * Cinco atributos privados: `usuario` y `maquina` (objetos `Jugador`), `rondas_para_ganar` (`int`), `ultima_ronda` (un objeto `Ronda`) y `cantidad_rondas` (`int`).
> * Un constructor que recibe los dos jugadores y cuántas rondas hay que ganar (3 si no se dice), y crea una partida sin rondas.
> * Métodos para consultar la última ronda y la cantidad de rondas jugadas.
> * Un método `jugar_ronda`, que recibe las dos jugadas, crea la ronda, le suma un punto a quien la gana (a nadie si es empate) y retorna el resultado.
> * Un método `terminada`, que dice si alguno ya ganó las rondas necesarias.
> * Un método `get_ganador`, que retorna el jugador con más puntos.
> * Un método que devuelve el marcador en texto, por ejemplo `Ana 2 - 1 Máquina`.
>
> La clase **supone** que los jugadores llegan con 0 puntos, que las jugadas son válidas y que `get_ganador` se pide cuando la partida terminó.

### Diagrama UML

![UML de Partida](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/uml_partida.png)

`= 3` en el constructor indica el valor por defecto de ese parámetro.

### Las tres preguntas

| Pregunta | Con `Jugador` | Con `Ronda` |
|---|---|---|
| ¿Quién conoce a quién? | La partida conoce a sus dos jugadores | La partida conoce su última ronda |
| ¿Quién crea el objeto? | **Otro** (el programa principal): llegan ya creados al constructor. Es **agregación** | **La partida**, dentro de `jugar_ronda`. Es **composición** |
| ¿Quién le pide qué a quién? | Sus puntos, su nombre y que sumen un punto | Su resultado |

### La clase

````python
class Partida:
    def __init__(self, usuario, maquina, rondas_para_ganar=3):
        """
        Supone: usuario y maquina son objetos Jugador con 0 puntos;
        rondas_para_ganar > 0.
        Garantiza: crea una partida que todavía no tiene rondas.
        """
        self.__usuario = usuario
        self.__maquina = maquina
        self.__rondas_para_ganar = rondas_para_ganar
        self.__ultima_ronda = None
        self.__cantidad_rondas = 0

    def get_ultima_ronda(self):
        """Garantiza: retorna la última ronda jugada, o None si no hay."""
        return self.__ultima_ronda

    def get_cantidad_rondas(self):
        return self.__cantidad_rondas

    def jugar_ronda(self, jugada_usuario, jugada_maquina):
        """
        Supone: la partida no ha terminado; las jugadas son válidas.
        Garantiza: crea la ronda, le suma un punto a quien la gana (a
        nadie si es empate) y retorna su resultado.
        """
        ronda = Ronda(self.__usuario, jugada_usuario,
                      self.__maquina, jugada_maquina)
        resultado = ronda.get_resultado()
        if resultado == GANA_USUARIO:
            self.__usuario.sumar_punto()
        elif resultado == GANA_MAQUINA:
            self.__maquina.sumar_punto()
        self.__ultima_ronda = ronda
        self.__cantidad_rondas = self.__cantidad_rondas + 1
        return resultado

    def terminada(self):
        """
        Garantiza: retorna True si alguno de los dos ya ganó
        rondas_para_ganar rondas, y False si no.
        """
        return (self.__usuario.get_puntos() == self.__rondas_para_ganar
                or self.__maquina.get_puntos() == self.__rondas_para_ganar)

    def get_ganador(self):
        """
        Supone: la partida terminó.
        Garantiza: retorna el jugador que tiene más puntos.
        """
        ganador = self.__maquina
        if self.__usuario.get_puntos() > self.__maquina.get_puntos():
            ganador = self.__usuario
        return ganador

    def __str__(self):
        """Garantiza: retorna el marcador, por ejemplo "Ana 2 - 1 Máquina"."""
        return (self.__usuario.get_nombre() + " "
                + str(self.__usuario.get_puntos()) + " - "
                + str(self.__maquina.get_puntos()) + " "
                + self.__maquina.get_nombre())
````

El método `jugar_ronda` reúne todo lo construido hasta ahora:

| Línea | Qué pasa | A quién se le pide |
|---|---|---|
| `ronda = Ronda(self.__usuario, jugada_usuario, self.__maquina, jugada_maquina)` | La partida **crea** una ronda y le entrega sus dos jugadores | A la clase `Ronda` |
| `resultado = ronda.get_resultado()` | Le pregunta a la ronda quién ganó | A la ronda |
| `self.__usuario.sumar_punto()` | Le pide al ganador que sume su punto | Al jugador |
| `self.__ultima_ronda = ronda` | Guarda la ronda para poder mostrarla después | — |

Quien usa la partida escribe una línea, `partida.jugar_ronda(PIEDRA, TIJERA)`, y **recibe solo dos parámetros**: las jugadas. Los nombres y los puntos ya los tiene la partida, a través de sus jugadores.

`self.__ultima_ronda` empieza en `None`, que es el valor de Python para decir "todavía no hay objeto".

La última línea es una **llamada encadenada**. Se lee de izquierda a derecha, y cada punto es una petición a un objeto distinto:

| Parte | ¿A quién se le pide? | ¿Qué retorna? |
|---|---|---|
| `partida.get_ganador()` | A la partida | Un objeto `Jugador` |
| `partida.get_ganador().get_nombre()` | A ese jugador | Un `str`: `"Ana"` |

Para saber qué método se puede escribir después de un punto, hay que preguntar: **¿de qué clase es lo que retornó la parte anterior?**

### Pruebas

La prueba de `Partida` ejercita **las tres clases y las reglas a la vez**.

| Caso | Resultado esperado |
|---|---|
| Partida nueva | No terminada; 0 rondas; última ronda `None`; marcador `Ana 0 - 0 Máquina` |
| Gana el usuario | Suma él y solo él |
| Empate | No suma nadie, pero la ronda cuenta |
| Gana la máquina | Suma ella |
| Con el marcador 2 - 1 (límite) | Todavía no termina |
| Con el marcador 3 - 1 | Termina; el ganador es el mismo objeto `ana` |
| Una partida a 1 ronda que gana la máquina | Termina en la primera ronda |

## Parte 7 — La integración: el programa principal

![Mapa de clases: todas probadas, y el programa principal](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/mapa_poo_integracion.png)

Las tres clases están escritas y probadas. Falta el programa que las usa.

Hay dos papeles distintos. Quien **escribió** las clases conoce sus atributos y sus detalles. El **programa principal** es su cliente: solo las usa, a través de sus métodos públicos. Hace tres cosas:

1. **Crea** los dos jugadores y la partida.
2. **Pide**: le encarga cada ronda a la partida.
3. **Muestra** lo que la partida responde.

| El programa principal sí | El programa principal no |
|---|---|
| Crea los objetos `Jugador` y la `Partida` | Crea objetos `Ronda`: los crea la partida |
| Le pide a la partida `jugar_ronda(...)` | Decide quién ganó ni le suma puntos a nadie |
| Imprime la ronda y el marcador | Arma esos textos: los arman `Ronda` y `Partida` |

Como en el otro notebook, las jugadas están escritas en dos arreglos para que la salida sea siempre la misma.

````python
RONDAS_PARA_GANAR = 3


def main():
    # Las jugadas del caso calculado a mano
    jugadas_usuario = [PIEDRA, PIEDRA, TIJERA, PAPEL, PAPEL, TIJERA]
    jugadas_maquina = [PAPEL, TIJERA, PIEDRA, PAPEL, PIEDRA, PAPEL]

    # 1. Crear
    usuario = Jugador("Ana")
    maquina = Jugador("Máquina")
    partida = Partida(usuario, maquina, RONDAS_PARA_GANAR)

    i = 0
    while not partida.terminada():
        # 2. Pedir
        partida.jugar_ronda(jugadas_usuario[i], jugadas_maquina[i])
        # 3. Mostrar
        print(partida.get_ultima_ronda())
        print("Marcador:", partida)
        i = i + 1

    print()
    print("Ganador de la partida:", partida.get_ganador().get_nombre())


main()
````

### Una ronda, de punta a punta

El programa principal escribió **una línea**: `partida.jugar_ronda(PIEDRA, PAPEL)`. Detrás de ella colaboraron las tres clases y una función:

![Quién le pide qué a quién al jugar una ronda](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/peticion_jugar_ronda.png)

| # | Quién pide | A quién | Qué le pide | Para qué |
|---|---|---|---|---|
| 1 | Programa principal | `Partida` | `jugar_ronda(PIEDRA, PAPEL)` | Es lo único que el programa sabe pedir |
| 2 | `Partida` | la clase `Ronda` | `Ronda(usuario, PIEDRA, maquina, PAPEL)` | Crear la ronda |
| 3 | `Ronda` | la función `definir_ganador` | `definir_ganador(PIEDRA, PAPEL)` | Saber quién ganó |
| 4 | `Partida` | `Ronda` | `get_resultado()` | Saber a quién sumarle |
| 5 | `Partida` | `Jugador` (la máquina) | `sumar_punto()` | Que el ganador sume su punto |

Después, al imprimir, hay dos peticiones más: `print(partida.get_ultima_ronda())` hace que la ronda les pida el nombre a sus jugadores, y `print(partida)` hace que la partida les pida nombre y puntos.

Ninguna clase hace todo el trabajo. Cada una hace una parte pequeña y le pide el resto a quien sabe hacerlo. **Eso es un programa orientado a objetos: objetos que se piden cosas entre sí.**

En la versión con funciones, todos los datos pasaban por el programa principal, que los llevaba de una función a otra. Aquí **los objetos se hablan entre ellos**, y el programa principal solo da la orden inicial.

### El programa de verdad: con teclado y con azar

El programa real cambia solo de dónde salen las jugadas, igual que en la versión con funciones:

```python
partida.jugar_ronda(leer_jugada(), jugar_maquina())
```

Las clases no cambian. El programa completo, separado en archivos (`reglas.py`, `jugador.py`, `ronda.py`, `partida.py`, `main.py` y sus pruebas), está en [`version_poo/codigo`](https://github.com/algoritmos-UdeA/logica1-20262/tree/main/clases/13/teoria/app_piedra-papel-tijera/version_poo/codigo). La página [Versión POO](https://github.com/algoritmos-UdeA/logica1-20262/blob/main/clases/13/teoria/app_piedra-papel-tijera/version_poo/README.md) explica cómo ejecutarlo.

Al separar en archivos aparece un detalle que confirma el diseño: **un archivo importa a otro solo si crea objetos de esa clase**. `partida.py` importa `ronda.py` (composición: crea rondas), pero no importa `jugador.py` (agregación: recibe a los jugadores ya creados).

## Parte 8 — Las dos versiones, frente a frente

### La misma ronda, escrita de dos formas

```python
# Con funciones
resultado = definir_ganador(jugada_usuario, jugada_maquina)
puntos_usuario = sumar_si_gana(puntos_usuario, resultado, GANA_USUARIO)
puntos_maquina = sumar_si_gana(puntos_maquina, resultado, GANA_MAQUINA)
print(texto_ronda(nombre_usuario, jugada_usuario,
                  nombre_maquina, jugada_maquina, resultado))
print("Marcador:", texto_marcador(nombre_usuario, puntos_usuario,
                                  nombre_maquina, puntos_maquina))

# Con objetos
partida.jugar_ronda(jugada_usuario, jugada_maquina)
print(partida.get_ultima_ronda())
print("Marcador:", partida)
```

| | Con funciones | Con objetos |
|---|---|---|
| ¿Dónde están los datos de un jugador? | En dos variables sueltas | En un objeto `Jugador` |
| ¿Quién suma el punto? | El programa principal, con `puntos = sumar_si_gana(...)`, una vez por jugador | `Partida.jugar_ronda`, que le pide `sumar_punto()` al que ganó |
| ¿Cuántos datos recibe cada tarea? | `texto_ronda` recibe 5; `texto_marcador` y `nombre_ganador`, 4 | `jugar_ronda` recibe 2: las jugadas |
| ¿Cómo se imprime? | Con funciones que arman el texto | `print(ronda)` y `print(partida)`, gracias a `__str__` |
| ¿Quién decide si terminó? | `partida_terminada(puntos_usuario, puntos_maquina, RONDAS_PARA_GANAR)` | `partida.terminada()` |

### El tablero de incomodidades

![Tablero: versión estructurada frente a versión POO](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_piedra-papel-tijera/images/tablero_juego.png)

Tres incomodidades quedaron resueltas. La cuarta fila es distinta: las reglas **no necesitaban** resolverse, y por eso siguen siendo funciones.

La última fila es la prueba de fuego de un diseño: **¿qué pasa cuando hay que agregar algo?** Si cada jugador llevara sus partidas ganadas, con funciones cambiarían variables, parámetros y llamadas en varias partes. Con objetos cambia `Jugador` (un atributo y un método) y una línea del programa principal; `Ronda` y `Partida` **no cambian**. El ejercicio 1 de la parte 10 lo comprueba.

> **La programación orientada a objetos no reemplaza a las funciones.** Este programa tiene tres clases y cuatro funciones, y las dos cosas están bien donde están.

`'str' object has no attribute 'get_nombre'`. Lo engañoso es **dónde** falla: la ronda se creó sin problema, porque el constructor solo guarda lo que recibe. El error aparece más tarde, cuando `__str__` intenta `self.__usuario.get_nombre()` y `self.__usuario` resulta ser el texto `"Ana"`.

`'NoneType' object has no attribute 'get_resultado'` significa: "esto es `None`, y a `None` no se le puede pedir nada". `get_ultima_ronda()` retornó `None` porque no se ha jugado ninguna ronda. Se revisa antes de usar:

```python
ronda = partida.get_ultima_ronda()
if ronda != None:
    print(ronda.get_resultado())
```

### 4. Cambiar un atributo privado desde afuera

> ⚠️ **Error frecuente.** Intentar "arreglar" los puntos de un jugador sin pasar por sus métodos.

````python
ana = Jugador("Ana")

ana.__puntos = 10        # no da error...
print(ana.get_puntos())  # ...pero los puntos verdaderos no cambiaron
print(ana)
````

La segunda partida **nace terminada**. Los puntos están en los jugadores, y son los mismos dos objetos: Ana llega con el punto que ganó antes. La partida *supone* jugadores con 0 puntos, y aquí no se cumplió.

Es el efecto, esta vez no deseado, de que la partida guarde referencias y no copias. El ejercicio 4 de la parte 10 propone una solución, y deja una pregunta de diseño: ¿los puntos de una partida deberían estar en el jugador o en la partida?

### Resumen

| Qué pasa | Qué significa | Cómo se corrige |
|---|---|---|
| `AttributeError: 'str' object has no attribute ...` | Se entregó un dato (texto, número) donde se esperaba un objeto. Puede fallar lejos de donde se entregó | Entregar el objeto, no uno de sus datos |
| `AttributeError: 'NoneType' object has no attribute ...` | Se le pidió un método a `None`: todavía no hay objeto | Revisar `if objeto != None:` antes de usarlo |
| Un método "no hace nada" y no hay mensaje | Faltaron los paréntesis | `objeto.metodo()` |
| Un atributo "no cambia" y no hay mensaje | Se intentó cambiar un atributo privado desde afuera | Usar los métodos de la clase |
| Un objeto llega con datos de antes y no hay mensaje | Varias variables o varios objetos señalan al mismo objeto | Decidir si se quería el mismo objeto o uno nuevo |

Tres de los cinco **no dan mensaje**. Para encontrarlos sirve dibujar los objetos y sus flechas, o verlos en Python Tutor.

### ✏️ Ejercicio 1 — Partidas ganadas

Se quiere que cada jugador lleve cuántas partidas ha ganado. Agregue el atributo `partidas_ganadas` (empieza en 0), un método para consultarlo y un método `sumar_partida()`. ¿En cuál clase van? ¿Cambia alguna de las otras dos?

<details>
<summary>Ver solución</summary>

Van en `Jugador`, porque es un dato del jugador. `Ronda` y `Partida` **no cambian**.

```python
class Jugador:
    def __init__(self, nombre):
        self.__nombre = nombre
        self.__puntos = 0
        self.__partidas_ganadas = 0

    def get_nombre(self):
        return self.__nombre

    def get_puntos(self):
        return self.__puntos

    def get_partidas_ganadas(self):
        return self.__partidas_ganadas

    def sumar_punto(self):
        self.__puntos = self.__puntos + 1

    def sumar_partida(self):
        self.__partidas_ganadas = self.__partidas_ganadas + 1

    def __str__(self):
        return self.__nombre + ": " + str(self.__puntos)
```

El programa principal agrega una línea al terminar la partida, y la prueba la usa:

```python
ana = Jugador("Ana")
maquina = Jugador("Máquina")
partida = Partida(ana, maquina, 1)
partida.jugar_ronda(PIEDRA, TIJERA)
partida.get_ganador().sumar_partida()
assert ana.get_partidas_ganadas() == 1
assert maquina.get_partidas_ganadas() == 0
```

`partida.get_ganador().sumar_partida()` es una llamada encadenada: la partida entrega al jugador que ganó, y a ese jugador se le pide que sume una partida.

</details>

### ✏️ Ejercicio 2 — `get_perdedor`

Agregue a `Partida` un método `get_perdedor()` que retorne el jugador con menos puntos. Supone, como `get_ganador`, que la partida terminó.

<details>
<summary>Ver solución</summary>

Se agrega dentro de la clase `Partida`. Retorna **el objeto**, no su nombre:

```python
    def get_perdedor(self):
        perdedor = self.__usuario
        if self.__usuario.get_puntos() > self.__maquina.get_puntos():
            perdedor = self.__maquina
        return perdedor
```

```python
ana = Jugador("Ana")
maquina = Jugador("Máquina")
partida = Partida(ana, maquina, 1)
partida.jugar_ronda(PIEDRA, TIJERA)
assert partida.get_perdedor() is maquina
assert partida.get_ganador() is not partida.get_perdedor()
```

</details>

### ✏️ Ejercicio 3 — Prediga la salida

Sin ejecutar, escriba qué imprimen las tres últimas líneas. Después compruebe.

```python
ana = Jugador("Ana")
maquina = Jugador("Máquina")
partida = Partida(ana, maquina, 2)

partida.jugar_ronda(PAPEL, PIEDRA)
ana.sumar_punto()

print(partida)
print(partida.terminada())
print(partida.get_cantidad_rondas())
```

<details>
<summary>Ver solución</summary>

```
Ana 2 - 0 Máquina
True
1
```

* Ana ganó la ronda (papel le gana a piedra), y la partida le sumó un punto.
* `ana.sumar_punto()` le suma otro **por fuera de la partida**. Como la partida guarda una referencia al mismo objeto, lo "ve": el marcador dice 2.
* Con 2 puntos y la meta en 2, la partida terminó, aunque solo se jugó **una** ronda.

El programa principal no debería hacer esto: los puntos de una partida los suma la partida. El ejemplo muestra que el encapsulamiento protege los atributos, pero no impide usar mal los métodos públicos.

</details>

### ✏️ Ejercicio 4 — Una segunda partida

En la parte 9 una segunda partida nacía terminada, porque los jugadores conservaban sus puntos. Agregue a `Jugador` un método `reiniciar()` que deje los puntos en 0, y úselo para jugar dos partidas seguidas con los mismos jugadores.

<details>
<summary>Ver solución</summary>

Se agrega dentro de la clase `Jugador`:

```python
    def reiniciar(self):
        self.__puntos = 0
```

```python
ana = Jugador("Ana")
maquina = Jugador("Máquina")
primera = Partida(ana, maquina, 1)
primera.jugar_ronda(PIEDRA, TIJERA)

ana.reiniciar()
maquina.reiniciar()
segunda = Partida(ana, maquina, 1)
assert segunda.terminada() == False
assert str(segunda) == "Ana 0 - 0 Máquina"
```

**Pregunta para pensar.** Esta solución funciona, pero obliga al programa principal a acordarse de reiniciar. ¿Y si los puntos de una partida no estuvieran en el `Jugador`, sino en la `Partida`? Dibuje el diagrama UML con ese cambio: ¿qué atributos pasarían de una clase a la otra?, ¿qué métodos cambiarían? No hay una única respuesta correcta; lo importante es poder explicar la que se escoja.

</details>
