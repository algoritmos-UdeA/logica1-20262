Dos ideas se repiten en todo el notebook:

* **Una función puede modificar un arreglo que recibe**, y el cambio lo ve quien la llamó. Cuando lo hace, su especificación dice **MODIFICA**.
* **Una función no puede modificar un número que recibe.** Si calcula un valor nuevo (por ejemplo, cuántos inscritos hay ahora), lo tiene que **retornar**, y quien la llama tiene que guardarlo.

## Parte 1 — El problema y un caso calculado a mano

### Enunciado

> Se desea un programa para administrar las notas de un curso con cupo para `N = 10` estudiantes. De cada estudiante se conoce su **Id** (un texto, por ejemplo `"007"`), su **nombre** y su **nota** (entre 0.0 y 5.0).
>
> El programa debe permitir:
>
> 1. **Agregar** un estudiante. No se puede inscribir un Id repetido ni pasar del cupo.
> 2. **Eliminar** un estudiante, dado su Id.
> 3. **Actualizar** la nota de un estudiante, dado su Id.
> 4. **Buscar** un estudiante por su Id.
> 5. **Listar** el curso.
> 6. **Ordenar** el curso por nota, de mayor a menor.
> 7. **Ordenar** el curso por nombre, en orden alfabético.
>
> El programa empieza con cuatro estudiantes inscritos: Ana (`"000"`, 3.0), Beto (`"123"`, 4.5), Carla (`"987"`, 2.0) y Dario (`"007"`, 4.0).

### Un caso para verificar a mano

Antes de programar se calcula a mano qué debe pasar con una secuencia de operaciones. Esa secuencia se usará al final para probar el programa completo.

| # | Operación | Cómo queda el curso | Inscritos |
|---|---|---|---|
| | (al comenzar) | Ana 3.0, Beto 4.5, Carla 2.0, Dario 4.0 | 4 |
| 1 | Agregar a Elena (`"555"`, 3.8) | Ana, Beto, Carla, Dario, **Elena 3.8** | 5 |
| 2 | Agregar a otro estudiante con el Id `"123"` | No cambia: ese Id ya está | 5 |
| 3 | Eliminar el Id `"123"` (Beto) | Ana, Carla, Dario, Elena | 4 |
| 4 | Actualizar la nota de `"987"` (Carla) a 3.5 | Ana 3.0, **Carla 3.5**, Dario 4.0, Elena 3.8 | 4 |
| 5 | Buscar el Id `"007"` | No cambia. Dario está en la posición 2 | 4 |
| 6 | Ordenar por nota | Dario 4.0, Elena 3.8, Carla 3.5, Ana 3.0 | 4 |
| 7 | Ordenar por nombre | Ana 3.0, Carla 3.5, Dario 4.0, Elena 3.8 | 4 |

Las posiciones se cuentan desde 0. Al eliminar a Beto, los que estaban después se corren una posición a la izquierda: por eso Dario pasa de la posición 3 a la 2.

## Parte 2 — El diseño: los datos, las tareas y las funciones

### 1. Los datos

De cada estudiante hay tres datos de **tipos distintos**: el Id y el nombre son textos, y la nota es un número. Un arreglo guarda datos de un solo tipo, así que se usan **tres arreglos paralelos**: el estudiante de la posición `i` tiene su Id en `ids[i]`, su nombre en `nombres[i]` y su nota en `notas[i]`.

![Tres arreglos paralelos con cupo 10 y cantidad 4](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/arreglo_parcial.png)

Los arreglos tienen `N = 10` casillas, pero no todas están ocupadas. La variable `cantidad` dice **cuántas lo están**. Los estudiantes ocupan las posiciones `0` a `cantidad - 1`; el resto está libre. Esto se llama un arreglo **parcialmente lleno**.

| Dato | Cómo se guarda | Tipo |
|---|---|---|
| El cupo | Constante `N` | `int` |
| Los Id | Arreglo `ids`, de `N` casillas | Textos. El Id `"007"` se guarda como texto para no perder los ceros |
| Los nombres | Arreglo `nombres`, de `N` casillas | Textos |
| Las notas | Arreglo `notas`, de `N` casillas | Números con decimales |
| Cuántos inscritos hay | Variable `cantidad` | `int` |

Hay que fijarse en algo que más adelante va a pesar: **un estudiante no existe como un solo dato**. Sus tres datos están en tres arreglos distintos, y lo único que los une es la posición.

### 2. Las tareas

Cada opción del enunciado se vuelve una función. Para cada una se decide qué recibe, si modifica los arreglos y qué retorna.

| Tarea | Función | ¿Modifica los arreglos? | Retorna |
|---|---|---|---|
| Buscar por Id | `buscar_id` | No | La posición, o -1 si no está |
| Agregar | `agregar_estudiante` | Sí | La cantidad nueva |
| Eliminar | `eliminar_estudiante` | Sí | La cantidad nueva |
| Actualizar la nota | `actualizar_nota` | Sí (`notas`) | `True` si lo encontró, `False` si no |
| Listar | `mostrar_listado` | No | Nada: imprime |
| Ordenar por nota | `ordenar_por_nota` | Sí | Nada |
| Ordenar por nombre | `ordenar_por_nombre` | Sí | Nada |

Al escribirlas aparecen cuatro **funciones de apoyo**, que el enunciado no menciona pero que evitan repetir código: `intercambiar`, `intercambiar_estudiantes`, `posicion_mayor_nota` y `posicion_menor_nombre`.

### 3. Tres decisiones de diseño

1. **Todas las funciones reciben `cantidad`** y recorren hasta ahí, no hasta el final del arreglo. Las casillas libres no son estudiantes.
2. **Las funciones que cambian la cantidad la retornan.** `agregar_estudiante` y `eliminar_estudiante` no pueden cambiar la variable `cantidad` de quien las llama, así que entregan el valor nuevo.
3. **Solo `mostrar_listado` imprime.** Las demás retornan resultados, y por eso se pueden probar con `assert`.

El programa principal llama a siete funciones, una por opción. Tres de ellas se apoyan en `buscar_id`, y las dos de ordenar comparten `intercambiar_estudiantes`.

### 5. El plan de construcción

| Paso | Qué se construye | Necesita que ya exista |
|---|---|---|
| 1 | Crear el curso, `buscar_id` y `agregar_estudiante` | Nada |

## Parte 3 — Paso 1: crear el curso, buscar y agregar

### Crear el curso

Un curso vacío son tres arreglos con sus casillas libres y una cantidad en 0. Las casillas libres se marcan con `""` en los textos y `0.0` en las notas.

````python
N = 10      # cupo máximo del curso

ids = [""] * N
nombres = [""] * N
notas = [0.0] * N
cantidad = 0

print(ids)
print(notas)
print(cantidad)
````

### `buscar_id`

Es una búsqueda lineal: recorre los inscritos hasta encontrar el Id, y retorna **la posición**. Con la posición se llega a los otros dos datos del estudiante.

````python
def buscar_id(ids, cantidad, id_buscado):
    """
    Busca un Id entre los estudiantes inscritos (búsqueda lineal).

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        Retorna la posición de id_buscado entre 0 y cantidad - 1,
        o -1 si no está. No modifica ids.
    """
    pos = -1
    for i in range(cantidad):
        if ids[i] == id_buscado:
            pos = i
            break
    return pos
````

Dos detalles:

* El recorrido es `range(cantidad)`, no `range(len(ids))`: solo se busca entre los inscritos.
* La función retorna `-1` cuando no encuentra. Es un valor que no puede ser una posición, así que quien la llama lo reconoce.

### `agregar_estudiante`

````python
def agregar_estudiante(ids, nombres, notas, cantidad, id_nuevo, nombre, nota):
    """
    Agrega un estudiante al final de los inscritos.

    Supone:
        ids, nombres y notas tienen el mismo tamaño (el cupo).
        0 <= cantidad <= len(ids).
    Garantiza:
        Si hay cupo y id_nuevo no está inscrito, MODIFICA ids, nombres y
        notas (guarda los datos en la posición cantidad) y retorna
        cantidad + 1. Si no, no modifica nada y retorna cantidad.
    """
    if cantidad < len(ids) and buscar_id(ids, cantidad, id_nuevo) == -1:
        ids[cantidad] = id_nuevo
        nombres[cantidad] = nombre
        notas[cantidad] = nota
        cantidad = cantidad + 1
    return cantidad
````

La función revisa dos condiciones antes de guardar: que **haya cupo** (`cantidad < len(ids)`) y que el **Id no esté inscrito** (para eso usa `buscar_id`). Si se cumplen, escribe en la posición `cantidad`, que es la primera libre, **en los tres arreglos**.

Recibe siete parámetros: los tres arreglos, la cantidad y los tres datos del estudiante.

Lo más importante es la última línea: `return cantidad`. La función **modificó los arreglos**, y eso lo ve quien la llamó. Pero no puede modificar la variable `cantidad` de quien la llamó: le entrega el valor nuevo, y hay que guardarlo.

````python
cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "000", "Ana", 3.0)
cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "123", "Beto", 4.5)
cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "987", "Carla", 2.0)
cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "007", "Dario", 4.0)

print(cantidad)
print(ids)
print(nombres)
print(notas)
````

### Las pruebas y dos funciones de ayuda

Casi todas las pruebas necesitan el mismo punto de partida: el curso con los cuatro estudiantes. Para no repetirlo, se escribe una función `curso_inicial()` que lo arma y retorna las cuatro cosas. La otra, `empieza_con`, compara las primeras casillas de un arreglo con los valores esperados.

````python
N = 10


def empieza_con(lista, esperado):
    """
    Supone: len(esperado) <= len(lista).
    Garantiza: retorna True si las primeras posiciones de lista son
    iguales, una por una, a las de esperado. No modifica nada.
    """
    iguales = True
    for k in range(len(esperado)):
        if lista[k] != esperado[k]:
            iguales = False
    return iguales


def curso_inicial():
    """Garantiza: retorna (ids, nombres, notas, cantidad) con los 4 estudiantes de ejemplo."""
    ids = [""] * N
    nombres = [""] * N
    notas = [0.0] * N
    cantidad = 0
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "000", "Ana", 3.0)
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "123", "Beto", 4.5)
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "987", "Carla", 2.0)
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "007", "Dario", 4.0)
    return ids, nombres, notas, cantidad
````

`return ids, nombres, notas, cantidad` retorna cuatro valores a la vez, y quien llama los recibe en cuatro variables: `ids, nombres, notas, cantidad = curso_inicial()`.

| Función | Caso | Resultado esperado |
|---|---|---|
| `buscar_id` | Típico: uno del medio | Su posición |
| | Límites: el primero y el último | 0 y 3 |
| | No está | -1 |
| | Se busca `""`, el valor de las casillas libres | -1: las libres no cuentan |
| `agregar_estudiante` | Típico | Queda en la posición 4; la cantidad pasa a 5 |
| | Id repetido | No agrega |
| | Curso lleno | No agrega |

````python
# buscar_id
ids, nombres, notas, cantidad = curso_inicial()
assert cantidad == 4
assert buscar_id(ids, cantidad, "987") == 2         # típico
assert buscar_id(ids, cantidad, "000") == 0         # límite: el primero
assert buscar_id(ids, cantidad, "007") == 3         # límite: el último
assert buscar_id(ids, cantidad, "555") == -1        # no está
assert buscar_id(ids, cantidad, "") == -1           # las casillas libres no cuentan
print("buscar_id: pruebas correctas")
````

### `eliminar_estudiante`

En un arreglo no se puede "sacar" una casilla. Para eliminar a un estudiante se **corren una posición a la izquierda** todos los que estaban después, y la última casilla ocupada queda libre.

````python
def eliminar_estudiante(ids, nombres, notas, cantidad, id_buscado):
    """
    Elimina un estudiante por Id, corriendo los siguientes a la izquierda.

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        Si id_buscado está inscrito, MODIFICA ids, nombres y notas (los
        estudiantes que estaban después quedan una posición a la
        izquierda y la última ocupada se libera) y retorna cantidad - 1.
        Si no está, no modifica nada y retorna cantidad.
    """
    pos = buscar_id(ids, cantidad, id_buscado)
    if pos != -1:
        for i in range(pos, cantidad - 1):
            ids[i] = ids[i + 1]
            nombres[i] = nombres[i + 1]
            notas[i] = notas[i + 1]
        cantidad = cantidad - 1
        ids[cantidad] = ""
        nombres[cantidad] = ""
        notas[cantidad] = 0.0
    return cantidad
````

Así se elimina el Id `"123"` (Beto, posición 1) de un curso con 4 inscritos. El ciclo va de `pos` a `cantidad - 2`:

| Momento | `ids` (primeras 4 casillas) | Qué pasó |
|---|---|---|
| Antes | `"000"`, `"123"`, `"987"`, `"007"` | `buscar_id` retornó `pos = 1` |
| `i = 1` | `"000"`, **`"987"`**, `"987"`, `"007"` | La casilla 1 copia a la 2 |
| `i = 2` | `"000"`, `"987"`, **`"007"`**, `"007"` | La casilla 2 copia a la 3 |
| Después del ciclo | `"000"`, `"987"`, `"007"`, **`""`** | `cantidad` pasa a 3 y la casilla 3 se libera |

Lo mismo se hace, en el mismo ciclo, con `nombres` y con `notas`. Y como la cantidad cambió, la función **la retorna**.

### `actualizar_nota`

````python
def actualizar_nota(ids, notas, cantidad, id_buscado, nota):
    """
    Cambia la nota de un estudiante.

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        Si id_buscado está inscrito, MODIFICA notas (su nota pasa a ser
        nota) y retorna True. Si no, no modifica nada y retorna False.
    """
    pos = buscar_id(ids, cantidad, id_buscado)
    encontrado = False
    if pos != -1:
        notas[pos] = nota
        encontrado = True
    return encontrado
````

Esta función no cambia la cantidad, así que no tiene que retornarla. Aprovecha el `return` para avisar **si pudo hacerlo**: `True` si encontró al estudiante, `False` si no. Solo recibe los arreglos que necesita: `ids`, para buscar, y `notas`, para cambiar.

````python
ids, nombres, notas, cantidad = curso_inicial()

print(actualizar_nota(ids, notas, cantidad, "987", 3.5))
print(notas)
print(actualizar_nota(ids, notas, cantidad, "555", 5.0))
````

### `mostrar_listado`

````python
def mostrar_listado(ids, nombres, notas, cantidad):
    """
    Supone: 0 <= cantidad <= len(ids).
    Garantiza: imprime una línea por estudiante inscrito (Id, nombre y
    nota). No modifica nada. No retorna nada.
    """
    print("Id", "\t", "Nombre", "\t", "Nota")
    for i in range(cantidad):
        print(ids[i], "\t", nombres[i], "\t", notas[i])
````

Es la única función que imprime. `"\t"` es un tabulador: separa las columnas. Como no retorna nada, no se prueba con `assert`: se ejecuta y se mira la salida.

## Parte 5 — Paso 3: ordenar

![Mapa: paso 3, ordenar](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/mapa_estructurada_paso3.png)

Ordenar el curso es ordenar **tres arreglos a la vez**, sin que ningún estudiante pierda sus datos. Se construye de abajo hacia arriba: primero intercambiar, después encontrar el mayor, y por último ordenar.

### `intercambiar`

Es la función de apoyo más pequeña, y sirve para cualquier arreglo.

````python
def intercambiar(lista, i, j):
    """
    Intercambia los elementos de las posiciones i y j de una lista.

    Supone:
        i y j son posiciones válidas de lista (0 a len(lista) - 1).
    Garantiza:
        MODIFICA lista: el valor que estaba en i queda en j, y el que
        estaba en j queda en i. No retorna nada.
    """
    temp = lista[i]
    lista[i] = lista[j]
    lista[j] = temp
````

````python
X = [29, 10, 14]
intercambiar(X, 0, 2)
assert X == [14, 10, 29]
intercambiar(X, 1, 1)                # límite: una posición consigo misma
assert X == [14, 10, 29]
print("intercambiar: pruebas correctas")
````

### `intercambiar_estudiantes`

````python
def intercambiar_estudiantes(ids, nombres, notas, i, j):
    """
    Supone: i y j son posiciones válidas de los tres arreglos.
    Garantiza: MODIFICA ids, nombres y notas: los estudiantes de las
    posiciones i y j quedan intercambiados en los tres. No retorna nada.
    """
    intercambiar(ids, i, j)
    intercambiar(nombres, i, j)
    intercambiar(notas, i, j)
````

Mover a un estudiante de posición exige **tres intercambios**, uno por arreglo, con las mismas dos posiciones.

Beto aparece con 3.0 y Ana con 4.5: **cada uno quedó con la nota del otro**, y no hay ningún mensaje. Por eso existe `intercambiar_estudiantes`: quien ordena la llama una vez y no puede olvidar un arreglo.

### `posicion_mayor_nota` y `ordenar_por_nota`

````python
def posicion_mayor_nota(notas, inicio, cantidad):
    """
    Supone: 0 <= inicio < cantidad <= len(notas).
    Garantiza: retorna la posición de la mayor nota entre inicio y
    cantidad - 1 (si se repite, la primera). No modifica notas.
    """
    pos_mayor = inicio
    for j in range(inicio + 1, cantidad):
        if notas[j] > notas[pos_mayor]:
            pos_mayor = j
    return pos_mayor
````

````python
def ordenar_por_nota(ids, nombres, notas, cantidad):
    """
    Ordena el curso de mayor a menor nota (selección).

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        MODIFICA ids, nombres y notas: los inscritos quedan de mayor a
        menor nota, y cada estudiante conserva su Id, nombre y nota.
        Las posiciones libres no cambian. No retorna nada.
    """
    for i in range(cantidad - 1):
        pos_mayor = posicion_mayor_nota(notas, i, cantidad)
        intercambiar_estudiantes(ids, nombres, notas, i, pos_mayor)
````

Es el método de **selección**: en cada pasada se busca la mayor nota entre las que faltan y se trae a su lugar. Con las notas del curso inicial:

| Pasada | Se busca desde | Mayor nota (posición) | Intercambio | Cómo queda |
|---|---|---|---|---|
| (antes) | | | | Ana 3.0, Beto 4.5, Carla 2.0, Dario 4.0 |
| `i = 0` | 0 | 4.5 (posición 1) | 0 con 1 | **Beto 4.5**, Ana 3.0, Carla 2.0, Dario 4.0 |
| `i = 1` | 1 | 4.0 (posición 3) | 1 con 3 | Beto 4.5, **Dario 4.0**, Carla 2.0, Ana 3.0 |
| `i = 2` | 2 | 3.0 (posición 3) | 2 con 3 | Beto 4.5, Dario 4.0, **Ana 3.0**, Carla 2.0 |

`posicion_mayor_nota` solo mira el arreglo `notas`. `intercambiar_estudiantes` mueve los tres.

### `posicion_menor_nombre` y `ordenar_por_nombre`

````python
def posicion_menor_nombre(nombres, inicio, cantidad):
    """
    Supone: 0 <= inicio < cantidad <= len(nombres). Los nombres empiezan
    con mayúscula y no llevan tilde (si no, < no sigue el orden del
    diccionario).
    Garantiza: retorna la posición del nombre que va primero en orden
    alfabético entre inicio y cantidad - 1. No modifica nombres.
    """
    pos_menor = inicio
    for j in range(inicio + 1, cantidad):
        if nombres[j] < nombres[pos_menor]:
            pos_menor = j
    return pos_menor
````

````python
def ordenar_por_nombre(ids, nombres, notas, cantidad):
    """
    Ordena el curso alfabéticamente por nombre (selección).

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        MODIFICA ids, nombres y notas: los inscritos quedan en orden
        alfabético, y cada estudiante conserva su Id, nombre y nota.
        Las posiciones libres no cambian. No retorna nada.
    """
    for i in range(cantidad - 1):
        pos_menor = posicion_menor_nombre(nombres, i, cantidad)
        intercambiar_estudiantes(ids, nombres, notas, i, pos_menor)
````

> 🔮 **Prediga.** Compare `ordenar_por_nombre` con `ordenar_por_nota`, y `posicion_menor_nombre` con `posicion_mayor_nota`. ¿En qué se diferencian?

Son **casi iguales**. Solo cambian el arreglo que se mira (`nombres` en vez de `notas`) y la comparación (`<` en vez de `>`). El operador `<` compara textos en orden alfabético, siempre que empiecen con mayúscula y no tengan tildes, como dice la especificación.

## Parte 6 — La integración: el programa principal

![Mapa: paso 4, el programa principal](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/mapa_estructurada_paso4.png)

Todas las funciones están probadas. El programa principal las reúne, y tiene una responsabilidad que ninguna función puede asumir: **guardar los datos del curso** (los tres arreglos y la cantidad) mientras el programa está abierto.

Para que el notebook siempre dé la misma salida, aquí no hay menú ni teclado: el programa ejecuta, en orden, las siete operaciones del caso calculado a mano en la parte 1.

````python
N = 10      # cupo máximo del curso


def main():
    # Los datos del curso: tres arreglos y una cantidad
    ids = [""] * N
    nombres = [""] * N
    notas = [0.0] * N
    cantidad = 0

    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "000", "Ana", 3.0)
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "123", "Beto", 4.5)
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "987", "Carla", 2.0)
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "007", "Dario", 4.0)
    print("--- Curso inicial ---")
    mostrar_listado(ids, nombres, notas, cantidad)

    print("--- 1. Agregar a Elena (555) ---")
    antes = cantidad
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "555", "Elena", 3.8)
    print("¿Se agregó?", cantidad != antes, "| inscritos:", cantidad)

    print("--- 2. Agregar a otro con el Id 123 ---")
    antes = cantidad
    cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "123", "Otro", 1.0)
    print("¿Se agregó?", cantidad != antes, "| inscritos:", cantidad)

    print("--- 3. Eliminar el Id 123 ---")
    antes = cantidad
    cantidad = eliminar_estudiante(ids, nombres, notas, cantidad, "123")
    print("¿Se eliminó?", cantidad != antes, "| inscritos:", cantidad)

    print("--- 4. Actualizar la nota de 987 a 3.5 ---")
    print("¿Se actualizó?", actualizar_nota(ids, notas, cantidad, "987", 3.5))

    print("--- 5. Buscar el Id 007 ---")
    pos = buscar_id(ids, cantidad, "007")
    print("Posición", pos, ":", ids[pos], nombres[pos], notas[pos])

    print("--- 6. Ordenar por nota ---")
    ordenar_por_nota(ids, nombres, notas, cantidad)
    mostrar_listado(ids, nombres, notas, cantidad)

    print("--- 7. Ordenar por nombre ---")
    ordenar_por_nombre(ids, nombres, notas, cantidad)
    mostrar_listado(ids, nombres, notas, cantidad)


main()
````

### Una operación, de punta a punta

Esto pasa cuando el programa principal elimina al estudiante `"123"`:

| # | Quién llama | A quién | Le entrega | Recibe |
|---|---|---|---|---|
| 1 | Programa principal | `eliminar_estudiante` | Los tres arreglos, la cantidad y el Id | — |
| 2 | `eliminar_estudiante` | `buscar_id` | `ids`, la cantidad y el Id | La posición: 1 |

| 4 | `eliminar_estudiante` | Programa principal | — | La cantidad nueva: 4 |
| 5 | Programa principal | (él mismo) | Guarda la cantidad en su variable | — |

Los arreglos quedaron cambiados desde el paso 3. La cantidad, en cambio, **viajó de vuelta** en el paso 4 y solo quedó actualizada cuando el programa principal la guardó en el paso 5.

## Parte 7 — Lo que resulta incómodo

El programa funciona y está probado. Aun así, al escribirlo aparecieron varias incomodidades. No son errores: son consecuencias de guardar los datos de un estudiante en tres arreglos.

| # | Incomodidad | Dónde se vio |
|---|---|---|
| 1 | **Un estudiante no existe como tal.** Sus tres datos están en tres arreglos, unidos solo por la posición | Parte 2; para mostrar a un estudiante hay que escribir `ids[pos], nombres[pos], notas[pos]` |
| 2 | **Cada intercambio son tres.** Si se olvida un arreglo, un estudiante queda con la nota de otro y no hay error | Parte 5, `intercambiar_estudiantes` |
| 3 | **`cantidad` va a todas partes.** Todas las funciones la reciben, y las que la cambian la retornan; si no se guarda, no hay error | Parte 3, `agregar_estudiante` |
| 4 | **Una función de ordenar por cada criterio.** `ordenar_por_nota` y `ordenar_por_nombre` son casi iguales | Parte 5 |

Hay una quinta, que se ve con una prueba:

Escriba una función `calcular_promedio(notas, cantidad)` que retorne el promedio de las notas de los inscritos. Si el curso está vacío, retorna 0.0. Escriba primero la especificación y las pruebas.

<details>
<summary>Ver solución</summary>

Es un acumulador, que recorre hasta `cantidad` y cuida la división por cero:

```python
def calcular_promedio(notas, cantidad):
    promedio = 0.0
    if cantidad > 0:
        suma = 0.0
        for i in range(cantidad):
            suma = suma + notas[i]
        promedio = suma / cantidad
    return promedio
```


### ✏️ Ejercicio 2 — `contar_aprobados`

Escriba una función `contar_aprobados(notas, cantidad, nota_minima)` que retorne cuántos inscritos tienen una nota mayor o igual a `nota_minima`.

<details>
<summary>Ver solución</summary>

```python
def contar_aprobados(notas, cantidad, nota_minima):
    aprobados = 0
    for i in range(cantidad):
        if notas[i] >= nota_minima:
            aprobados = aprobados + 1
    return aprobados
```

```python
ids, nombres, notas, cantidad = curso_inicial()
assert contar_aprobados(notas, cantidad, 3.0) == 3
assert contar_aprobados(notas, cantidad, 5.0) == 0
assert contar_aprobados(notas, 0, 3.0) == 0
```

### ✏️ Ejercicio 4 — ¿Qué habría que cambiar?

Responda sin programar. Se pide guardar también el **correo** de cada estudiante. ¿Qué datos nuevos hacen falta? ¿Cuáles de las diez funciones del curso habría que cambiar, y por qué?

Hace falta un cuarto arreglo, `correos = [""] * N`, y cambian **6 de las 10 funciones**, además del programa principal:

|---|---|
