## Parte 0 — El punto de partida: lo que incomoda sin objetos

En la versión con arreglos, los datos del curso son **tres arreglos paralelos** y una variable: el estudiante de la posición `i` tiene su Id en `ids[i]`, su nombre en `nombres[i]` y su nota en `notas[i]`, y `cantidad` dice cuántas casillas están ocupadas.

![Tres arreglos paralelos con cupo 10 y cantidad 4](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/arreglo_parcial.png)

Para ordenar el curso hay que cambiar estudiantes de posición, y eso exige intercambiar **en los tres arreglos**.

````python
def intercambiar(lista, i, j):
    """
    Intercambia los elementos de las posiciones i y j de una lista.

    Supone:
        i y j son posiciones válidas de lista (0 a len(lista) - 1).
    Garantiza:
        MODIFICA lista: el valor que estaba en i queda en j, y el que
        estaba en j queda en i. No retorna nada.
    """
    temp = lista[i]
    lista[i] = lista[j]
    lista[j] = temp
````

> ⚠️ **Error frecuente.** Se intercambian las posiciones 0 y 1, pero se olvida el arreglo de notas.

````python
ids = ["000", "123", "987", "007"]
nombres = ["Ana", "Beto", "Carla", "Dario"]
notas = [3.0, 4.5, 2.0, 4.0]
cantidad = 4

intercambiar(ids, 0, 1)
intercambiar(nombres, 0, 1)
# falta: intercambiar(notas, 0, 1)

for i in range(cantidad):
    print(ids[i], nombres[i], notas[i])
````

Esa versión funciona y está probada, pero deja estas incomodidades:

| # | Incomodidad |
|---|---|
| 1 | **Un estudiante no existe como tal**: sus tres datos están en tres arreglos, unidos solo por la posición |
| 2 | **Cada intercambio son tres**, y si se olvida uno no hay error |

La programación orientada a objetos propone otra forma de organizar el programa: **juntar cada dato con lo que se hace con él**.

## Parte 2 — El diseño: del enunciado a las clases y su relación

El enunciado no dice qué clases escribir. Para descubrirlas se **subrayan los sustantivos y los verbos**:

* Los **sustantivos** (cosas) son candidatos a **clases** o a **atributos**.
* Los **verbos** (acciones) son candidatos a **métodos**.

### 1. Marcar el enunciado

En **negrita**, los sustantivos; en *cursiva*, los verbos.

> Se desea un programa para *administrar* las **notas** de un **curso** con **cupo** para 10 **estudiantes**. De cada estudiante se conoce su **Id**, su **nombre** y su **nota** (entre 0.0 y 5.0).
>
> El programa debe permitir *agregar* un estudiante, *eliminar* un estudiante, *actualizar* la nota de un estudiante, *buscar* un estudiante por su Id, *listar* el curso, y *ordenar* el curso por nota o por nombre.

### 2. Decidir qué es cada sustantivo

Para cada sustantivo se hacen tres preguntas, en orden:

1. **¿Tiene varios datos propios y algo que sabe hacer?** Entonces es una **clase**.
2. **¿Es un dato sencillo que describe a otra cosa?** Entonces es un **atributo** de esa cosa.
3. **¿Se puede calcular a partir de otros datos?** Entonces no se guarda: es un **método**.

| Sustantivo | Decisión | Por qué |
|---|---|---|
| **estudiante** | **Clase `Estudiante`** | Tiene tres datos (Id, nombre, nota) y sabe cuidar su nota |
| Id, nombre | Atributos de `Estudiante` (`str`) | Datos sencillos que describen al estudiante |
| nota | Atributo de `Estudiante` (`float`) | Un dato sencillo, con una regla: va entre 0.0 y 5.0 |
| **curso** | **Clase `Curso`** | Guarda a los estudiantes y sabe agregar, eliminar, buscar, listar y ordenar |
| estudiantes (el grupo) | Atributo de `Curso`: **un arreglo de objetos `Estudiante`** | Describe al curso, pero no es un dato sencillo |
| cupo | El tamaño de ese arreglo | Se le da al curso al crearlo |
| (cuántos inscritos hay) | Atributo `cantidad` de `Curso` (`int`) | El enunciado no lo nombra, pero el arreglo no está siempre lleno |
| programa | Se descarta | Es el programa mismo |

Quedan **dos clases**: `Estudiante` y `Curso`.

La pista de que hay una **relación entre clases** está en el atributo cuyo tipo es otra clase: los *estudiantes* de un curso son objetos `Estudiante`.

### 3. Decidir de quién es cada verbo

Un método va en **la clase que tiene los datos necesarios para hacerlo**.

| Verbo del enunciado | ¿Dónde va? | Por qué ahí |
|---|---|---|
| cambiar la nota, cuidando el rango | Método `set_nota(nota)` de `Estudiante` | El estudiante tiene la nota y conoce su regla |
| agregar, eliminar | Métodos `agregar_estudiante(estudiante)` y `eliminar_estudiante(id)` de `Curso` | El curso tiene el arreglo y la cantidad |
| buscar | Método `buscar_id(id)` de `Curso` | Hay que recorrer a todos los inscritos |
| actualizar la nota de un estudiante | Método `actualizar_nota(id, nota)` de `Curso` | El curso sabe **encontrarlo**; el cambio se lo pide al estudiante |
| listar, ordenar | Métodos `mostrar_listado()`, `ordenar_por_nota()` y `ordenar_por_nombre()` de `Curso` | Trabajan con todo el grupo |
| intercambiar dos casillas | **Función** `intercambiar(lista, i, j)` | No necesita datos guardados: sirve para cualquier arreglo |

> **"Actualizar la nota" aparece dos veces, y no es un error.** Son dos tareas distintas que colaboran: el curso sabe *cuál* estudiante es, y el estudiante sabe *si la nota es válida*. Cada clase hace la parte para la que tiene los datos.

### 4. Decidir la relación

Hay una sola relación, y para ella se responden **tres preguntas**:

| Pregunta | Respuesta |
|---|---|
| ¿Quién conoce a quién? | El curso conoce a sus estudiantes. Un estudiante **no sabe** en qué curso está |
| ¿Quién crea el objeto? | El curso **no** crea estudiantes: los recibe ya creados en `agregar_estudiante` |
| ¿Quién le pide qué a quién? | El curso les pide a sus estudiantes su Id (para buscar), su nota y su nombre (para ordenar), que cambien su nota y que se conviertan en texto (para listar) |

La segunda pregunta decide el tipo de relación:

| Relación | Símbolo | Cuándo |
|---|---|---|
| **Composición** | Rombo negro | La clase **crea** el objeto que guarda |
| **Agregación** | Rombo blanco | La clase **recibe** el objeto ya creado y lo guarda como parte de un grupo |
| **Asociación** | Flecha | La clase **recibe** el objeto ya creado y lo conoce para pedirle cosas |

Como el curso recibe a sus estudiantes ya creados y los guarda en un grupo, la relación es de **agregación**. El rombo se dibuja del lado de la clase que guarda.

![Diagrama UML de las clases Estudiante y Curso](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/uml_estudiante_curso.png)

El diagrama se lee así:

* Arriba, el **nombre** de la clase; en el medio (azul), los **atributos**; abajo (amarillo), los **métodos**.
* `-` significa **privado** (solo se usa dentro de la clase) y `+` significa **público** (se puede usar desde afuera).
* Un atributo se escribe `nombre : tipo`. Un método se escribe `nombre(parámetros) : tipo de lo que retorna`.
* El **constructor** lleva el nombre de la clase y no tiene tipo de retorno. Los métodos que no retornan nada (como `mostrar_listado`) no llevan tipo.

## Parte 3 — Lo que no cambia: `intercambiar` sigue siendo una función

Para ordenar hace falta intercambiar dos casillas de un arreglo. Esa función ya existe, y se usa **sin cambiarle una letra**.

````python
def intercambiar(lista, i, j):
    """
    Intercambia los elementos de las posiciones i y j de una lista.

    Supone:
        i y j son posiciones válidas de lista (0 a len(lista) - 1).
    Garantiza:
        MODIFICA lista: el valor que estaba en i queda en j, y el que
        estaba en j queda en i. No retorna nada.
    """
    temp = lista[i]
    lista[i] = lista[j]
    lista[j] = temp
````

````python
X = [29, 10, 14]
intercambiar(X, 0, 2)
assert X == [14, 10, 29]
print("intercambiar: pruebas correctas")
````

No se convierte en método porque no tiene a quién pertenecer: no usa datos de ningún estudiante ni de ningún curso. **La programación orientada a objetos no reemplaza a las funciones.**

## Parte 4 — Paso 1: la clase `Estudiante`

![Mapa de clases: paso 1, Estudiante](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/mapa_poo_paso1.png)

`Estudiante` no conoce a ninguna otra clase. Se escribe y se prueba sola. Resuelve la primera incomodidad: **los tres datos quedan juntos**.

### Enunciado

> Diseñe una clase llamada `Estudiante`. La clase contiene:
>
> * Dos constantes, `NOTA_MINIMA` y `NOTA_MAXIMA`, con los valores 0.0 y 5.0.
> * Tres atributos privados: `id` (`str`), `nombre` (`str`) y `nota` (`float`).
> * Un constructor que recibe el Id, el nombre y la nota.
> * Métodos para consultar cada atributo.
> * Un método `set_nota`, que cambia la nota solo si está entre la mínima y la máxima, y retorna `True` si la cambió o `False` si no.
> * Un método que devuelve al estudiante en texto, como una línea del listado.
>
> El constructor **supone** que la nota que recibe es válida; no la revisa.

### Diagrama UML

![UML de Estudiante](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/uml_estudiante.png)

### La clase

````python
class Estudiante:
    NOTA_MINIMA = 0.0
    NOTA_MAXIMA = 5.0

    def __init__(self, id, nombre, nota):
        """
        Supone: id y nombre son textos; nota está entre NOTA_MINIMA y
        NOTA_MAXIMA (el constructor no valida).
        Garantiza: crea un estudiante con esos datos.
        """
        self.__id = id
        self.__nombre = nombre
        self.__nota = nota

    def get_id(self):
        return self.__id

    def get_nombre(self):
        return self.__nombre

    def get_nota(self):
        return self.__nota

    def set_nota(self, nota):
        """
        Garantiza: si nota está entre NOTA_MINIMA y NOTA_MAXIMA, la
        guarda y retorna True. Si no, no cambia nada y retorna False.
        """
        cambiada = False
        if Estudiante.NOTA_MINIMA <= nota <= Estudiante.NOTA_MAXIMA:
            self.__nota = nota
            cambiada = True
        return cambiada

    def __str__(self):
        """Garantiza: retorna 'Id \\t nombre \\t nota', una línea del listado."""
        return (self.__id + " \t " + self.__nombre
                + " \t " + str(self.__nota))
````

Tres cosas para notar:

* **Las constantes** se escriben dentro de la clase y fuera de los métodos, y se usan con el nombre de la clase: `Estudiante.NOTA_MAXIMA`. Hay una sola para todos los estudiantes.
* **No hay `set_id` ni `set_nombre`.** Un estudiante no cambia de Id ni de nombre: lo que no debe cambiar, no tiene método para cambiarlo.
* **`set_nota` cuida la regla del enunciado.** Como `__nota` es privado, este método es la única puerta para cambiar la nota, y la regla queda escrita **en un solo lugar**.

### Uso

````python
ana = Estudiante("000", "Ana", 3.0)
print(ana)
print(ana.get_nombre(), "tiene", ana.get_nota())

print(ana.set_nota(4.2))
print(ana)
````

## Parte 5 — Paso 2: la clase `Curso`, que guarda un arreglo de estudiantes

![Mapa de clases: paso 2, Curso](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/mapa_poo_paso2.png)

Aquí aparece la relación: un objeto que **guarda un grupo de objetos de otra clase**.

### Enunciado

> Diseñe una clase llamada `Curso`. La clase contiene:
>
> * Dos atributos privados: `estudiantes`, un arreglo de objetos `Estudiante`, y `cantidad` (`int`), que dice cuántos hay inscritos.
> * Un constructor que recibe el cupo y crea un curso vacío con esa capacidad.
> * Un método para consultar la cantidad, y un método `get_estudiante` que recibe una posición y retorna el estudiante que está ahí.
> * Un método `buscar_id`, que recibe un Id y retorna su posición, o -1 si no está inscrito.
> * Un método `agregar_estudiante`, que recibe un objeto `Estudiante` ya creado. Si hay cupo y su Id no está inscrito, lo guarda y retorna `True`; si no, retorna `False`.
> * Un método `eliminar_estudiante`, que recibe un Id. Si está inscrito, lo elimina y retorna `True`; si no, retorna `False`.
> * Un método `actualizar_nota`, que recibe un Id y una nota. Retorna `True` si pudo cambiarla, y `False` si el Id no existe o la nota no es válida.
> * Un método `mostrar_listado`, que imprime una línea por estudiante.
> * Métodos `ordenar_por_nota` (de mayor a menor) y `ordenar_por_nombre` (alfabético), con sus métodos de apoyo `posicion_mayor_nota` y `posicion_menor_nombre`.

### Diagrama UML

![UML de Curso](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/uml_curso.png)

### Las tres preguntas

| Pregunta | Respuesta |
|---|---|
| ¿Quién conoce a quién? | El curso conoce a sus estudiantes. Un estudiante no sabe en qué curso está |
| ¿Quién crea el objeto? | **Otro** (el programa principal): los estudiantes llegan ya creados a `agregar_estudiante`. Es **agregación** |
| ¿Quién le pide qué a quién? | El curso les pide `get_id()`, `get_nota()`, `get_nombre()`, `set_nota(...)` y su texto |

### La clase

````python
class Curso:
    def __init__(self, cupo):
        """
        Supone: cupo > 0.
        Garantiza: crea un curso vacío con capacidad para cupo estudiantes.
        """
        self.__estudiantes = [None] * cupo
        self.__cantidad = 0

    def get_cantidad(self):
        return self.__cantidad

    def get_estudiante(self, posicion):
        """
        Supone: 0 <= posicion < cantidad.
        Garantiza: retorna el estudiante de esa posición.
        """
        return self.__estudiantes[posicion]

    def buscar_id(self, id):
        """
        Garantiza: retorna la posición del estudiante con ese Id
        (búsqueda lineal), o -1 si no está inscrito.
        """
        pos = -1
        for i in range(self.__cantidad):
            if self.__estudiantes[i].get_id() == id:
                pos = i
                break
        return pos

    def agregar_estudiante(self, estudiante):
        """
        Garantiza: si hay cupo y su Id no está inscrito, guarda el
        estudiante al final de los inscritos y retorna True. Si no, no
        cambia nada y retorna False.
        """
        agregado = False
        if (self.__cantidad < len(self.__estudiantes)
                and self.buscar_id(estudiante.get_id()) == -1):
            self.__estudiantes[self.__cantidad] = estudiante
            self.__cantidad = self.__cantidad + 1
            agregado = True
        return agregado

    def eliminar_estudiante(self, id):
        """
        Garantiza: si el Id está inscrito, lo elimina corriendo a la
        izquierda los estudiantes que estaban después y retorna True.
        Si no, no cambia nada y retorna False.
        """
        pos = self.buscar_id(id)
        eliminado = False
        if pos != -1:
            for i in range(pos, self.__cantidad - 1):
                self.__estudiantes[i] = self.__estudiantes[i + 1]
            self.__cantidad = self.__cantidad - 1
            self.__estudiantes[self.__cantidad] = None
            eliminado = True
        return eliminado

    def actualizar_nota(self, id, nota):
        """
        Garantiza: si el Id está inscrito y la nota es válida (la valida
        set_nota), cambia su nota y retorna True. Si no, retorna False.
        """
        pos = self.buscar_id(id)
        actualizada = False
        if pos != -1:
            actualizada = self.__estudiantes[pos].set_nota(nota)
        return actualizada

    def mostrar_listado(self):
        """Garantiza: imprime una línea por estudiante inscrito. No retorna nada."""
        print("Id", "\t", "Nombre", "\t", "Nota")
        for i in range(self.__cantidad):
            print(self.__estudiantes[i])

    def posicion_mayor_nota(self, inicio=0):
        """
        Supone: 0 <= inicio < cantidad.
        Garantiza: retorna la posición de la mayor nota entre inicio y
        cantidad - 1 (si se repite, la primera).
        """
        pos_mayor = inicio
        for j in range(inicio + 1, self.__cantidad):
            if (self.__estudiantes[j].get_nota()
                    > self.__estudiantes[pos_mayor].get_nota()):
                pos_mayor = j
        return pos_mayor

    def ordenar_por_nota(self):
        """
        Garantiza: los inscritos quedan de mayor a menor nota (selección).
        Cada intercambio mueve un estudiante completo. No retorna nada.
        """
        for i in range(self.__cantidad - 1):
            pos_mayor = self.posicion_mayor_nota(i)
            intercambiar(self.__estudiantes, i, pos_mayor)

    def posicion_menor_nombre(self, inicio=0):
        """
        Supone: 0 <= inicio < cantidad. Los nombres empiezan con mayúscula
        y no llevan tilde (si no, < no sigue el orden del diccionario).
        Garantiza: retorna la posición del nombre que va primero en orden
        alfabético entre inicio y cantidad - 1.
        """
        pos_menor = inicio
        for j in range(inicio + 1, self.__cantidad):
            if (self.__estudiantes[j].get_nombre()
                    < self.__estudiantes[pos_menor].get_nombre()):
                pos_menor = j
        return pos_menor

    def ordenar_por_nombre(self):
        """
        Garantiza: los inscritos quedan en orden alfabético (selección).
        No retorna nada.
        """
        for i in range(self.__cantidad - 1):
            pos_menor = self.posicion_menor_nombre(i)
            intercambiar(self.__estudiantes, i, pos_menor)
````

La clase es larga, pero cada método es **el mismo algoritmo** de la versión con arreglos: búsqueda lineal, correr a la izquierda, selección. Lo que cambia es dónde están los datos.

| En la versión con arreglos | En la clase `Curso` |
|---|---|
| Tres arreglos: `ids`, `nombres`, `notas` | **Un** arreglo: `self.__estudiantes`, con un objeto por casilla |
| Casillas libres con `""` y `0.0` | Casillas libres con `None`, el valor de Python para "no hay objeto" |
| La variable `cantidad`, que se pasa y se retorna | El atributo `self.__cantidad`, que el curso actualiza por dentro |
| `ids[i] == id_buscado` | `self.__estudiantes[i].get_id() == id` |
| Cada función recibe los arreglos y la cantidad | Cada método ya los tiene en `self`, y recibe solo lo nuevo |

### Así queda en la memoria

![Un objeto Curso con un arreglo de referencias a objetos Estudiante](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/curso_objetos.png)

Cada casilla ocupada guarda una **referencia**: una flecha hacia un objeto `Estudiante`. Los datos de un estudiante viajan juntos, dentro de su objeto.

### Crear el curso y agregar

Primero se crean los estudiantes. Después se le **entregan** al curso.

La línea `curso.get_estudiante(pos).get_nombre()` se lee de **izquierda a derecha**, y cada punto es una petición a un objeto distinto:

| Parte | ¿A quién se le pide? | ¿Qué retorna? |
|---|---|---|
| `curso.get_estudiante(pos)` | Al curso | Un objeto `Estudiante` |
| `curso.get_estudiante(pos).get_nombre()` | A ese estudiante | Un `str`: `"Dario"` |

Para saber qué método se puede escribir después de un punto, hay que preguntar: **¿de qué clase es lo que retornó la parte anterior?**

### Ordenar: un solo intercambio

![Intercambio de dos referencias en el arreglo de objetos](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/intercambio_objetos.png)

Para cambiar dos estudiantes de posición basta **un intercambio**: las dos casillas cambian de referencia, y cada estudiante sigue con sus datos. No hay forma de dejar a alguien con la nota de otro.

Y se hace con la misma función `intercambiar` de la parte 3, sin modificarla.

````python
curso = Curso(10)
curso.agregar_estudiante(Estudiante("000", "Ana", 3.0))
curso.agregar_estudiante(Estudiante("123", "Beto", 4.5))
curso.agregar_estudiante(Estudiante("987", "Carla", 2.0))
curso.agregar_estudiante(Estudiante("007", "Dario", 4.0))

curso.ordenar_por_nota()
curso.mostrar_listado()

curso.ordenar_por_nombre()
curso.mostrar_listado()
````

Dentro de la clase, un método le pide ayuda a otro método **del mismo objeto** con `self.`: `ordenar_por_nota` llama a `self.posicion_mayor_nota(i)`.

`ordenar_por_nota` y `ordenar_por_nombre` siguen siendo casi iguales. Esa incomodidad **no desapareció**; la parte 7 vuelve sobre ella.

### Pruebas

Las pruebas son las mismas de la versión con arreglos, escritas con objetos, más un caso nuevo: la nota inválida. Usan cuatro funciones de ayuda: una arma el curso inicial, y las otras tres sacan los Id, los nombres y las notas del curso en arreglos, para compararlos con lo esperado.

````python
N = 10


def curso_inicial():
    """Garantiza: retorna un Curso de cupo N con los 4 estudiantes de ejemplo."""
    curso = Curso(N)
    curso.agregar_estudiante(Estudiante("000", "Ana", 3.0))
    curso.agregar_estudiante(Estudiante("123", "Beto", 4.5))
    curso.agregar_estudiante(Estudiante("987", "Carla", 2.0))
    curso.agregar_estudiante(Estudiante("007", "Dario", 4.0))
    return curso


def ids_del_curso(curso):
    """Garantiza: retorna una lista con los Id de los inscritos, en orden."""
    ids = [""] * curso.get_cantidad()
    for i in range(curso.get_cantidad()):
        ids[i] = curso.get_estudiante(i).get_id()
    return ids


def nombres_del_curso(curso):
    """Garantiza: retorna una lista con los nombres de los inscritos, en orden."""
    nombres = [""] * curso.get_cantidad()
    for i in range(curso.get_cantidad()):
        nombres[i] = curso.get_estudiante(i).get_nombre()
    return nombres


def notas_del_curso(curso):
    """Garantiza: retorna una lista con las notas de los inscritos, en orden."""
    notas = [0.0] * curso.get_cantidad()
    for i in range(curso.get_cantidad()):
        notas[i] = curso.get_estudiante(i).get_nota()
    return notas
````

La prueba de `Curso` ya es una **integración**: ejercita las dos clases y la función `intercambiar` a la vez.

| Método | Casos |
|---|---|
| `buscar_id` | Uno del medio, el primero, el último, uno que no está, y un curso vacío |
| `agregar_estudiante` | Típico, Id repetido y curso lleno |
| `eliminar_estudiante` | Uno del medio, el último y uno que no está |
| `actualizar_nota` | Válida, Id que no está y **nota inválida** |
| Ordenar | Por nota, por nombre, con notas repetidas, curso vacío y un solo estudiante |

````python
# buscar_id
curso = curso_inicial()
assert curso.get_cantidad() == 4
assert curso.buscar_id("987") == 2          # típico
assert curso.buscar_id("000") == 0          # límite: el primero
assert curso.buscar_id("007") == 3          # límite: el último
assert curso.buscar_id("555") == -1         # no está
assert Curso(N).buscar_id("000") == -1      # curso vacío

# agregar_estudiante
curso = curso_inicial()
assert curso.agregar_estudiante(Estudiante("555", "Elena", 3.8)) == True
assert curso.get_cantidad() == 5 and curso.get_estudiante(4).get_nombre() == "Elena"
assert curso.agregar_estudiante(Estudiante("123", "Otro", 1.0)) == False   # Id repetido
assert curso.get_cantidad() == 5 and curso.get_estudiante(1).get_nombre() == "Beto"
for k in range(5, N):                       # llenar el cupo
    curso.agregar_estudiante(Estudiante("X" + str(k), "E" + str(k), 1.0))
assert curso.get_cantidad() == N
assert curso.agregar_estudiante(Estudiante("999", "Lleno", 5.0)) == False  # sin cupo
assert curso.get_cantidad() == N and curso.buscar_id("999") == -1

# eliminar_estudiante
curso = curso_inicial()
assert curso.eliminar_estudiante("123") == True                 # del medio
assert ids_del_curso(curso) == ["000", "987", "007"]
assert nombres_del_curso(curso) == ["Ana", "Carla", "Dario"]
assert notas_del_curso(curso) == [3.0, 2.0, 4.0]
assert curso.eliminar_estudiante("007") == True                 # el último
assert ids_del_curso(curso) == ["000", "987"]
assert curso.eliminar_estudiante("555") == False                # no está
assert curso.get_cantidad() == 2

# actualizar_nota
curso = curso_inicial()
assert curso.actualizar_nota("987", 3.5) == True
assert curso.get_estudiante(2).get_nota() == 3.5
assert curso.actualizar_nota("555", 5.0) == False               # no está
assert curso.actualizar_nota("987", 7.0) == False               # nota inválida
assert notas_del_curso(curso) == [3.0, 4.5, 3.5, 4.0]

# ordenar_por_nota (de mayor a menor)
curso = curso_inicial()
curso.ordenar_por_nota()
assert notas_del_curso(curso) == [4.5, 4.0, 3.0, 2.0]
assert nombres_del_curso(curso) == ["Beto", "Dario", "Ana", "Carla"]
assert ids_del_curso(curso) == ["123", "007", "000", "987"]

# ordenar_por_nombre (alfabético)
curso = curso_inicial()
curso.ordenar_por_nota()                    # se desordena primero
curso.ordenar_por_nombre()
assert nombres_del_curso(curso) == ["Ana", "Beto", "Carla", "Dario"]
assert ids_del_curso(curso) == ["000", "123", "987", "007"]
assert notas_del_curso(curso) == [3.0, 4.5, 2.0, 4.0]

# ordenar_por_nota con notas repetidas: cada nombre sigue con su nota
curso = Curso(3)
curso.agregar_estudiante(Estudiante("1", "Ana", 4.0))
curso.agregar_estudiante(Estudiante("2", "Beto", 3.0))
curso.agregar_estudiante(Estudiante("3", "Carla", 4.0))
curso.ordenar_por_nota()
assert notas_del_curso(curso) == [4.0, 4.0, 3.0]
assert nombres_del_curso(curso) == ["Ana", "Carla", "Beto"]

# Casos especiales de ordenar: curso vacío y un solo estudiante
curso = Curso(N)
curso.ordenar_por_nota()
assert curso.get_cantidad() == 0
curso.agregar_estudiante(Estudiante("000", "Ana", 3.0))
curso.ordenar_por_nombre()
assert nombres_del_curso(curso) == ["Ana"]

print("Pruebas de curso: todas pasaron")
````

## Parte 6 — La integración: el programa principal

![Mapa de clases: todas probadas, y el programa principal](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/mapa_poo_integracion.png)

Las dos clases están escritas y probadas. Falta el programa que las usa.

Hay dos papeles distintos. Quien **escribió** las clases conoce sus atributos y sus detalles. El **programa principal** es su cliente: solo las usa, a través de sus métodos públicos. Hace tres cosas:

1. **Crea** el curso y los estudiantes.
2. **Pide**: le encarga cada operación al curso.
3. **Muestra** lo que el curso responde.

| El programa principal sí | El programa principal no |
|---|---|
| Crea los objetos `Estudiante` y el `Curso` | Guarda arreglos ni lleva la cuenta de cuántos hay |
| Le pide al curso `agregar_estudiante(...)`, `ordenar_por_nota()`… | Recorre, corre casillas ni intercambia |
| Revisa si la respuesta fue `True` o `False` | Revisa si una nota es válida |

El programa ejecuta, en orden, las siete operaciones del caso calculado a mano.

````python
N = 10      # cupo máximo del curso


def main():
    # 1. Crear
    curso = Curso(N)
    curso.agregar_estudiante(Estudiante("000", "Ana", 3.0))
    curso.agregar_estudiante(Estudiante("123", "Beto", 4.5))
    curso.agregar_estudiante(Estudiante("987", "Carla", 2.0))
    curso.agregar_estudiante(Estudiante("007", "Dario", 4.0))
    print("--- Curso inicial ---")
    curso.mostrar_listado()

    # 2. Pedir   3. Mostrar
    print("--- 1. Agregar a Elena (555) ---")
    elena = Estudiante("555", "Elena", 3.8)
    print("¿Se agregó?", curso.agregar_estudiante(elena), "| inscritos:", curso.get_cantidad())

    print("--- 2. Agregar a otro con el Id 123 ---")
    otro = Estudiante("123", "Otro", 1.0)
    print("¿Se agregó?", curso.agregar_estudiante(otro), "| inscritos:", curso.get_cantidad())

    print("--- 3. Eliminar el Id 123 ---")
    print("¿Se eliminó?", curso.eliminar_estudiante("123"), "| inscritos:", curso.get_cantidad())

    print("--- 4. Actualizar la nota de 987 a 3.5 ---")
    print("¿Se actualizó?", curso.actualizar_nota("987", 3.5))

    print("--- 5. Buscar el Id 007 ---")
    pos = curso.buscar_id("007")
    estudiante = curso.get_estudiante(pos)
    print("Posición", pos, ":", estudiante.get_id(), estudiante.get_nombre(), estudiante.get_nota())

    print("--- 6. Ordenar por nota ---")
    curso.ordenar_por_nota()
    curso.mostrar_listado()

    print("--- 7. Ordenar por nombre ---")
    curso.ordenar_por_nombre()
    curso.mostrar_listado()


main()
````

## Parte 7 — Las dos versiones, frente a frente

### Las mismas operaciones, escritas de dos formas

| Operación | Con arreglos paralelos | Con objetos |
|---|---|---|
| Crear el curso | `ids = [""] * N`<br>`nombres = [""] * N`<br>`notas = [0.0] * N`<br>`cantidad = 0` | `curso = Curso(N)` |
| Agregar | `cantidad = agregar_estudiante(ids, nombres, notas, cantidad, "555", "Elena", 3.8)` | `curso.agregar_estudiante(Estudiante("555", "Elena", 3.8))` |
| Eliminar | `cantidad = eliminar_estudiante(ids, nombres, notas, cantidad, "123")` | `curso.eliminar_estudiante("123")` |
| Actualizar | `actualizar_nota(ids, notas, cantidad, "987", 3.5)` | `curso.actualizar_nota("987", 3.5)` |
| Buscar | `pos = buscar_id(ids, cantidad, "007")` | `pos = curso.buscar_id("007")` |
| Listar | `mostrar_listado(ids, nombres, notas, cantidad)` | `curso.mostrar_listado()` |
| Ordenar | `ordenar_por_nota(ids, nombres, notas, cantidad)` | `curso.ordenar_por_nota()` |

Los algoritmos son los mismos. Lo que cambió es **dónde viven los datos**: los métodos ya los tienen en `self`, y solo reciben lo nuevo.

### El tablero de incomodidades

![Tablero de incomodidades: arreglos paralelos frente a objetos](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_curso/images/tablero_final.png)

Tres incomodidades quedaron resueltas, y también la del correo: con objetos cambia `Estudiante` (un atributo, un método para consultarlo y quizá su texto) y **`Curso` no cambia**, porque guarda estudiantes sin importarle cuántos datos tenga cada uno. El ejercicio 1 de la parte 9 lo comprueba.

### Lo que los objetos no resuelven

La fila 4 queda en rojo: **sigue habiendo un método de ordenar por cada criterio**. `ordenar_por_nota` y `ordenar_por_nombre` son casi iguales; solo cambia la comparación. Si se pidiera ordenar por Id, habría que escribir un tercero.

Los objetos no eliminan esa repetición. Para eso hacen falta herramientas que este curso no cubre, como entregarle al método la forma de comparar. Conviene saberlo: **la programación orientada a objetos organiza mejor los datos, pero no lo resuelve todo.**

`set_nota` rechaza el 9.5, pero el estudiante **nació** con 9.0. El enunciado lo advertía: el constructor *supone* una nota válida y no la revisa. Es una decisión de diseño, y deja la responsabilidad en quien crea el objeto. El programa principal debería revisar la nota antes de crear al estudiante.

### Resumen

| Qué pasa | Qué significa | Cómo se corrige |
|---|---|---|
| `TypeError: ... takes 2 positional arguments but 4 were given` | Se entregaron datos sueltos donde se esperaba un objeto | Crear el objeto y entregarlo |
| `AttributeError: 'NoneType' object has no attribute ...` | Se le pidió un método a `None`: en esa casilla no hay objeto | Usar solo posiciones entre 0 y `cantidad - 1` |
| El programa dice que hizo algo que no hizo, y no hay mensaje | No se revisó el `True` o `False` que retornó el método | `if curso.agregar_estudiante(...):` |
| Un objeto cambia "solo", y no hay mensaje | Dos lugares guardan una referencia al mismo objeto | Decidir si se quería el mismo objeto o uno nuevo |
| Un dato inválido entra, y no hay mensaje | No se cumplió lo que el constructor supone | Revisar antes de crear el objeto |

Tres de los cinco **no dan mensaje**. Para encontrarlos sirve dibujar los objetos y sus flechas, o verlos en Python Tutor.

class Estudiante:
    NOTA_MINIMA = 0.0

    def __init__(self, id, nombre, nota, correo):

        self.__nombre = nombre
        self.__nota = nota
        self.__correo = correo

    def get_id(self):

        cambiada = False
        if Estudiante.NOTA_MINIMA <= nota <= Estudiante.NOTA_MAXIMA:

            cambiada = True

    def contar_aprobados(self, nota_minima):

        return aprobados
