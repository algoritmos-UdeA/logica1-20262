## Parte 2 — Paradigmas de programación

Un **paradigma** es una forma de ver el problema y de organizar su solución. No es un lenguaje ni una instrucción: es qué se pregunta uno primero, antes de escribir código.

| Paradigma | Idea central | La pregunta de partida | Lenguajes |
|---|---|---|---|
| Programación estructurada | La solución se arma con secuencia, selección e iteración, organizadas en funciones | ¿Qué pasos se siguen, y en qué orden? | C, Pascal, Fortran |
| Programación funcional | La solución se descompone en funciones, como en matemáticas | ¿Qué transformaciones se aplican? | Haskell, Lisp, Erlang, Elixir |
| Programación orientada a objetos (POO) | El mundo es un conjunto de objetos, con datos y comportamiento, que interactúan | ¿Quiénes participan, y qué sabe hacer cada uno? | Java, C++, C#, Smalltalk |
| Programación reactiva | El programa reacciona a un flujo de datos que cambia con el tiempo | ¿A qué eventos se reacciona? | Elm; bibliotecas como RxJS y RxJava |

Una analogía: atender un restaurante para 20 personas. En **estructurada** es una receta (primero esto, luego aquello, se repite para cada plato); en **funcional**, estaciones (lo crudo sale picado, lo picado sale cocido); en **POO**, roles (el mesero toma pedidos, el cocinero cocina, la caja cobra); en **reactiva**, pedidos (la cocina espera y cada pedido que llega dispara el trabajo).

Python permite varios paradigmas (es **multiparadigma**). Hasta ahora el curso ha usado la programación estructurada.

### El mismo problema, dos paradigmas

Calcular el promedio de las notas 3.0, 4.5, 2.0 y 4.0. En programación estructurada, los **datos están sueltos** y la operación está aparte: tiene que saber dónde están los datos.

![Programación estructurada: datos sueltos y operación aparte](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/paradigma_estructurada.png)

````python
notas = [3.0, 4.5, 2.0, 4.0]
suma = 0.0
for i in range(len(notas)):
    suma = suma + notas[i]
print(suma / len(notas))
````

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=notas%20%3D%20%5B3.0%2C%204.5%2C%202.0%2C%204.0%5D%0Asuma%20%3D%200.0%0Afor%20i%20in%20range%28len%28notas%29%29%3A%0A%20%20%20%20suma%20%3D%20suma%20%2B%20notas%5Bi%5D%0Aprint%28suma%20%2F%20len%28notas%29%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

En POO, el objeto `curso` **tiene sus notas** y **sabe calcular su promedio**: los datos y la operación viajan juntos. El código de la clase se explica en las partes siguientes; por ahora basta comparar cómo se usa: `curso.calcular_promedio()`.

![POO: el curso tiene sus datos y sabe calcular](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/paradigma_poo.png)

````python
class Curso:
    def __init__(self, notas):
        self.notas = notas

    def calcular_promedio(self):
        suma = 0.0
        for i in range(len(self.notas)):
            suma = suma + self.notas[i]
        return suma / len(self.notas)


curso = Curso([3.0, 4.5, 2.0, 4.0])
print(curso.calcular_promedio())
````

## Parte 3 — Clase y objeto

### ¿Cómo se guardan juntos datos de distinto tipo?

Un estudiante tiene un Id (texto), un nombre (texto) y una nota (real). Un arreglo no sirve para guardarlos juntos: sus datos son todos del mismo tipo.

| Estructura | Tipos de datos | Cada dato se identifica por | Ejemplo |
|---|---|---|---|
| Arreglo | Uno solo | Su posición | `notas[2]` |
| Registro (`struct` en C, `record` en Pascal) | Varios | Su nombre | `ana.nota` |
| Clase (POO) | Varios | Su nombre, y además tiene métodos propios | `ana.set_nota(4.0)` |

Python permite listas con tipos mezclados, como `["000", "Ana", 3.0]`, pero esa lista identifica cada dato por su posición: `ana[2]` no dice que es la nota. En el curso, las listas se usan como arreglos: un solo tipo. Una **clase** es como un registro que además sabe hacer cosas.

### El plano y las casas

Una **clase** es como el plano de una casa: dice qué tiene cada casa (color, puerta) y qué se puede hacer con ella (abrir, cerrar). Un **objeto** es cada casa construida con ese plano: todas tienen la misma forma, pero cada una tiene sus propios valores. De una clase se crean muchos objetos, también llamados **instancias**.

![Un plano, muchas casas](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/plano_casas.png)

### Estado, comportamiento e identidad

Todo objeto tiene tres características:

| Un objeto tiene | Qué es | En las casas |
|---|---|---|
| **Estado** | Los valores de sus datos en un momento dado | `casa_1`: color azul, puerta cerrada. Si se pinta o se abre la puerta, su estado cambia |
| **Comportamiento** | Lo que sabe hacer | Abrir y cerrar la puerta |
| **Identidad** | Es uno solo, distinto de cualquier otro, aunque tenga los mismos datos | La dirección: dos casas idénticas siguen siendo dos casas |

![Estado, comportamiento e identidad de una casa](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/objeto_casa.png)

### Del plano de la casa al estudiante

Lo mismo con los estudiantes de un curso: la clase es `Estudiante`; los objetos, `lisa` y `bart`. El estado de `lisa` es su Id `"742"`, su nombre `"Lisa"` y su nota 5.0; su comportamiento, consultar y cambiar la nota (`get_nota()`, `set_nota()`). Si Bart copia las notas de Lisa, siguen siendo dos estudiantes: esa es la identidad.

![La clase Estudiante y dos objetos](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/estudiante_objetos.png)

De aquí en adelante, las clases y los objetos se dibujan en **UML**: sin dibujos, solo con cajas.

## Parte 4 — De la idea a UML y de UML a Python

### ¿Qué permite la POO?

| Propiedad | Qué permite | En los estudiantes |
|---|---|---|
| **Abstracción** | Quedarse solo con lo que importa para el problema | Se guardan su Id, su nombre y su nota, no su estatura |
| **Encapsulamiento** | Juntar los datos con sus métodos y proteger los datos | La nota solo se cambia con `set_nota()` |
| Herencia | Crear una clase a partir de otra | Un estudiante de posgrado es un estudiante, con algo más: su tesis |
| Polimorfismo | Un mismo método, con comportamiento distinto según el objeto | `calcular_nota_final()` en pregrado y en posgrado |

En este curso se ven la **abstracción** y el **encapsulamiento**. La herencia y el polimorfismo corresponden a cursos posteriores.

### Encapsulamiento: los datos se protegen

El objeto guarda sus datos en el centro: son **privados**. Alrededor están sus métodos: son **públicos**. El código de afuera no toca los datos directamente: le pide a un método lo que necesita. Como el sistema de notas de la universidad: nadie entra a la base de datos a escribir una nota; se consulta y se cambia a través del sistema, que revisa que la nota sea válida.

![Encapsulamiento: datos privados, métodos públicos](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/encapsulamiento.png)

### ¿Por qué UML?

Entre 1989 y 1994, los métodos para dibujar programas orientados a objetos pasaron de menos de 10 a más de 50: cada equipo dibujaba distinto. Booch, Rumbaugh y Jacobson unificaron sus métodos, y en 1997 el OMG (Object Management Group) adoptó **UML** (Lenguaje Unificado de Modelado) como el estándar para dibujar un programa orientado a objetos **antes de escribirlo**.

![Línea de tiempo de UML](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/linea_tiempo_uml.png)

### La plantilla de una clase

En UML, cada clase es una caja de tres partes: el **nombre**, los **atributos** (sus características) y los **métodos** (lo que sabe hacer).

![La plantilla de una clase en UML](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/plantilla_uml.png)

| Elemento | Se escribe | Ejemplo |
|---|---|---|
| Atributo | `visibilidad nombre : tipo` | `- nota : float` |
| Constructor | `visibilidad NombreDeLaClase(parámetros)` | `+ Estudiante(id: str, nombre: str, nota: float)` |
| Método | `visibilidad nombre(parámetros) : tipo de retorno` | `+ set_nota(nota: float) : bool` |

Un método que no devuelve nada **no lleva tipo de retorno** en UML; en Python, devuelve `None`.

### Visibilidad

| Símbolo | Visibilidad | Quién puede usarlo | En Python |
|---|---|---|---|
| `+` | Público | Cualquier código | Nombre normal: `base` |
| `-` | Privado | Solo la propia clase | Dos guiones bajos: `__base` |
| `#` | Protegido | La clase y las que heredan de ella | Un guion bajo: `_base` (solo convención) |
| `~` | Paquete | Las clases del mismo paquete | No existe en Python |

En este curso se usan `+` para los métodos y `-` para los atributos.

### De UML a Python

![Cada parte del diagrama UML en Python](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_a_python.png)

| Elemento | En UML | En Python |
|---|---|---|
| Clase | `NombreDeLaClase` | `class NombreDeLaClase:` |
| Constructor | `+ NombreDeLaClase(parametros)` | `def __init__(self, parametros):` |
| Atributo | `- atributo : tipo` | `self.__atributo = valor`, dentro de `__init__` |
| Método | `+ metodo(parametros) : tipo` | `def metodo(self, parametros):` |
| Crear un objeto | — | `objeto = NombreDeLaClase(valores)` |
| Usar un método | — | `objeto.metodo(valores)` |

Lo que más cambia:

* El constructor se escribe `__init__`, pero se usa con el nombre de la clase: `r1 = Rectangulo(3.0, 4.0)`.
* Los atributos no se declaran aparte: nacen dentro de `__init__`, con `self`.
* Todo método recibe `self` como primer parámetro: es el objeto que se está usando. En UML, `self` no aparece.

## Parte 5 — Ejemplo 1: `Rectangulo`

### Enunciado

> Diseñe una clase llamada `Rectangulo` para representar un rectángulo. La clase contiene:
>
> * Dos atributos, `base` y `altura`.
> * Un constructor que crea un rectángulo con la base y la altura dadas.
> * Un método `calcular_area()` que retorna el área.
> * Un método `calcular_perimetro()` que retorna el perímetro.
>
> Dibuje el diagrama UML y luego impleméntela. Pruébela con dos rectángulos: de 3.0 × 4.0 y de 5.0 × 2.0.

![El rectángulo, su área y su perímetro](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/figura_rectangulo.png)

### Diagrama UML

El enunciado no dice los tipos: se deciden al diseñar. La base y la altura pueden tener decimales: `float`. Por ahora, los atributos son públicos (`+`).

![UML de Rectangulo con atributos públicos](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_rectangulo.png)

### La clase en Python

Es la plantilla aplicada: el nombre en `class`, los atributos dentro de `__init__` con `self`, y cada método en su `def`, con `self` como primer parámetro. Esta celda solo **define** la clase: no imprime nada.

````python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return self.base * self.altura

    def calcular_perimetro(self):
        return 2 * self.base + 2 * self.altura
````

### Ejemplo 1 mejorado: atributos privados

**Privado** (`-`): solo los métodos de la clase pueden leer o cambiar el atributo. Desde fuera se usan métodos: `get_` para consultar y `set_` para cambiar, **validando**. En Python, un atributo es privado si su nombre empieza con dos guiones bajos: `self.__base`.

![UML de Rectangulo con atributos privados](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_rectangulo_privado.png)

En esta implementación, `set_base` y `set_altura` devuelven `True` si aceptaron el valor y `False` si lo rechazaron, con una bandera (`cambiada`). Validar es una decisión de diseño; en la industria, en vez de devolver `False`, se lanzaría un error (una *excepción*), que el curso no ve.

````python
class Rectangulo:
    def __init__(self, base, altura):
        self.__base = base
        self.__altura = altura

    def get_base(self):
        return self.__base

    def get_altura(self):
        return self.__altura

    def set_base(self, base):
        cambiada = False
        if base > 0:
            self.__base = base
            cambiada = True
        return cambiada

    def set_altura(self, altura):
        cambiada = False
        if altura > 0:
            self.__altura = altura
            cambiada = True
        return cambiada

    def calcular_area(self):
        return self.__base * self.__altura

    def calcular_perimetro(self):
        return 2 * self.__base + 2 * self.__altura
````

* `__str__` devuelve el texto que muestran `print(v)` y `str(v)`. Sin él, `print` mostraría algo como `<__main__.Ventilador object at 0x...>`.

### La clase en Python

````python
class Ventilador:
    LENTO = 1
    MEDIO = 2
    RAPIDO = 3

    def __init__(self, velocidad=1, radio=5.0, color="azul", encendido=False):
        self.__velocidad = velocidad
        self.__radio = radio
        self.__color = color
        self.__encendido = encendido

    def get_velocidad(self):
        return self.__velocidad

    def get_radio(self):
        return self.__radio

    def get_color(self):
        return self.__color

    def esta_encendido(self):
        return self.__encendido

    def set_velocidad(self, velocidad):
        cambiada = False
        if (velocidad >= Ventilador.LENTO and
                velocidad <= Ventilador.RAPIDO):
            self.__velocidad = velocidad
            cambiada = True
        return cambiada

    def set_color(self, color):
        self.__color = color

    def encender(self):
        self.__encendido = True

    def apagar(self):
        self.__encendido = False

    def __str__(self):
        estado = "apagado"
        if self.__encendido:
            estado = "encendido"
        return ("Ventilador " + self.__color
                + ", radio " + str(self.__radio)
                + ", velocidad " + str(self.__velocidad)
                + ", " + estado)
````

### Enunciado

> Diseñe dos clases:

> * Un constructor que crea una cuenta con su titular y su saldo inicial.
> * Métodos para consultar el titular y el saldo.
> * Un método `depositar` que suma un valor al saldo.
> * Un método `retirar` que resta un valor del saldo, solo si el saldo alcanza; devuelve `True` o `False`.
> * Un método `transferir` que pasa un valor de esta cuenta a otra cuenta, solo si el saldo alcanza; devuelve `True` o `False`.

>

### La clase `Titular`

No tiene nada nuevo: atributos privados, constructor y métodos para consultar.

![UML de Titular](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_titular.png)

* **¿Y los setters?** El enunciado solo pide consultar: una cédula no cambia. No todo atributo necesita un *setter*.
* **¿Por qué la cédula es `str` y no `int`?** Porque no se hacen cuentas con ella y puede empezar con cero.
* **¿Por qué separar al titular de la cuenta?** Porque una persona puede tener varias cuentas: sus datos se escriben una sola vez.

````python
class Titular:
    def __init__(self, nombre, cedula):
        self.__nombre = nombre
        self.__cedula = cedula

    def get_nombre(self):
        return self.__nombre

    def get_cedula(self):
        return self.__cedula
````

### La clase `CuentaBancaria`

Lo nuevo: **un tipo puede ser una clase**. El atributo `titular` guarda un objeto `Titular` completo, y `transferir` recibe otra cuenta como parámetro (`destino : CuentaBancaria`). En el diagrama, los tipos que son clases van en negrita.

![UML de CuentaBancaria](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_cuenta.png)

* **¿Y `set_saldo`?** No existe: en un banco nadie escribe el saldo a mano. Solo cambia con `depositar`, `retirar` y `transferir`.
* **¿Qué devuelve `get_titular()`?** Un objeto `Titular`, no un texto. Para obtener el nombre: `cuenta.get_titular().get_nombre()`.
* **¿Por qué `depositar` no lleva tipo de retorno?** Porque no devuelve nada; `retirar` sí: avisa con `True` o `False` si el saldo alcanzó.

### Las dos clases relacionadas

![Asociación entre CuentaBancaria y Titular](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_cuenta_titular.png)

* **Asociación**: una línea entre dos clases que se relacionan. La palabra `titular`, junto a la flecha, es el nombre del atributo que guarda la relación.
* **La flecha** indica quién conoce a quién: la cuenta conoce a su titular (tiene el atributo `titular`); el titular no sabe qué cuentas tiene.
* **Multiplicidad**: cuántos objetos participan. `1` es exactamente uno; `1..*`, uno o más: una cuenta tiene un titular, y un titular puede tener una o más cuentas.

Diseñar diagramas UML para programas grandes se estudia en Ingeniería de Software. Aquí basta la idea: cuando un programa crece, sus clases se relacionan, y UML permite dibujar esas relaciones.

### La clase en Python

* `self.__titular` guarda el objeto `Titular` completo, no solo el nombre. En Python Tutor se ve como una flecha de la cuenta al titular: la misma flecha del diagrama UML.
* `__str__` escribe `self.__titular.get_nombre()` porque el nombre es privado de `Titular`: la cuenta se lo pide con su método, como lo haría cualquiera desde fuera.
* `transferir` es un método que **usa otros métodos**: `self.retirar(valor)` en esta cuenta y `destino.depositar(valor)` en la otra.

````python
class CuentaBancaria:
    def __init__(self, titular, saldo=0.0):
        self.__titular = titular
        self.__saldo = saldo

    def get_titular(self):
        return self.__titular

    def get_saldo(self):
        return self.__saldo

    def depositar(self, valor):
        self.__saldo = self.__saldo + valor

    def retirar(self, valor):
        retirado = False
        if valor <= self.__saldo:
            self.__saldo = self.__saldo - valor
            retirado = True
        return retirado

    def transferir(self, destino, valor):
        transferido = False
        if self.retirar(valor):
            destino.depositar(valor)
            transferido = True
        return transferido

    def __str__(self):
        return (self.__titular.get_nombre()
                + ": $" + str(self.__saldo))
````

## Parte 10 — Ejemplo 6: la clase `Banco`

### El sistema crece

El arreglo de cuentas que vivía suelto en el programa pasa a ser un **atributo** de una clase `Banco`.

![Banco, CuentaBancaria y Titular](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_banco.png)

* **Composición** (◆): el banco es dueño de sus cuentas. Las **crea**, y si el banco desaparece, sus cuentas también. Una cuenta pertenece a un solo banco. El rombo va del lado del todo (el banco), y `cuentas` es el nombre del atributo.
* El titular **no** es parte del banco: Homero sigue existiendo aunque el banco cierre. Por eso entre la cuenta y el titular hay una asociación, sin rombo.

### La clase `Banco`

![UML de Banco](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/images/uml_banco_clase.png)

`cuentas : CuentaBancaria[*]` es un atributo que guarda **muchas** cuentas: `[*]` es la misma multiplicidad `0..*` de la flecha del diagrama anterior, escrita dentro de la caja.

| Atributo | Tipo | Qué guarda |
|---|---|---|
| `nombre` | `str` | El nombre del banco |
| `cuentas` | `CuentaBancaria[*]` | El arreglo de cuentas, con tamaño `cupo` |
| `cantidad` | `int` | Cuántas cuentas hay abiertas (empieza en 0) |

| Método | Devuelve | Qué hace |
|---|---|---|
| `Banco(nombre, cupo)` | — | Crea el banco sin cuentas, con espacio para `cupo` cuentas |
| `get_nombre()` | `str` | Consulta el nombre |
| `get_cantidad()` | `int` | Consulta cuántas cuentas hay |
| `get_cuenta(posicion)` | `CuentaBancaria` | Consulta la cuenta de esa posición |
| `abrir_cuenta(titular, saldo)` | `bool` | Crea una cuenta en la siguiente posición libre; `False` si no hay cupo |
| `posicion_mayor_saldo(inicio)` | `int` | Busca la posición del mayor saldo desde `inicio` |
| `ordenar_por_saldo()` | — | Ordena las cuentas de mayor a menor saldo |
| `dar_bono(valor)` | — | Deposita `valor` en cada cuenta |

* **¿Por qué `abrir_cuenta` recibe un `Titular` y no una `CuentaBancaria`?** Porque el banco **crea** la cuenta (por eso es composición); el titular, en cambio, ya existe fuera del banco.
* **¿Por qué `posicion_mayor_saldo` tiene `inicio = 0`?** Para usar el mismo método en dos casos: buscar el mayor de todo el banco y ordenar.

### La clase en Python

Se reutiliza la función `intercambiar` de la clase 12, con su variable temporal `temp`.

* **¿Dónde quedó `cuentas = [None] * 3`?** Dentro del constructor: ahora es `self.__cuentas`, y el tamaño (`cupo`) lo decide quien crea el banco.
* **¿Por qué `len(self.__cuentas)` y no `cupo` en `abrir_cuenta`?** Porque `cupo` es un parámetro: deja de existir cuando el constructor termina. El arreglo sí recuerda su tamaño.
* **¿Por qué `range(..., self.__cantidad)` y no `len(...)` en los recorridos?** Porque solo se recorren las cuentas abiertas: las demás casillas siguen en `None`. Es el arreglo **parcialmente lleno**, ahora dentro de la clase.
* **¿Qué hace `self.posicion_mayor_saldo(i)`?** El banco usa uno de sus propios métodos: con `self.` se llama a otro método del mismo objeto.
* **¿Cuántos intercambios hay por cada paso?** Uno: se intercambia la cuenta completa, con su titular y su saldo.

````python
def intercambiar(lista, i, j):
    """Intercambia los elementos de las posiciones i y j (la misma función de ordenamientos.py)."""
    temp = lista[i]
    lista[i] = lista[j]
    lista[j] = temp


class Banco:
    def __init__(self, nombre, cupo):
        self.__nombre = nombre
        self.__cuentas = [None] * cupo
        self.__cantidad = 0

    def get_nombre(self):
        return self.__nombre

    def get_cantidad(self):
        return self.__cantidad

    def get_cuenta(self, posicion):
        return self.__cuentas[posicion]

    def abrir_cuenta(self, titular, saldo=0.0):
        abierta = False
        if self.__cantidad < len(self.__cuentas):
            self.__cuentas[self.__cantidad] = CuentaBancaria(titular, saldo)
            self.__cantidad = self.__cantidad + 1
            abierta = True
        return abierta

    def posicion_mayor_saldo(self, inicio=0):
        pos_mayor = inicio
        for j in range(inicio + 1, self.__cantidad):
            if self.__cuentas[j].get_saldo() > self.__cuentas[pos_mayor].get_saldo():
                pos_mayor = j
        return pos_mayor

    def ordenar_por_saldo(self):
        for i in range(self.__cantidad - 1):
            pos_mayor = self.posicion_mayor_saldo(i)
            intercambiar(self.__cuentas, i, pos_mayor)

    def dar_bono(self, valor):
        for i in range(self.__cantidad):
            self.__cuentas[i].depositar(valor)
````

### Pruebas

Las mismas pruebas del Ejemplo 5, y además lo nuevo: ordenar y el cupo del banco (5 cuentas).

| # | Qué se hace | Resultado esperado | Por qué |
|---|---|---|---|
| 1 | Se abren las cuentas de Homero (15000.0), Marge (15000.0) y el señor Burns (1000000.0) | 3 cuentas abiertas | Los mismos datos del Ejemplo 5 |
| 2 | Se busca la posición del mayor saldo | 2 (el señor Burns) | Igual que en el Ejemplo 5 |
| 3 | Se da un bono de 10000.0 a cada cuenta | Homero queda con 25000.0 | Igual que en el Ejemplo 5 |
| 4 | Se ordenan las cuentas por saldo | El señor Burns queda en la posición 0 | Nuevo: ordenar |
| 5 | Se abren las cuentas de Lisa y Bart | `True` las dos: el banco queda lleno (5 de 5) | Caso límite: se llena el cupo |
| 6 | Se intenta abrir la cuenta de Maggie | `False`: no hay cupo | Caso especial: arreglo lleno |

Cada `assert` corresponde a una fila de la tabla.

### Composición y agregación

¿Cómo saber qué rombo va? Mirando **quién crea el objeto**, en el método que lo guarda:

| Relación | Método | Qué hace | Si el todo desaparece |
|---|---|---|---|
| **Composición** (◆) | `abrir_cuenta` | Crea la cuenta adentro, con `CuentaBancaria(titular, saldo)` | Sus cuentas también desaparecen; una cuenta pertenece a un solo banco |
| **Agregación** (◇) | `agregar_estudiante` | Guarda un estudiante que ya fue creado afuera | Los estudiantes siguen existiendo, y un estudiante puede estar en varios cursos |

Los errores más peligrosos son los que **no dan error**: el programa sigue corriendo con un resultado equivocado.

### Olvidar `self` en un método

> ⚠️ **Error frecuente.** `calcular_area` no recibe `self`.

````python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area():              # falta self
        return self.base * self.altura


r1 = Rectangulo(3.0, 4.0)
print(r1.calcular_area())
````

### Usar un atributo sin `self.`

> ⚠️ **Error frecuente.** Dentro del método se escribe `base` en vez de `self.base`.

````python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return base * altura          # falta self.


r1 = Rectangulo(3.0, 4.0)
print(r1.calcular_area())
````

`base` sin `self.` es una variable local que no existe. **Corrección**: `return self.base * self.altura`.

### Olvidar los paréntesis al llamar un método

`print(r1.calcular_area)`, sin paréntesis, **no da error**: no ejecuta el método, sino que muestra el método mismo, algo como `<bound method Rectangulo.calcular_area of <__main__.Rectangulo object at 0x...>>`. **Corrección**: `print(r1.calcular_area())`.

### Comparar objetos con `==`

> ⚠️ **Error frecuente.** ¿Dos rectángulos con los mismos valores son "iguales"?

````python
class Rectangulo:
    def __init__(self, base, altura):
        self.__base = base
        self.__altura = altura

    def get_base(self):
        return self.__base

    def get_altura(self):
        return self.__altura

    def set_base(self, base):
        cambiada = False
        if base > 0:
            self.__base = base
            cambiada = True
        return cambiada

    def set_altura(self, altura):
        cambiada = False
        if altura > 0:
            self.__altura = altura
            cambiada = True
        return cambiada

    def calcular_area(self):
        return self.__base * self.__altura

    def calcular_perimetro(self):
        return 2 * self.__base + 2 * self.__altura


r1 = Rectangulo(3.0, 4.0)
r3 = Rectangulo(3.0, 4.0)
print(r1 == r3)
print(r1.get_base() == r3.get_base() and r1.get_altura() == r3.get_altura())
````

### Dos métodos con el mismo nombre

> ⚠️ **Error frecuente.** Se intenta escribir dos constructores, como en Java (*sobrecarga*).

````python
class Ventilador:
    def __init__(self):
        self.__velocidad = 1

    def __init__(self, velocidad):        # reemplaza al anterior
        self.__velocidad = velocidad


v = Ventilador()
````

En Python, el segundo `def` con el mismo nombre **reemplaza** al primero, así que el constructor sin parámetros ya no existe: "missing 1 required positional argument". **Corrección**: un solo método con valor por defecto, como el constructor del ventilador: `def __init__(self, velocidad=1):`.

### Cambiar un atributo privado desde afuera

`r1.__base = -5.0`, fuera de la clase, **no da error**: crea otro atributo aparte y la base verdadera no cambia (ver la Parte 5). **Corrección**: usar el *setter*, `r1.set_base(5.0)`.

| `r1 == r3` con los mismos valores | <span style="color:#B85450">**No da error**</span>: da `False` | Comparar los valores con los *getters* |
| Dos métodos con el mismo nombre | El segundo reemplaza al primero: `TypeError: ... missing 1 required positional argument` | Un solo método con valor por defecto |
| `r1.__base = -5.0`, fuera de la clase | <span style="color:#B85450">**No da error**</span>: crea otro atributo; la base verdadera no cambia | `r1.set_base(5.0)` |

## Resumen

La idea central: **los datos y lo que se hace con ellos viven juntos, en el objeto**.

| Concepto | En UML | En Python |
|---|---|---|
| Clase | Caja de tres partes: nombre, atributos, métodos | `class Rectangulo:` |
| Objeto | `r1 : Rectangulo` (subrayado) | `r1 = Rectangulo(3.0, 4.0)` |
| Constructor | `+ Rectangulo(base: float, altura: float)` | `def __init__(self, base, altura):` |
| Atributo privado | `- base : float` | `self.__base = base` |
| Método | `+ calcular_area() : float` | `def calcular_area(self):` |
| Método sin retorno | `+ encender()` (sin tipo) | Sin `return`: devuelve `None` |
| *Getter* / *setter* | `+ get_base() : float` / `+ set_base(base: float) : bool` | `def get_base(self):` / `def set_base(self, base):` |
| Constante de clase | `+ LENTO : int = 1` (subrayada) | `LENTO = 1` dentro de la clase |
| Asociación | `CuentaBancaria ──> Titular` | Un atributo que guarda otro objeto |
| Composición | `Banco ◆── CuentaBancaria` | El método crea el objeto: `CuentaBancaria(...)` |
| Agregación | `Curso ◇── Estudiante` | El método guarda un objeto ya creado |

* Cada clase va en su propia celda (o en su archivo), y sus pruebas con `assert` aparte.
* Los atributos son privados (`__`); desde fuera se usan los métodos. Los *setters* que validan devuelven `True` o `False`.
* `__str__` da el texto de `print(objeto)` y de `str(objeto)`.
* Un arreglo de objetos parcialmente lleno se crea con `[None] * cupo` y una `cantidad`, y se recorre hasta `cantidad`.

**Lo que sigue:** construir en Python el `Curso` del diagrama, con el mismo menú de siempre.
