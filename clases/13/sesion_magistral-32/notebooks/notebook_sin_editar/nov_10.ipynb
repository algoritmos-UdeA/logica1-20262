# Clase recangulo

```python
class Rectangulo:
    # ----------- Atributos ----------- 

    # Constructos
    def __init__(self, base = 1, altura = 1):
        # Atributos
        self.base = base 
        self.altura = altura

    # ----------- Metodos -----------
    def calcular_area(self):
        area = self.base*self.altura
        return area
        
    def calcular_perimetro(self):
        return 2*(self.base + self.altura)

    def __str__(self):
        str_out = f"- base: {self.base}" + "\n"
        str_out += f"- altura: {self.altura}" + "\n"
        return str_out
    
```

Creando objetos

```python
# Objetos
R1 = Rectangulo(3,2)
R2 = Rectangulo(1,2)
R3 = Rectangulo()

# Metodo
area_R1 = R1.calcular_area()
area_R2 = R2.calcular_area()
perimetro_R3 = R3.calcular_perimetro()

A = [1,2,3]
print(A)
print(R1)
print(R2)
```

Hacer pruebas

```python
r1 = Rectangulo(3.0, 4.0)
r2 = Rectangulo(5.0, 2.0)
assert r1.calcular_area() == 12.0
assert r2.calcular_area() == 10.0
assert r1.calcular_perimetro() == 14.0
assert r2.calcular_perimetro() == 14.0

r3 = Rectangulo(3.0, 4.0)
print(r1 is r3)
print("Rectangulo: todas las pruebas pasaron")
```

Rescribira clase con atributos privados

```python
class Rectangulo:
    # ----------- Atributos ----------- 

    # Constructos
    def __init__(self, base = 1, altura = 1):
        # Atributos
        self.__base = base 
        self.__altura = altura

    # ----------- Metodos -----------
    # Metodos accesores
    def set_base(self, base):
        self.__base = base

    def set_altura(self, altura):
        self.__altura = altura

    def get_base(self):
        return self.__base

    def get_altura(self):
        return self.__altura
        
    # Otros ---
    def calcular_area(self):
        area = self.__base*self.__altura
        return area
        
    def calcular_perimetro(self):
        return 2*(self.__base + self.__altura)

    def __str__(self):
        str_out = f"- base: {self.__base}" + "\n"
        str_out += f"- altura: {self.__altura}" + "\n"
        return str_out
    
```

```python
r5 = Rectangulo(4.5,2)
print(r5)
print(r5.get_base()) # print(r5.__base)
r5.set_base(8) # r5.__base = 8
r5.set_altura(6) # r5.__altura = 6
print(r5)
print(r5.calcular_area())
print(r5.get_altura())
print(r5.get_base())
```

## Aplicacion tipo UBER

```python
import math

class Punto:
    # ----------- Atributos ----------- 

    # Constructos
    def __init__(self, x = 0, y = 0):
        # Atributos
        self.__x = x 
        self.__y = y

    # ----------- Metodos -----------
    # Metodos accesores
    def set_x(self, x):
        self.__x = x

    def set_y(self, y):
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y
        
    def calcular_distancia(self, P):
        delta_x = P.__x - self.__x
        delta_y = P.__y - self.__y
        d = delta_x**2 + delta_y**2
        d = math.sqrt(d)
        return d

    def calcular_radio(self):
        r = math.sqrt(self.__x**2 + self.__y**2)
        return r
        
    def __str__(self):
        str_punto = f"({self.__x},{self.__y})" 
        return str_punto 
```

```python
A = Punto()
B = Punto(1,1)
C = Punto(-4,-3)
assert round(A.calcular_distancia(B),2) == 1.41
assert round(A.calcular_radio(),2) == 0.00
dBC = B.calcular_distancia(C)
assert round(dBC,2) == 6.40
print("Todas las pruebas OK")
```

Cliente

```python
class Cliente:
    # ----------- Atributos ----------- 

    # Constructos
    def __init__(self, cedula = None, nombre = None):
        # Atributos
        self.__cedula = cedula 
        self.__nombre = nombre
        self.__posicion = Punto(0,0)

    # ----------- Metodos -----------
    # Metodos accesores
    def set_cedula(self, cedula):
        self.__cedula = cedula

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def set_posicion(self, x, y):
        self.__posicion.set_x(x)
        self.__posicion.set_y(y)

    def get_cedula(self):
        return self.__cedula

    def get_nombre(self):
        return self.__nombre
        
    def get_posicion(self):
        return self.__posicion
```

```python
cliente1 = Cliente("666","juaja")
cliente1.set_posicion(2,3)
print(cliente1.get_cedula())
print(cliente1.get_nombre())
print(cliente1.get_posicion())
```
