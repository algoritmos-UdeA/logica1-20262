## Parte 0 — Repaso: una clase sola

Antes de relacionar varias clases conviene tener clara una sola. Este es el vocabulario que se usa en todo el notebook:

| Palabra | Qué es | En el ejemplo |
|---|---|---|
| **Clase** | El molde: describe qué características tienen y qué saben hacer todos los objetos de un tipo | `Rectangulo` |
| **Objeto** | Un ejemplar concreto, creado a partir de la clase | `r1`, un rectángulo de base 3.0 y altura 4.0 |
| **Atributos** | Sus características: los datos que cada objeto guarda | `base`, `altura` |
| **Métodos** | Lo que sabe hacer: funciones que se le piden al objeto con un punto | `r1.calcular_area()` |
| **Constructor** | El método `__init__`, que se ejecuta al crear el objeto y le da sus valores iniciales | `Rectangulo(3.0, 4.0)` |
| **`self`** | Dentro de la clase, el objeto al que se le pidió el método | en `r1.calcular_area()`, `self` es `r1` |
| **Encapsulamiento** | Los atributos se ocultan (en Python, con `__` al comienzo del nombre) y solo se llega a ellos con los métodos | `r1.get_base()`, `r1.set_base(5.0)` |

### Enunciado

> Diseñe una clase llamada `Rectangulo`. La clase contiene:
>
> * Dos atributos privados, `base` y `altura` (`float`).
> * Un constructor que crea un rectángulo con la base y la altura dadas.
> * Métodos para consultar y para cambiar cada atributo.
> * Un método que retorna el área y otro que retorna el perímetro.
> * Un método que devuelve una descripción del rectángulo en texto.
>
> La clase **supone** que la base y la altura son positivas. No lo verifica: es responsabilidad de quien la usa.

### Diagrama UML

![UML de Rectangulo](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/uml_rectangulo.png)

El diagrama se lee así:

* Arriba, el **nombre** de la clase; en el medio (azul), los **atributos**; abajo (amarillo), los **métodos**.
* `-` significa **privado** (solo se usa dentro de la clase) y `+` significa **público** (se puede usar desde afuera).
* Un atributo se escribe `nombre : tipo`. Un método se escribe `nombre(parámetros) : tipo de lo que retorna`.
* El **constructor** lleva el nombre de la clase y no tiene tipo de retorno.
* Los métodos que no retornan nada (como `set_base`) no llevan tipo.

### La clase

````python
class Rectangulo:
    def __init__(self, base, altura):
        self.__base = base
        self.__altura = altura

    def get_base(self):
        return self.__base

    def get_altura(self):
        return self.__altura

    def set_base(self, base):
        self.__base = base

    def set_altura(self, altura):
        self.__altura = altura

    def calcular_area(self):
        return self.__base * self.__altura

    def calcular_perimetro(self):
        return 2 * self.__base + 2 * self.__altura

    def __str__(self):
        return f"Rectangulo de base {self.__base} y altura {self.__altura}"
````

### Lo que ya se sabía de las listas

Hasta ahora la lista se usó como **arreglo**: un grupo de datos del mismo tipo, de tamaño fijo, al que se entra por posición.

| Lo que se hacía | Cómo |
|---|---|
| Crear un arreglo de `N` casillas | `notas = [0.0] * N` |
| Leer o cambiar una casilla | `notas[i]`, `notas[i] = 4.5` |
| Saber cuántas casillas tiene | `len(notas)` |
| Recorrerlo | `for i in range(N):` o `for nota in notas:` |

Con esas reglas el arreglo nunca cambiaba de tamaño. Era una decisión del curso para aprender a recorrer, buscar y ordenar con las propias manos. Pero la lista de Python sabe hacer mucho más, **porque es un objeto de la clase `list`**, y como todo objeto tiene métodos.

Tres cosas para notar:

* **No aparecen atributos.** La lista guarda sus datos en alguna parte, pero no se ven desde afuera ni hace falta saber cómo están guardados. Eso es el **encapsulamiento** visto desde el lado de quien usa la clase: basta conocer los métodos.
* `elemento` no tiene tipo porque una lista acepta datos de cualquier clase.
* Con cada método conviene hacerse dos preguntas: **¿modifica la lista?** y **¿qué retorna?**

| Método | ¿Modifica la lista? | ¿Qué retorna? |
|---|---|---|

| `sort()` | Sí | Nada (`None`) |
| `reverse()` | Sí | Nada (`None`) |
| `index(elemento)` | No | La posición (`int`) |
| `count(elemento)` | No | La cantidad (`int`) |

### Agregar elementos

Ya no hace falta saber el tamaño desde el comienzo: la lista se crea vacía y **crece**.

````python
nombres = []                      # una lista vacía; también sirve list()
print(nombres, len(nombres))

nombres.append("Tommy")           # agrega al final
nombres.append("CJ")
print(nombres, len(nombres))

nombres.insert(0, "Catalina")     # inserta en la posición 0; los demás se corren
print(nombres, len(nombres))

nombres.extend(["Franklin", "CJ"])   # agrega al final todos los de otra lista
print(nombres, len(nombres))
````

### Una variable no contiene al objeto: lo señala

Esta es la idea más importante para lo que sigue, y ya se había visto con los arreglos: `=` **no copia** una lista.

````python
a = ["Tommy", "CJ"]
b = a                    # b NO es una copia: es otro nombre para la misma lista
b.append("Catalina")

print(a)
print(b)
print(a is b)            # ¿a y b son el MISMO objeto?
````

## Parte 2 — El problema: una aplicación de transporte

Antes de escribir cualquier clase hace falta un **enunciado**: qué hace el programa, qué datos maneja, qué se le puede pedir y qué supone.

> Una empresa de transporte quiere un programa para administrar sus viajes dentro de una ciudad, que se representa como un plano: cada lugar es un **punto** con coordenadas `x` y `y`, medidas en kilómetros.
>
> La **plataforma** registra **clientes** y **conductores**. De cada cliente se conoce la cédula, el nombre y la posición donde se encuentra. De cada conductor se conoce la cédula, el nombre, la placa de su vehículo, la posición donde se encuentra y si está disponible. Todos empiezan en el punto (0, 0) y pueden cambiar de posición; un conductor recién registrado está disponible.
>
> Un cliente puede **solicitar un viaje** hacia un destino. La plataforma le asigna el conductor disponible **más cercano** al cliente; si no hay ninguno disponible, no hay viaje. De cada **viaje** se conoce el cliente, el conductor, el origen (la posición del cliente al solicitarlo) y el destino. La distancia del viaje es la que hay en línea recta entre el origen y el destino, y su valor es una tarifa base de \$4000 más \$1500 por cada kilómetro. Mientras dura el viaje, el conductor no está disponible.
>
> Al **finalizar el viaje**, el cliente y el conductor quedan en el destino y el conductor vuelve a estar disponible. Un cliente con un viaje sin finalizar no puede solicitar otro.
>
> La plataforma debe permitir consultar los clientes, los conductores y los viajes.
>
> **Se supone** que los datos que recibe el programa son correctos (cédulas que no se repiten, números donde se piden números). Las clases no lo verifican: es responsabilidad de quien las usa.

### Un caso para verificar a mano

Un buen enunciado permite calcular un ejemplo **antes** de programar. Ese ejemplo se usará después para probar.

Lucia está en (0, 0). Hay dos conductores: Tommy en (3, 4) y CJ en (1, 1). Lucia pide ir a (6, 8).

La distancia en línea recta entre dos puntos se calcula con el teorema de Pitágoras:

$$
d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}
$$

| Pregunta | Cálculo | Resultado |
|---|---|---|
| ¿A qué distancia está Tommy de Lucia? | $\sqrt{3^2 + 4^2} = \sqrt{25}$ | 5 km |
| ¿A qué distancia está CJ de Lucia? | $\sqrt{1^2 + 1^2} = \sqrt{2}$ | 1.41 km |
| ¿A quién le asignan? | El menor entre 5 y 1.41 | **CJ** |
| ¿Cuánto mide el viaje? | De (0, 0) a (6, 8): $\sqrt{6^2 + 8^2} = \sqrt{100}$ | **10 km** |
| ¿Cuánto vale? | 4000 + 1500 × 10 | **\$19 000** |
| ¿Dónde quedan al finalizar? | En el destino | Lucia y CJ en (6, 8); CJ disponible |

## Parte 3 — El diseño: del enunciado a las clases y sus relaciones

El enunciado no dice qué clases escribir. Para descubrirlas hay una técnica sencilla que sirve de punto de partida: **subrayar los sustantivos y los verbos** del enunciado.

* Los **sustantivos** (cosas) son candidatos a **clases** o a **atributos**.
* Los **verbos** (acciones) son candidatos a **métodos**.

### 1. Marcar el enunciado

En **negrita**, los sustantivos; en *cursiva*, los verbos.

> Una **empresa** de transporte quiere un **programa** para *administrar* sus **viajes** dentro de una **ciudad**, que se representa como un **plano**: cada lugar es un **punto** con **coordenadas** `x` y `y`, medidas en **kilómetros**.
>
> La **plataforma** *registra* **clientes** y **conductores**. De cada cliente se conoce la **cédula**, el **nombre** y la **posición** donde se encuentra. De cada conductor se conoce la cédula, el nombre, la **placa** de su vehículo, la posición donde se encuentra y si está **disponible**. Todos empiezan en el punto (0, 0) y pueden *cambiar de posición*.
>
> Un cliente puede *solicitar* un **viaje** hacia un **destino**. La plataforma le *asigna* el conductor disponible *más cercano* al cliente. De cada viaje se conoce el cliente, el conductor, el **origen** y el destino. La **distancia** del viaje *se calcula* entre el origen y el destino, y su **valor** *se calcula* con una **tarifa base** más un **valor por kilómetro**.
>
> Al *finalizar* el viaje, el cliente y el conductor quedan en el destino y el conductor vuelve a estar disponible.
>
> La plataforma debe permitir *consultar* los clientes, los conductores y los viajes.

### 2. Decidir qué es cada sustantivo

Para cada sustantivo se hacen tres preguntas, en orden:

1. **¿Tiene varios datos propios y algo que sabe hacer?** Entonces es una **clase**.
2. **¿Es un dato sencillo que describe a otra cosa?** Entonces es un **atributo** de esa cosa.
3. **¿Se puede calcular a partir de otros datos?** Entonces no se guarda: es un **método**.

| Sustantivo | Decisión | Por qué |
|---|---|---|
| empresa, programa | Se descarta | Es quien pide el programa y el programa mismo: no son datos |
| ciudad, plano | Se descarta | Solo es el espacio donde están los puntos; no tiene datos propios |
| kilómetros | Se descarta | Es la unidad de medida |
| **punto** | **Clase `Punto`** | Tiene dos datos (`x`, `y`) y sabe calcular la distancia a otro punto |
| coordenadas `x`, `y` | Atributos de `Punto` | Datos sencillos (`float`) que describen al punto |
| **cliente** | **Clase `Cliente`** | Tiene varios datos y sabe cambiar de posición |
| **conductor** | **Clase `Conductor`** | Tiene varios datos y sabe cambiar de posición, ocuparse y liberarse |
| cédula, nombre, placa | Atributos (`str`) | Datos sencillos que describen al cliente o al conductor |
| disponible | Atributo de `Conductor` (`bool`) | Un dato sencillo: sí o no |
| posición | Atributo de `Cliente` y de `Conductor`, **de tipo `Punto`** | Describe dónde está, pero no es un dato sencillo: es un punto |
| **viaje** | **Clase `Viaje`** | Tiene varios datos y sabe calcular su distancia y su valor, y finalizar |
| origen, destino | Atributos de `Viaje`, **de tipo `Punto`** | Son lugares, es decir, puntos |
| distancia, valor | **Métodos** de `Viaje` | No se guardan: se calculan con el origen, el destino y las tarifas |
| tarifa base, valor por kilómetro | Constantes de la clase `Viaje` | Son iguales para todos los viajes |
| **plataforma** | **Clase `Plataforma`** | Guarda los clientes, los conductores y los viajes, y sabe registrar, buscar y asignar |

Quedan **cinco clases**: `Punto`, `Cliente`, `Conductor`, `Viaje` y `Plataforma`.

La pista de que hay una **relación entre clases** está en los atributos cuyo tipo es otra clase: la *posición* de un cliente es un `Punto`; el *cliente* de un viaje es un `Cliente`.

### 3. Decidir de quién es cada verbo

Un método va en **la clase que tiene los datos necesarios para hacerlo**.

| Verbo del enunciado | Método | ¿En cuál clase? | Por qué ahí |
|---|---|---|---|
| calcular la distancia entre dos lugares | `calcular_distancia(otro)` | `Punto` | El punto tiene las coordenadas |
| cambiar de posición | `set_posicion(x, y)` | `Cliente` y `Conductor` | Cada uno tiene su posición |
| dejar de estar disponible, volver a estarlo | `ocupar()`, `liberar()` | `Conductor` | El conductor guarda si está disponible |
| calcular la distancia y el valor del viaje | `calcular_distancia()`, `calcular_valor()` | `Viaje` | El viaje tiene el origen, el destino y las tarifas |
| finalizar | `finalizar()` | `Viaje` | El viaje conoce al cliente, al conductor y el destino |
| registrar | `registrar_cliente(cliente)`, `registrar_conductor(conductor)` | `Plataforma` | La plataforma guarda las listas |
| asignar el más cercano | `conductor_mas_cercano(punto)` | `Plataforma` | Solo ella conoce a **todos** los conductores |
| solicitar un viaje | `solicitar_viaje(cedula, x_destino, y_destino)` | `Plataforma` | Hay que buscar al cliente y escoger entre todos los conductores |
| consultar | `mostrar_clientes()`, `mostrar_conductores()`, `mostrar_viajes()` | `Plataforma` | Ella guarda las listas |

> **¿Por qué "solicitar un viaje" no es un método del cliente, si es el cliente quien lo solicita?** Porque para atender la solicitud hay que revisar todos los conductores, y un cliente no los conoce. La plataforma sí. En el diseño, un verbo no va en la clase de quien lo hace en la vida real, sino en la clase que tiene los datos.

### 4. Decidir las relaciones

Cada vez que una clase tiene un atributo cuyo tipo es otra clase, hay una relación. Para cada una se responden **tres preguntas**:

1. **¿Quién conoce a quién?** La clase que tiene el atributo conoce a la otra, y no al revés.
2. **¿Quién crea el objeto?** Esto decide el tipo de relación.
3. **¿Quién le pide qué a quién?** Los métodos que una clase usa de la otra.

La segunda pregunta decide cómo se dibuja:

| Relación | Símbolo | Cuándo |
|---|---|---|
| **Composición** | Rombo negro | La clase **crea** el objeto que guarda |
| **Agregación** | Rombo blanco | La clase **recibe** el objeto ya creado y lo guarda en un grupo |
| **Asociación** | Flecha | La clase **recibe** el objeto ya creado y lo conoce para pedirle cosas |

El rombo se dibuja del lado de la clase que guarda al objeto.

| Relación | ¿Quién conoce a quién? | ¿Quién crea el objeto? | Tipo | ¿Qué le pide? |
|---|---|---|---|---|
| `Cliente` — `Punto` | El cliente conoce su punto | El cliente, en su constructor | Composición | `set_x`, `set_y` |
| `Conductor` — `Punto` | El conductor conoce su punto | El conductor, en su constructor | Composición | `set_x`, `set_y` |
| `Viaje` — `Punto` | El viaje conoce su origen y su destino | El viaje, en su constructor | Composición | `calcular_distancia`, `get_x`, `get_y` |
| `Viaje` — `Cliente` | El viaje conoce a su cliente | Nadie dentro del viaje: lo recibe | Asociación | `get_posicion`, `set_posicion`, `get_nombre` |
| `Viaje` — `Conductor` | El viaje conoce a su conductor | Nadie dentro del viaje: lo recibe | Asociación | `ocupar`, `liberar`, `set_posicion`, `get_nombre` |
| `Plataforma` — `Cliente` | La plataforma conoce a sus clientes | Los recibe ya creados | Agregación | `get_cedula`, `get_posicion` |
| `Plataforma` — `Conductor` | La plataforma conoce a sus conductores | Los recibe ya creados | Agregación | `get_cedula`, `esta_disponible`, `get_posicion` |
| `Plataforma` — `Viaje` | La plataforma conoce sus viajes | La plataforma, en `solicitar_viaje` | Composición | `esta_finalizado`, `get_cliente`, `finalizar` |

Hay que fijarse en la dirección: el viaje conoce a su cliente, pero **el cliente no conoce sus viajes**; el cliente conoce su punto, pero **el punto no sabe de quién es**. Por eso `Punto` es la clase más independiente de todas.

| 4 | `Viaje` | `Punto`, `Cliente`, `Conductor` | Un objeto que recibe a otros ya creados (asociación) |
| 5 | `Plataforma` | `Viaje` y, a través de ella, todas | Un objeto que guarda listas de objetos (agregación y composición) |
| 6 | Programa principal | `Plataforma`, `Cliente`, `Conductor` | La integración |

**¿Por qué en ese orden?** Si `Cliente` falla y `Punto` no estaba probada, no se sabe cuál de las dos tiene el error. Si `Punto` ya pasó sus pruebas, el error está en `Cliente`. Probar de abajo hacia arriba convierte un problema grande en varios pequeños.

> **Una advertencia sobre la técnica.** Subrayar sustantivos y verbos es un punto de partida, no una receta: hay que decidir, descartar y a veces devolverse. Durante la construcción aparecerá un método que el enunciado no mencionaba (`buscar_viaje_activo`), y es normal: el diseño se ajusta mientras se programa.

## Parte 4 — Paso 1: la clase `Punto`

![Mapa de clases: paso 1, Punto](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/mapa_paso1.png)

`Punto` no conoce a ninguna otra clase. Se escribe y se prueba sola.

### Enunciado

> Diseñe una clase llamada `Punto`, que representa un lugar de la ciudad. La clase contiene:
>
> * Dos atributos privados, `x` y `y` (`float`): las coordenadas en kilómetros.
> * Un constructor que crea un punto con las coordenadas dadas. Si no se dan, crea el punto (0.0, 0.0).
> * Métodos para consultar y para cambiar cada coordenada.
> * Un método `calcular_distancia` que recibe **otro punto** y retorna la distancia en línea recta entre los dos.
> * Un método que devuelve el punto en texto, por ejemplo `(3.0, 4.0)`.

### Diagrama UML

![UML de Punto](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/uml_punto.png)

`= 0.0` en el constructor indica el valor por defecto de ese parámetro.

### La clase

Los textos entre `"""` son la **especificación** de cada método, como en las funciones: *Supone* dice qué espera recibir y *Garantiza* dice qué hace.

````python
import math


class Punto:
    def __init__(self, x=0.0, y=0.0):
        """
        Garantiza: crea el punto (x, y). Si no se dan valores, crea el
        punto (0.0, 0.0).
        """
        self.__x = x
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def set_x(self, x):
        self.__x = x

    def set_y(self, y):
        self.__y = y

    def calcular_distancia(self, otro):
        """
        Supone: otro es un objeto Punto.
        Garantiza: retorna la distancia en línea recta entre este punto
        y el otro. No modifica ninguno de los dos.
        """
        dx = self.__x - otro.get_x()
        dy = self.__y - otro.get_y()
        return math.sqrt(dx ** 2 + dy ** 2)

    def __str__(self):
        return f"({self.__x}, {self.__y})"
````

Hay un detalle nuevo en `calcular_distancia(self, otro)`: el método trabaja con **dos objetos de la misma clase**.

* `self` es el punto al que se le pidió el método. Como el código está dentro de la clase, usa sus atributos directamente: `self.__x`.
* `otro` es el punto que llegó como parámetro. Sus coordenadas se piden con sus métodos: `otro.get_x()`.

En `a.calcular_distancia(b)`, `self` es `a` y `otro` es `b`.

### Pruebas

## Parte 5 — Paso 2: la clase `Cliente`, que tiene un `Punto`

![Mapa de clases: paso 2, Cliente](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/mapa_paso2.png)

Aquí aparece la primera relación: un objeto que **guarda a otro objeto**.

### Enunciado

> Diseñe una clase llamada `Cliente`. La clase contiene:
>
> * Tres atributos privados: `cedula` (`str`), `nombre` (`str`) y `posicion`, que es un objeto de la clase `Punto`.
> * Un constructor que recibe la cédula y el nombre, y crea al cliente ubicado en el punto (0.0, 0.0).
> * Métodos para consultar la cédula, el nombre y la posición.
> * Un método `set_posicion` que recibe las coordenadas `x` y `y` y deja al cliente en ese lugar.
> * Un método que devuelve al cliente en texto, por ejemplo `1001 Lucia en (0.0, 0.0)`.

### Diagrama UML

![UML de Cliente](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/uml_cliente.png)

### Las tres preguntas

| Pregunta | Respuesta |
|---|---|
| ¿Quién conoce a quién? | El cliente conoce su punto. El punto no sabe que pertenece a un cliente |
| ¿Quién crea el objeto? | El cliente, en su constructor: `self.__posicion = Punto(0.0, 0.0)`. Es **composición** |
| ¿Quién le pide qué a quién? | El cliente le pide a su punto que cambie (`set_x`, `set_y`) y que se convierta en texto |

### La clase

````python
class Cliente:
    def __init__(self, cedula, nombre):
        """
        Garantiza: crea un cliente ubicado en el punto (0.0, 0.0).
        """
        self.__cedula = cedula
        self.__nombre = nombre
        self.__posicion = Punto(0.0, 0.0)

    def get_cedula(self):
        return self.__cedula

    def get_nombre(self):
        return self.__nombre

    def get_posicion(self):
        """Garantiza: retorna el objeto Punto donde está el cliente."""
        return self.__posicion

    def set_posicion(self, x, y):
        """
        Garantiza: el cliente queda en (x, y). No crea un punto nuevo:
        le pide a su punto que cambie sus coordenadas.
        """
        self.__posicion.set_x(x)
        self.__posicion.set_y(y)

    def __str__(self):
        return f"{self.__cedula} {self.__nombre} en {self.__posicion}"
````

Tres líneas merecen atención, porque en ellas el cliente **le pide cosas a su punto**:

| Línea | Qué pasa |
|---|---|

## Parte 6 — Paso 3: la clase `Conductor`

![Mapa de clases: paso 3, Conductor](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/mapa_paso3.png)

El conductor también tiene una posición. No hay que inventar nada: **se usa otra vez la clase `Punto`**, que ya está probada. Esa es una de las ventajas de separar el problema en clases.

### Enunciado

> Diseñe una clase llamada `Conductor`. La clase contiene:
>
> * Cinco atributos privados: `cedula`, `nombre` y `placa` (`str`); `posicion`, que es un objeto de la clase `Punto`; y `disponible` (`bool`).
> * Un constructor que recibe la cédula, el nombre y la placa, y crea al conductor disponible y ubicado en (0.0, 0.0).
> * Métodos para consultar cada atributo.
> * Un método `set_posicion` que recibe las coordenadas `x` y `y` y deja al conductor en ese lugar.
> * Un método `ocupar`, que deja al conductor como no disponible, y un método `liberar`, que lo deja disponible.
> * Un método que devuelve al conductor en texto, por ejemplo `2002 CJ (GRV992) en (1.0, 1.0) disponible`.

### Diagrama UML

![UML de Conductor](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/uml_conductor.png)

### Las tres preguntas

| Pregunta | Respuesta |
|---|---|
| ¿Quién conoce a quién? | El conductor conoce su punto |
| ¿Quién crea el objeto? | El conductor, en su constructor. Es **composición**, igual que en `Cliente` |
| ¿Quién le pide qué a quién? | El conductor le pide a su punto que cambie y que se convierta en texto |

### La clase

````python
class Conductor:
    def __init__(self, cedula, nombre, placa):
        """
        Garantiza: crea un conductor disponible, ubicado en (0.0, 0.0).
        """
        self.__cedula = cedula
        self.__nombre = nombre
        self.__placa = placa
        self.__posicion = Punto(0.0, 0.0)
        self.__disponible = True

    def get_cedula(self):
        return self.__cedula

    def get_nombre(self):
        return self.__nombre

    def get_placa(self):
        return self.__placa

    def get_posicion(self):
        """Garantiza: retorna el objeto Punto donde está el conductor."""
        return self.__posicion

    def set_posicion(self, x, y):
        """Garantiza: el conductor queda en (x, y)."""
        self.__posicion.set_x(x)
        self.__posicion.set_y(y)

    def esta_disponible(self):
        return self.__disponible

    def ocupar(self):
        """Garantiza: el conductor deja de estar disponible."""
        self.__disponible = False

    def liberar(self):
        """Garantiza: el conductor vuelve a estar disponible."""
        self.__disponible = True

    def __str__(self):
        estado = "ocupado"
        if self.__disponible:
            estado = "disponible"
        return (f"{self.__cedula} {self.__nombre} ({self.__placa}) "
                f"en {self.__posicion} {estado}")
````

## Parte 7 — Paso 4: la clase `Viaje`, que une a un cliente con un conductor

![Mapa de clases: paso 4, Viaje](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/mapa_paso4.png)

Hasta aquí cada clase creaba el objeto que guardaba. `Viaje` trae lo nuevo: **recibe objetos que ya existían**.

### Enunciado

> Diseñe una clase llamada `Viaje`. La clase contiene:
>
> * Dos constantes: `TARIFA_BASE`, con el valor 4000, y `VALOR_POR_KM`, con el valor 1500.
> * Cinco atributos privados: `cliente` (un objeto `Cliente`), `conductor` (un objeto `Conductor`), `origen` y `destino` (objetos `Punto`) y `finalizado` (`bool`).
> * Un constructor que recibe el cliente, el conductor y las coordenadas del destino. El origen es la posición que tiene el cliente en ese momento. El viaje empieza sin finalizar, y el conductor queda ocupado.
> * Métodos para consultar cada atributo.
> * Un método `calcular_distancia`, que retorna los kilómetros entre el origen y el destino.
> * Un método `calcular_valor`, que retorna la tarifa base más el valor por kilómetro multiplicado por la distancia.
> * Un método `finalizar`, que deja al cliente y al conductor en el destino, libera al conductor y marca el viaje como finalizado.
> * Un método que devuelve el viaje en texto.
>
> La clase **supone** que el conductor que recibe está disponible y que `finalizar` se pide una sola vez.

### Diagrama UML

![UML de Viaje](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/uml_viaje.png)

### Las tres preguntas

En esta clase hay **dos tipos de relación**, y la diferencia está en la segunda pregunta.

| Pregunta | Con `Cliente` y `Conductor` | Con `Punto` (origen y destino) |
|---|---|---|
| ¿Quién conoce a quién? | El viaje conoce a su cliente y a su conductor. Ellos no conocen al viaje | El viaje conoce su origen y su destino |
| ¿Quién crea el objeto? | **Nadie dentro del viaje**: llegan ya creados como parámetros del constructor. Es **asociación** | **El viaje**, en su constructor. Es **composición** |
| ¿Quién le pide qué a quién? | Al cliente: su posición y que se mueva. Al conductor: que se ocupe, que se libere y que se mueva | Al origen: la distancia hasta el destino |

### La clase

````python
class Viaje:
    TARIFA_BASE = 4000
    VALOR_POR_KM = 1500

    def __init__(self, cliente, conductor, x_destino, y_destino):
        """
        Supone: cliente es un objeto Cliente y conductor es un objeto
        Conductor que está disponible.
        Garantiza: crea un viaje sin finalizar, desde la posición actual
        del cliente hasta (x_destino, y_destino), y ocupa al conductor.
        """
        self.__cliente = cliente
        self.__conductor = conductor
        posicion = cliente.get_posicion()
        self.__origen = Punto(posicion.get_x(), posicion.get_y())
        self.__destino = Punto(x_destino, y_destino)
        self.__finalizado = False
        conductor.ocupar()

    def get_cliente(self):
        return self.__cliente

    def get_conductor(self):
        return self.__conductor

    def get_origen(self):
        return self.__origen

    def get_destino(self):
        return self.__destino

    def esta_finalizado(self):
        return self.__finalizado

    def calcular_distancia(self):
        """Garantiza: retorna los kilómetros entre el origen y el destino."""
        return self.__origen.calcular_distancia(self.__destino)

    def calcular_valor(self):
        """
        Garantiza: retorna el valor del viaje, que es la tarifa base más
        el valor por kilómetro multiplicado por la distancia.
        """
        return Viaje.TARIFA_BASE + Viaje.VALOR_POR_KM * self.calcular_distancia()

    def finalizar(self):
        """
        Supone: el viaje no ha finalizado.
        Garantiza: el cliente y el conductor quedan en el destino, el
        conductor vuelve a estar disponible y el viaje queda finalizado.
        """
        x = self.__destino.get_x()
        y = self.__destino.get_y()
        self.__cliente.set_posicion(x, y)
        self.__conductor.set_posicion(x, y)
        self.__conductor.liberar()
        self.__finalizado = True

    def __str__(self):
        estado = "en curso"
        if self.__finalizado:
            estado = "finalizado"
        return (f"{self.__cliente.get_nombre()} con "
                f"{self.__conductor.get_nombre()}: "
                f"{self.__origen} -> {self.__destino} | "
                f"{self.calcular_distancia():.2f} km | "
                f"${self.calcular_valor():.0f} | {estado}")
````

El constructor merece leerse línea por línea:

| Línea | Qué pasa |
|---|---|
| `self.__cliente = cliente` | **No se crea nada**: se guarda la referencia al cliente que llegó. El viaje y quien lo creó señalan al mismo objeto |
| `self.__conductor = conductor` | Lo mismo con el conductor |
| `posicion = cliente.get_posicion()` | El viaje le pide al cliente su punto |
| `self.__origen = Punto(posicion.get_x(), posicion.get_y())` | El viaje **crea un punto nuevo** con las mismas coordenadas. Más abajo se explica por qué no guarda el del cliente |
| `self.__destino = Punto(x_destino, y_destino)` | Crea otro punto nuevo para el destino |
| `conductor.ocupar()` | El viaje le pide al conductor que se ocupe |

Y `calcular_distancia` muestra otra vez la delegación: el viaje **no sabe** calcular distancias. Le pide a su origen que calcule la distancia hasta su destino, y `Punto` ya sabe hacerlo.

Las constantes se escriben dentro de la clase y fuera de los métodos, y se usan con el nombre de la clase: `Viaje.TARIFA_BASE`.

## Parte 8 — Paso 5: la clase `Plataforma`, que guarda listas de objetos

![Mapa de clases: paso 5, Plataforma](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/mapa_paso5.png)

Falta la clase que reúne todo. La plataforma guarda **grupos** de objetos, y para eso usa lo visto en la parte 1: listas con `append`.

### Enunciado

> Diseñe una clase llamada `Plataforma`. La clase contiene:
>
> * Tres atributos privados: `clientes`, `conductores` y `viajes`, que son listas de objetos `Cliente`, `Conductor` y `Viaje`.
> * Un constructor que crea la plataforma con las tres listas vacías.
> * Métodos `registrar_cliente` y `registrar_conductor`, que reciben un objeto ya creado y lo agregan a su lista.
> * Métodos `buscar_cliente` y `buscar_conductor`, que reciben una cédula y retornan el objeto, o `None` si no está.
> * Un método `conductor_mas_cercano`, que recibe un punto y retorna el conductor disponible más cercano a él, o `None` si no hay ninguno disponible.
> * Un método `buscar_viaje_activo`, que recibe la cédula de un cliente y retorna su viaje sin finalizar, o `None` si no tiene.
> * Un método `solicitar_viaje`, que recibe la cédula del cliente y las coordenadas del destino. Si el cliente está registrado, no tiene un viaje sin finalizar y hay un conductor disponible, crea el viaje con el conductor más cercano, lo guarda y lo retorna. Si no, retorna `None`.
> * Un método `finalizar_viaje`, que recibe la cédula del cliente, finaliza su viaje y retorna `True`; si el cliente no tiene un viaje sin finalizar, retorna `False`.
> * Métodos para mostrar los clientes, los conductores y los viajes.
>
> La clase **supone** que no se registran dos clientes ni dos conductores con la misma cédula.

`None` es el valor que Python usa para decir "no hay objeto". Se compara con `==` y `!=`.

### Diagrama UML

![UML de Plataforma](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/uml_plataforma.png)

### Las tres preguntas

| Pregunta | Con `Cliente` y `Conductor` | Con `Viaje` |
|---|---|---|
| ¿Quién conoce a quién? | La plataforma conoce a todos sus clientes y conductores | La plataforma conoce todos sus viajes |
| ¿Quién crea el objeto? | **Otro** (el programa principal): llegan ya creados a `registrar_cliente` y `registrar_conductor`. Es **agregación** | **La plataforma**, dentro de `solicitar_viaje`. Es **composición** |
| ¿Quién le pide qué a quién? | La cédula, para buscarlos; al conductor, si está disponible y dónde está | Si está finalizado, quién es su cliente, y que finalice |

### La clase

````python
class Plataforma:
    def __init__(self):
        """Garantiza: crea una plataforma sin clientes, conductores ni viajes."""
        self.__clientes = []
        self.__conductores = []
        self.__viajes = []

    def registrar_cliente(self, cliente):
        """
        Supone: cliente es un objeto Cliente y su cédula no está registrada.
        Garantiza: el cliente queda al final de la lista de clientes.
        """
        self.__clientes.append(cliente)

    def registrar_conductor(self, conductor):
        """
        Supone: conductor es un objeto Conductor y su cédula no está
        registrada.
        Garantiza: el conductor queda al final de la lista de conductores.
        """
        self.__conductores.append(conductor)

    def buscar_cliente(self, cedula):
        """
        Garantiza: retorna el cliente con esa cédula, o None si no está
        registrado.
        """
        encontrado = None
        for cliente in self.__clientes:
            if cliente.get_cedula() == cedula:
                encontrado = cliente
                break
        return encontrado

    def buscar_conductor(self, cedula):
        """
        Garantiza: retorna el conductor con esa cédula, o None si no está
        registrado.
        """
        encontrado = None
        for conductor in self.__conductores:
            if conductor.get_cedula() == cedula:
                encontrado = conductor
                break
        return encontrado

    def conductor_mas_cercano(self, punto):
        """
        Supone: punto es un objeto Punto.
        Garantiza: retorna el conductor disponible que está a menor
        distancia del punto, o None si no hay ninguno disponible. Si dos
        están a la misma distancia, retorna el que se registró primero.
        """
        mas_cercano = None
        menor_distancia = 0.0
        for conductor in self.__conductores:
            if conductor.esta_disponible():
                distancia = conductor.get_posicion().calcular_distancia(punto)
                if mas_cercano == None or distancia < menor_distancia:
                    mas_cercano = conductor
                    menor_distancia = distancia
        return mas_cercano

    def buscar_viaje_activo(self, cedula):
        """
        Garantiza: retorna el viaje sin finalizar del cliente con esa
        cédula, o None si no tiene.
        """
        encontrado = None
        for viaje in self.__viajes:
            if (not viaje.esta_finalizado()
                    and viaje.get_cliente().get_cedula() == cedula):
                encontrado = viaje
                break
        return encontrado

    def solicitar_viaje(self, cedula, x_destino, y_destino):
        """
        Garantiza: si el cliente está registrado, no tiene un viaje sin
        finalizar y hay un conductor disponible, crea el viaje con el
        conductor disponible más cercano al cliente, lo guarda y lo
        retorna. Si no, no cambia nada y retorna None.
        """
        viaje = None
        cliente = self.buscar_cliente(cedula)
        if cliente != None and self.buscar_viaje_activo(cedula) == None:
            conductor = self.conductor_mas_cercano(cliente.get_posicion())
            if conductor != None:
                viaje = Viaje(cliente, conductor, x_destino, y_destino)
                self.__viajes.append(viaje)
        return viaje

    def finalizar_viaje(self, cedula):
        """
        Garantiza: si el cliente con esa cédula tiene un viaje sin
        finalizar, lo finaliza y retorna True. Si no, no cambia nada y
        retorna False.
        """
        finalizado = False
        viaje = self.buscar_viaje_activo(cedula)
        if viaje != None:
            viaje.finalizar()
            finalizado = True
        return finalizado

    def mostrar_clientes(self):
        """Garantiza: imprime los clientes, uno por línea."""
        if len(self.__clientes) == 0:
            print("(no hay clientes)")
        for cliente in self.__clientes:
            print(cliente)

    def mostrar_conductores(self):
        """Garantiza: imprime los conductores, uno por línea."""
        if len(self.__conductores) == 0:
            print("(no hay conductores)")
        for conductor in self.__conductores:
            print(conductor)

    def mostrar_viajes(self):
        """Garantiza: imprime los viajes, uno por línea."""
        if len(self.__viajes) == 0:
            print("(no hay viajes)")
        for viaje in self.__viajes:
            print(viaje)
````

La clase es larga, pero cada método usa un patrón que ya se conoce.

**Registrar** es `append`: la lista crece con cada objeto que llega.

**Buscar por cédula** es la búsqueda lineal de la clase 12, sobre una lista de objetos. Lo que cambia es que la comparación se hace con un método del objeto (`cliente.get_cedula() == cedula`) y que se retorna **el objeto** en vez de su posición.

**Hallar el conductor más cercano** es la búsqueda del menor, con dos variables: `mas_cercano` guarda el mejor conductor visto hasta ahora y `menor_distancia` guarda a qué distancia está. `mas_cercano` empieza en `None`, y eso sirve para dos cosas: reconocer al primer conductor disponible (`mas_cercano == None`) y responder `None` si no había ninguno.

Con Tommy en (3, 4), CJ en (1, 1) y el punto (0, 0), el recorrido va así:

| Conductor | ¿Disponible? | Distancia | ¿Es menor que la que se tenía? | `mas_cercano` | `menor_distancia` |
|---|---|---|---|---|---|
| (antes de empezar) | | | | `None` | 0.0 |
| Tommy | Sí | 5.0 | Es el primero | Tommy | 5.0 |
| CJ | Sí | 1.41 | Sí: 1.41 < 5.0 | CJ | 1.41 |

**Solicitar un viaje** junta todo: busca al cliente, revisa que no tenga un viaje sin finalizar, busca al conductor más cercano y, solo si todo se cumple, **crea** el `Viaje` y lo guarda con `append`.

### Uso

El programa crea los clientes y los conductores, y se los **entrega** a la plataforma.

````python
plataforma = Plataforma()

lucia = Cliente("1001", "Lucia")
kendl = Cliente("1003", "Kendl")
kendl.set_posicion(10.0, 10.0)
plataforma.registrar_cliente(lucia)
plataforma.registrar_cliente(kendl)

tommy = Conductor("2001", "Tommy", "VCT986")
tommy.set_posicion(3.0, 4.0)
cj = Conductor("2002", "CJ", "GRV992")
cj.set_posicion(1.0, 1.0)
plataforma.registrar_conductor(tommy)
plataforma.registrar_conductor(cj)

plataforma.mostrar_clientes()
plataforma.mostrar_conductores()
````

### Pruebas

La prueba de `Plataforma` ejercita **las cinco clases a la vez**. Ya no prueba una pieza: prueba que las piezas encajan.

| Caso | Resultado esperado |
|---|---|
| Buscar una cédula registrada y una que no | El mismo objeto (`is`); `None` |
| Más cercano a (0, 0) y a (3, 3) | CJ; Tommy |
| Más cercano cuando no hay conductores | `None` |
| Lucia solicita ir a (6, 8) | Viaje con CJ por \$19 000; CJ queda ocupado |
| Lucia solicita otro sin finalizar | `None` |
| Solicita alguien que no está registrado | `None` |
| Kendl solicita con CJ ocupado | Le asignan a Tommy, el único disponible |
| Finalizar el viaje de Lucia, y volver a intentarlo | `True`; después `False` |
| Solicitar sin conductores disponibles | `None` |

````python
plataforma = Plataforma()

# Sin conductores no hay a quién asignar
assert plataforma.conductor_mas_cercano(Punto(0.0, 0.0)) == None

# Registrar y buscar
lucia = Cliente("1001", "Lucia")
kendl = Cliente("1003", "Kendl")
kendl.set_posicion(10.0, 10.0)
plataforma.registrar_cliente(lucia)
plataforma.registrar_cliente(kendl)

tommy = Conductor("2001", "Tommy", "VCT986")
tommy.set_posicion(3.0, 4.0)
cj = Conductor("2002", "CJ", "GRV992")
cj.set_posicion(1.0, 1.0)
plataforma.registrar_conductor(tommy)
plataforma.registrar_conductor(cj)

assert plataforma.buscar_cliente("1001") is lucia
assert plataforma.buscar_cliente("9999") == None
assert plataforma.buscar_conductor("2002") is cj
assert plataforma.buscar_conductor("9999") == None

# El conductor más cercano a (0, 0) es CJ (1.41 km); Tommy está a 5 km
assert plataforma.conductor_mas_cercano(Punto(0.0, 0.0)) is cj
# El más cercano a (3, 3) es Tommy (1 km)
assert plataforma.conductor_mas_cercano(Punto(3.0, 3.0)) is tommy

# Solicitar un viaje: Lucia va a (6, 8) con CJ
viaje = plataforma.solicitar_viaje("1001", 6.0, 8.0)
assert viaje != None
assert viaje.get_conductor() is cj
assert viaje.calcular_valor() == 19000.0
assert cj.esta_disponible() == False
assert plataforma.buscar_viaje_activo("1001") is viaje

# Un cliente con un viaje sin finalizar no puede solicitar otro
assert plataforma.solicitar_viaje("1001", 1.0, 1.0) == None

# Un cliente que no está registrado no puede solicitar
assert plataforma.solicitar_viaje("9999", 1.0, 1.0) == None

# CJ está ocupado: a Kendl le asignan a Tommy, el único disponible
viaje_kendl = plataforma.solicitar_viaje("1003", 13.0, 14.0)
assert viaje_kendl.get_conductor() is tommy
assert viaje_kendl.calcular_distancia() == 5.0

# Finalizar el viaje de Lucia
assert plataforma.finalizar_viaje("1001") == True
assert viaje.esta_finalizado() == True
assert cj.esta_disponible() == True
assert str(lucia.get_posicion()) == "(6.0, 8.0)"
assert plataforma.buscar_viaje_activo("1001") == None

# No se puede finalizar dos veces, ni el viaje de quien no tiene
assert plataforma.finalizar_viaje("1001") == False
assert plataforma.finalizar_viaje("9999") == False

# Sin conductores disponibles no hay viaje
cj.ocupar()
assert plataforma.solicitar_viaje("1001", 0.0, 0.0) == None

print("Pruebas de plataforma: todas pasaron")
````

## Parte 9 — La integración: el programa principal

![Mapa de clases: todas probadas, y el programa principal](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/mapa_integracion.png)

Las cinco clases están escritas y probadas. Falta el programa que las usa.

Hay dos papeles distintos. Quien **escribió** las clases conoce sus atributos y sus detalles. El **programa principal** es su cliente: solo las usa, a través de sus métodos públicos. En este problema hace tres cosas:

1. **Crea** la plataforma, los clientes y los conductores.
2. **Pide**: le encarga todo a la plataforma.
3. **Muestra** lo que la plataforma responde.

| El programa principal sí | El programa principal no |
|---|---|
| Crea objetos `Cliente` y `Conductor` | Crea objetos `Punto` ni `Viaje`: de eso se encargan las otras clases |
| Le pide a la plataforma `solicitar_viaje(...)` | Recorre los conductores ni calcula distancias |
| Revisa si la respuesta fue `None` | Toca atributos privados de ninguna clase |

````python
def main():
    # 1. Crear
    plataforma = Plataforma()

    lucia = Cliente("1001", "Lucia")
    kendl = Cliente("1003", "Kendl")
    kendl.set_posicion(10.0, 10.0)
    plataforma.registrar_cliente(lucia)
    plataforma.registrar_cliente(kendl)

    tommy = Conductor("2001", "Tommy", "VCT986")
    tommy.set_posicion(3.0, 4.0)
    cj = Conductor("2002", "CJ", "GRV992")
    cj.set_posicion(1.0, 1.0)
    catalina = Conductor("2003", "Catalina", "LBC008")
    catalina.set_posicion(9.0, 12.0)
    plataforma.registrar_conductor(tommy)
    plataforma.registrar_conductor(cj)
    plataforma.registrar_conductor(catalina)

    print("--- Conductores al comenzar ---")
    plataforma.mostrar_conductores()

    # 2. Pedir   3. Mostrar
    print("--- Lucia pide un viaje a (6, 8) ---")
    viaje = plataforma.solicitar_viaje("1001", 6.0, 8.0)
    if viaje == None:
        print("No se pudo crear el viaje")
    else:
        print(viaje)

    print("--- Kendl pide un viaje a (13, 14) ---")
    viaje = plataforma.solicitar_viaje("1003", 13.0, 14.0)
    if viaje == None:
        print("No se pudo crear el viaje")
    else:
        print(viaje)

    print("--- Finaliza el viaje de Lucia ---")
    plataforma.finalizar_viaje("1001")
    plataforma.mostrar_viajes()

    print("--- Conductores al terminar ---")
    plataforma.mostrar_conductores()

    print("--- Clientes al terminar ---")
    plataforma.mostrar_clientes()


main()
````

La salida coincide con lo calculado a mano en la parte 2: a Lucia le asignan a CJ, el viaje mide 10 km y vale \$19 000, y al final los dos están en (6.0, 8.0). A Kendl le asignan a Catalina, que está a 2.24 km de ella (Tommy está a 9.22 km y CJ ya estaba ocupado).

### Una petición, de punta a punta

El programa principal escribió **una línea**: `plataforma.solicitar_viaje("1001", 6.0, 8.0)`. Detrás de ella colaboraron las cinco clases. Este es el recorrido:

![Quién le pide qué a quién al solicitar un viaje](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/13/teoria/app_transporte/images/peticion_solicitar_viaje.png)

| # | Quién pide | A quién | Qué le pide | Para qué |
|---|---|---|---|---|
| 1 | Programa principal | `Plataforma` | `solicitar_viaje("1001", 6.0, 8.0)` | Es lo único que el programa sabe pedir |
| 2 | `Plataforma` | cada `Cliente` | `get_cedula()` y, al encontrarlo, `get_posicion()` | Hallar a Lucia y saber dónde está |
| 3 | `Plataforma` | cada `Conductor` | `esta_disponible()` y `get_posicion()` | Saber cuáles pueden ir y dónde están |
| 4 | `Plataforma` | el `Punto` de cada conductor | `calcular_distancia(punto de Lucia)` | Comparar y quedarse con el más cercano |
| 5 | `Plataforma` | la clase `Viaje` | `Viaje(lucia, cj, 6.0, 8.0)` | Crear el viaje |
| 6 | `Viaje` | `Cliente` | `get_posicion()` | Saber dónde empieza el viaje |
| 7 | `Viaje` | la clase `Punto` | `Punto(...)`, dos veces | Crear su origen y su destino |
| 8 | `Viaje` | `Conductor` | `ocupar()` | Que CJ deje de estar disponible |

Ninguna clase hace todo el trabajo. Cada una hace una parte pequeña y le pide el resto a quien sabe hacerlo. **Eso es un programa orientado a objetos: objetos que se piden cosas entre sí.**

Para ver este recorrido en movimiento sirve el enlace de Python Tutor de la celda de la parte 8 donde Lucia solicita su viaje: muestra cada llamada y cada flecha entre objetos.

| El programa principal | Un escenario fijo | `main.py`, con un menú que le pregunta al usuario |
| El código de las clases | El mismo | El mismo |

Al separar en archivos aparece un detalle que confirma el diseño: **un archivo importa a otro solo si crea objetos de esa clase**. `cliente.py` importa `punto.py` (composición), pero `plataforma.py` no importa `cliente.py` (agregación: los recibe ya creados).

### Resumen

| Mensaje | Qué significa | Cómo se corrige |
|---|---|---|
| `AttributeError: 'NoneType' object has no attribute ...` | Se le pidió un método a `None`: una búsqueda no encontró nada | Revisar `if objeto == None:` antes de usarlo |
| `AttributeError: 'str' object has no attribute ...` | Se entregó un dato (texto, número) donde se esperaba un objeto | Entregar el objeto, no uno de sus datos |
| `AttributeError: 'Cliente' object has no attribute ...` | Ese método es de otra clase | Preguntar de qué clase es el objeto y encadenar: `cliente.get_posicion().get_x()` |
| `TypeError: ... missing 1 required positional argument: 'self'` | Se le pidió el método a la clase | Pedírselo a un objeto: `lucia.get_nombre()` |
| `NameError: name '...' is not defined` (dentro de un método) | Falta `self.` | Escribir `self.__atributo` |
| Ningún mensaje, pero un objeto cambia "solo" | Dos variables señalan al mismo objeto (partes 5 y 7) | Decidir si se quería la referencia o un objeto nuevo |

El último es el más difícil, porque no avisa. Para encontrarlo sirve dibujar los objetos y sus flechas, o verlos en Python Tutor.
