## Parte 1 — El problema

## Parte 2 — Paso 1: la clase `Silla`

`ocupar` y `liberar` retornan `bool`: avisan si pudieron hacerlo. `__str__` retorna `"X"` u `"O"`.

````python
class Silla:

    def __init__(self, fila, numero):
        """
        Supone:
            fila es una letra mayuscula y numero es un entero >= 1.
        Garantiza:
            Crea una silla libre con esa fila y ese numero.
        """
        self.__fila = fila
        self.__numero = numero
        self.__ocupada = False

    def get_fila(self):
        return self.__fila

    def get_numero(self):
        return self.__numero

    def esta_libre(self):
        """Garantiza: retorna True si la silla no esta ocupada."""
        return not self.__ocupada

    def ocupar(self):
        """
        Garantiza:
            Si la silla estaba libre, la ocupa y retorna True.
            Si ya estaba ocupada, no cambia nada y retorna False.
        """
        ocupada_ahora = False
        if not self.__ocupada:
            self.__ocupada = True
            ocupada_ahora = True
        return ocupada_ahora

    def liberar(self):
        """
        Garantiza:
            Si la silla estaba ocupada, la libera y retorna True.
            Si ya estaba libre, no cambia nada y retorna False.
        """
        liberada = False
        if self.__ocupada:
            self.__ocupada = False
            liberada = True
        return liberada

    def __str__(self):
        texto = "O"
        if self.__ocupada:
            texto = "X"
        return texto
````

````python
s = Silla("B", 3)
print(s.get_fila(), s.get_numero(), s)
print(s.ocupar(), s)          # se ocupa
print(s.ocupar())             # ya estaba ocupada

assert not s.esta_libre()
assert s.liberar() and s.esta_libre()
assert not s.liberar()        # ya estaba libre
print("Paso 1: todas las pruebas pasaron")
````

## Parte 3 — Paso 2: la sala crea sus sillas

La matriz se crea **igual que cualquier matriz**: primero la lista de filas y después cada fila. La diferencia es que cada casilla recibe un **objeto nuevo**: `Silla(Sala.LETRAS[i], j + 1)`. La fila 0 es la letra A y la columna 0 es la silla 1.

````python
class Sala:

    LETRAS = "ABCDEFGHIJKL"     # la sala tiene a lo sumo 12 filas

    def __init__(self, filas, sillas_por_fila, precio):
        """
        Supone:
            1 <= filas <= 12, sillas_por_fila >= 1 y precio >= 0.
        Garantiza:
            Crea la sala con todas sus sillas libres (COMPOSICION: la
            sala crea sus sillas).
        """
        self.__filas = filas
        self.__sillas_por_fila = sillas_por_fila
        self.__precio = precio
        self.__vendidas = 0
        self.__sillas = [None] * filas
        for i in range(filas):
            self.__sillas[i] = [None] * sillas_por_fila
            for j in range(sillas_por_fila):
                self.__sillas[i][j] = Silla(Sala.LETRAS[i], j + 1)

    def get_filas(self):
        return self.__filas

    def get_sillas_por_fila(self):
        return self.__sillas_por_fila

    def __str__(self):
        texto = "  "
        for j in range(self.__sillas_por_fila):
            texto = texto + f" {j + 1}"
        for i in range(self.__filas):
            texto = texto + "\n" + Sala.LETRAS[i] + " "
            for j in range(self.__sillas_por_fila):
                texto = texto + " " + str(self.__sillas[i][j])
        return texto
````

## Parte 4 — Paso 3: dos formas de contar

El público dice **C4**: fila con letra y silla desde 1. La matriz guarda `sillas[2][3]`: desde 0. La conversión se hace **en un solo lugar**, `get_silla`.

![C4 para el público, sillas[2][3] en la matriz](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/14/teoria/images/sala_dos_numeraciones.png)

````python
class Sala:

    LETRAS = "ABCDEFGHIJKL"     # la sala tiene a lo sumo 12 filas

    def __init__(self, filas, sillas_por_fila, precio):
        """
        Supone:
            1 <= filas <= 12, sillas_por_fila >= 1 y precio >= 0.
        Garantiza:
            Crea la sala con todas sus sillas libres (COMPOSICION: la
            sala crea sus sillas).
        """
        self.__filas = filas
        self.__sillas_por_fila = sillas_por_fila
        self.__precio = precio
        self.__vendidas = 0
        self.__sillas = [None] * filas
        for i in range(filas):
            self.__sillas[i] = [None] * sillas_por_fila
            for j in range(sillas_por_fila):
                self.__sillas[i][j] = Silla(Sala.LETRAS[i], j + 1)

    def get_filas(self):
        return self.__filas

    def get_sillas_por_fila(self):
        return self.__sillas_por_fila

    def indice_fila(self, letra):
        """
        Garantiza:
            Retorna el indice (desde 0) de la fila con esa letra, o -1
            si la sala no tiene esa fila.
        """
        indice = -1
        for i in range(self.__filas):
            if Sala.LETRAS[i] == letra:
                indice = i
        return indice

    def es_silla_valida(self, letra, numero):
        """Garantiza: retorna True si la sala tiene la silla letra-numero."""
        return self.indice_fila(letra) != -1 and 1 <= numero <= self.__sillas_por_fila

    def get_silla(self, letra, numero):
        """
        Supone:
            es_silla_valida(letra, numero) es True.
        Garantiza:
            Retorna el objeto Silla de esa fila y ese numero.
        """
        return self.__sillas[self.indice_fila(letra)][numero - 1]

    def __str__(self):
        texto = "  "
        for j in range(self.__sillas_por_fila):
            texto = texto + f" {j + 1}"
        for i in range(self.__filas):
            texto = texto + "\n" + Sala.LETRAS[i] + " "
            for j in range(self.__sillas_por_fila):
                texto = texto + " " + str(self.__sillas[i][j])
        return texto
````

`indice_fila` es una **búsqueda**: recorre `LETRAS` hasta encontrar la letra, y retorna `-1` si la sala no tiene esa fila.

````python
sala = Sala(3, 4, 5000)
print(sala.indice_fila("A"), sala.indice_fila("C"), sala.indice_fila("D"))
print(sala.es_silla_valida("C", 4), sala.es_silla_valida("C", 5), sala.es_silla_valida("Z", 1))
silla = sala.get_silla("C", 4)
print(silla.get_fila(), silla.get_numero())

assert sala.indice_fila("D") == -1
assert not sala.es_silla_valida("A", 0)
print("Paso 3: todas las pruebas pasaron")
````

## Parte 5 — Paso 4: vender, liberar y contar

`vender` **le pide** a la silla que se ocupe; solo cuenta la venta si la silla aceptó. La sala no cambia el estado de una silla por fuera.

````python
class Sala:

    LETRAS = "ABCDEFGHIJKL"     # la sala tiene a lo sumo 12 filas

    def __init__(self, filas, sillas_por_fila, precio):
        """
        Supone:
            1 <= filas <= 12, sillas_por_fila >= 1 y precio >= 0.
        Garantiza:
            Crea la sala con todas sus sillas libres (COMPOSICION: la
            sala crea sus sillas).
        """
        self.__filas = filas
        self.__sillas_por_fila = sillas_por_fila
        self.__precio = precio
        self.__vendidas = 0
        self.__sillas = [None] * filas
        for i in range(filas):
            self.__sillas[i] = [None] * sillas_por_fila
            for j in range(sillas_por_fila):
                self.__sillas[i][j] = Silla(Sala.LETRAS[i], j + 1)

    def get_filas(self):
        return self.__filas

    def get_sillas_por_fila(self):
        return self.__sillas_por_fila

    def indice_fila(self, letra):
        """
        Garantiza:
            Retorna el indice (desde 0) de la fila con esa letra, o -1
            si la sala no tiene esa fila.
        """
        indice = -1
        for i in range(self.__filas):
            if Sala.LETRAS[i] == letra:
                indice = i
        return indice

    def es_silla_valida(self, letra, numero):
        """Garantiza: retorna True si la sala tiene la silla letra-numero."""
        return self.indice_fila(letra) != -1 and 1 <= numero <= self.__sillas_por_fila

    def get_silla(self, letra, numero):
        """
        Supone:
            es_silla_valida(letra, numero) es True.
        Garantiza:
            Retorna el objeto Silla de esa fila y ese numero.
        """
        return self.__sillas[self.indice_fila(letra)][numero - 1]

    def vender(self, letra, numero):
        """
        Garantiza:
            Si la silla existe y esta libre, la ocupa, cuenta la venta y
            retorna True. Si no, no cambia nada y retorna False.
        """
        vendida = False
        if self.es_silla_valida(letra, numero):
            if self.get_silla(letra, numero).ocupar():
                self.__vendidas = self.__vendidas + 1
                vendida = True
        return vendida

    def liberar(self, letra, numero):
        """
        Garantiza:
            Si la silla existe y esta ocupada, la libera, descuenta la
            venta y retorna True. Si no, no cambia nada y retorna False.
        """
        liberada = False
        if self.es_silla_valida(letra, numero):
            if self.get_silla(letra, numero).liberar():
                self.__vendidas = self.__vendidas - 1
                liberada = True
        return liberada

    def contar_libres_fila(self, letra):
        """
        Supone:
            la sala tiene una fila con esa letra.
        Garantiza:
            Retorna cuantas sillas libres tiene esa fila.
        """
        i = self.indice_fila(letra)
        libres = 0
        for j in range(self.__sillas_por_fila):
            if self.__sillas[i][j].esta_libre():
                libres = libres + 1
        return libres

    def contar_libres(self):
        """Garantiza: retorna cuantas sillas libres tiene toda la sala."""
        libres = 0
        for i in range(self.__filas):
            libres = libres + self.contar_libres_fila(Sala.LETRAS[i])
        return libres

    def get_recaudo(self):
        """Garantiza: retorna el dinero de las boletas vendidas."""
        return self.__vendidas * self.__precio

    def __str__(self):
        texto = "  "
        for j in range(self.__sillas_por_fila):
            texto = texto + f" {j + 1}"
        for i in range(self.__filas):
            texto = texto + "\n" + Sala.LETRAS[i] + " "
            for j in range(self.__sillas_por_fila):
                texto = texto + " " + str(self.__sillas[i][j])
        return texto
````

````python
sala = Sala(3, 4, 5000)
print(sala.vender("B", 2))        # True
print(sala.vender("B", 2))        # False: ya vendida
print(sala.vender("D", 1))        # False: no existe
print(sala.contar_libres_fila("B"), sala.contar_libres(), sala.get_recaudo())

assert sala.liberar("B", 2)
assert sala.get_recaudo() == 0
print("Paso 4: todas las pruebas pasaron")
````

````python
sala = Sala(5, 6, 5000)
ventas = [("A", 1), ("C", 4), ("B", 5), ("B", 6), ("C", 4), ("F", 1)]
for letra, numero in ventas:
    if sala.vender(letra, numero):
        print(f"{letra}{numero}: vendida")
    else:
        print(f"{letra}{numero}: no se pudo")

print(sala)
for i in range(sala.get_filas()):
    letra = Sala.LETRAS[i]
    print(f"Fila {letra}: {sala.contar_libres_fila(letra)} libres")
print("Libres en total:", sala.contar_libres())
print("Recaudo: $", sala.get_recaudo())
````

def fila_mas_llena(sala):
    mejor = Sala.LETRAS[0]
    for i in range(1, sala.get_filas()):
        letra = Sala.LETRAS[i]
        if sala.contar_libres_fila(letra) < sala.contar_libres_fila(mejor):
            mejor = letra
    return mejor

sala = Sala(3, 4, 5000)
