## Parte 2 — Paso 1: crear la matriz y verla

El constructor crea la lista de listas **por dentro**, en los mismos dos pasos de siempre. `__str__` arma un texto con una fila por línea; cada número ocupa 6 espacios y lleva un decimal (`:6.1f`).

````python
class Matriz:

    def __init__(self, filas, columnas, valor=0.0):
        """
        Supone:
            filas y columnas son enteros mayores que 0.
        Garantiza:
            Crea una matriz de filas x columnas con todas las casillas
            en valor (0.0 si no se da).
        """
        self.__filas = filas
        self.__columnas = columnas
        self.__datos = [0.0] * filas
        for i in range(filas):
            self.__datos[i] = [valor] * columnas

    def get_filas(self):
        return self.__filas

    def get_columnas(self):
        return self.__columnas

    def __str__(self):
        texto = ""
        for i in range(self.__filas):
            for j in range(self.__columnas):
                texto = texto + f"{self.__datos[i][j]:6.1f}"
            if i < self.__filas - 1:
                texto = texto + "\n"
        return texto
````

````python
m = Matriz(2, 3)
print(m.get_filas(), m.get_columnas())
print(m)
print(Matriz(2, 2, 1.5))

assert m.get_filas() == 2 and m.get_columnas() == 3
assert str(Matriz(1, 2, 3.0)) == "   3.0   3.0"
print("Paso 1: todas las pruebas pasaron")
````

## Parte 3 — Paso 2: leer y cambiar casillas

`set` **valida** la posición antes de guardar, con `es_posicion_valida`, y retorna `True` o `False`: es el mismo patrón de `set_base` del rectángulo de la clase 13. `get` **supone** que la posición es válida.

````python
class Matriz:

    def __init__(self, filas, columnas, valor=0.0):
        """
        Supone:
            filas y columnas son enteros mayores que 0.
        Garantiza:
            Crea una matriz de filas x columnas con todas las casillas
            en valor (0.0 si no se da).
        """
        self.__filas = filas
        self.__columnas = columnas
        self.__datos = [0.0] * filas
        for i in range(filas):
            self.__datos[i] = [valor] * columnas

    def get_filas(self):
        return self.__filas

    def get_columnas(self):
        return self.__columnas

    def es_posicion_valida(self, i, j):
        """Garantiza: retorna True si (i, j) es una casilla de la matriz."""
        return 0 <= i < self.__filas and 0 <= j < self.__columnas

    def get(self, i, j):
        """
        Supone:
            (i, j) es una posicion valida.
        Garantiza:
            Retorna el valor de la casilla (i, j).
        """
        return self.__datos[i][j]

    def set(self, i, j, valor):
        """
        Garantiza:
            Si (i, j) es una posicion valida, guarda valor en esa casilla
            y retorna True. Si no, no cambia nada y retorna False.
        """
        guardado = False
        if self.es_posicion_valida(i, j):
            self.__datos[i][j] = valor
            guardado = True
        return guardado

    def __str__(self):
        texto = ""
        for i in range(self.__filas):
            for j in range(self.__columnas):
                texto = texto + f"{self.__datos[i][j]:6.1f}"
            if i < self.__filas - 1:
                texto = texto + "\n"
        return texto
````

````python
m = Matriz(2, 3)
print(m.set(0, 0, 4.0))      # posicion valida
print(m.set(2, 0, 9.0))      # la fila 2 no existe
print(m.set(0, -1, 9.0))     # no se aceptan negativos
print(m)

assert m.get(0, 0) == 4.0
assert not m.set(2, 0, 9.0)
assert m.get(0, 0) == 4.0    # no cambio nada
print("Paso 2: todas las pruebas pasaron")
````

## Parte 4 — Paso 3: consultas que recorren la matriz

Dentro de la clase, los ciclos usan `self.__filas` y `self.__columnas`: **el tamaño ya lo conoce el objeto**, no hay que pasarlo ni averiguarlo con `len`.

````python
class Matriz:

    def __init__(self, filas, columnas, valor=0.0):
        """
        Supone:
            filas y columnas son enteros mayores que 0.
        Garantiza:
            Crea una matriz de filas x columnas con todas las casillas
            en valor (0.0 si no se da).
        """
        self.__filas = filas
        self.__columnas = columnas
        self.__datos = [0.0] * filas
        for i in range(filas):
            self.__datos[i] = [valor] * columnas

    def get_filas(self):
        return self.__filas

    def get_columnas(self):
        return self.__columnas

    def es_posicion_valida(self, i, j):
        """Garantiza: retorna True si (i, j) es una casilla de la matriz."""
        return 0 <= i < self.__filas and 0 <= j < self.__columnas

    def get(self, i, j):
        """
        Supone:
            (i, j) es una posicion valida.
        Garantiza:
            Retorna el valor de la casilla (i, j).
        """
        return self.__datos[i][j]

    def set(self, i, j, valor):
        """
        Garantiza:
            Si (i, j) es una posicion valida, guarda valor en esa casilla
            y retorna True. Si no, no cambia nada y retorna False.
        """
        guardado = False
        if self.es_posicion_valida(i, j):
            self.__datos[i][j] = valor
            guardado = True
        return guardado

    def sumar_fila(self, i):
        """
        Supone:
            0 <= i < filas.
        Garantiza:
            Retorna la suma de la fila i.
        """
        suma = 0.0
        for j in range(self.__columnas):
            suma = suma + self.__datos[i][j]
        return suma

    def sumar_columna(self, j):
        """
        Supone:
            0 <= j < columnas.
        Garantiza:
            Retorna la suma de la columna j.
        """
        suma = 0.0
        for i in range(self.__filas):
            suma = suma + self.__datos[i][j]
        return suma

    def es_cuadrada(self):
        """Garantiza: retorna True si la matriz tiene tantas filas como columnas."""
        return self.__filas == self.__columnas

    def __str__(self):
        texto = ""
        for i in range(self.__filas):
            for j in range(self.__columnas):
                texto = texto + f"{self.__datos[i][j]:6.1f}"
            if i < self.__filas - 1:
                texto = texto + "\n"
        return texto
````

````python
DATOS = [[3.0, 3.5, 2.5],
         [4.5, 4.0, 5.0],
         [2.0, 3.0, 3.5],
         [4.0, 4.5, 3.0]]
notas = Matriz(4, 3)
for i in range(4):
    for j in range(3):
        notas.set(i, j, DATOS[i][j])
print(notas)
for i in range(notas.get_filas()):
    print(f"Estudiante {i}: {notas.sumar_fila(i) / notas.get_columnas():.2f}")
print("Suma del parcial 1:", notas.sumar_columna(0))
print("Es cuadrada?", notas.es_cuadrada())

assert notas.sumar_fila(1) == 13.5
assert notas.sumar_columna(0) == 13.5
assert Matriz(3, 3).es_cuadrada()
print("Paso 3: todas las pruebas pasaron")
````

## Parte 5 — Paso 4: métodos que crean otra matriz

`transpuesta()` crea `Matriz(columnas, filas)`, la llena con `set` y la **retorna**. `sumar(otra)` recibe **otro objeto `Matriz`** y le pide sus valores con `otra.get(i, j)`. Ninguno de los dos modifica la matriz original.

````python
class Matriz:

    def __init__(self, filas, columnas, valor=0.0):
        """
        Supone:
            filas y columnas son enteros mayores que 0.
        Garantiza:
            Crea una matriz de filas x columnas con todas las casillas
            en valor (0.0 si no se da).
        """
        self.__filas = filas
        self.__columnas = columnas
        self.__datos = [0.0] * filas
        for i in range(filas):
            self.__datos[i] = [valor] * columnas

    def get_filas(self):
        return self.__filas

    def get_columnas(self):
        return self.__columnas

    def es_posicion_valida(self, i, j):
        """Garantiza: retorna True si (i, j) es una casilla de la matriz."""
        return 0 <= i < self.__filas and 0 <= j < self.__columnas

    def get(self, i, j):
        """
        Supone:
            (i, j) es una posicion valida.
        Garantiza:
            Retorna el valor de la casilla (i, j).
        """
        return self.__datos[i][j]

    def set(self, i, j, valor):
        """
        Garantiza:
            Si (i, j) es una posicion valida, guarda valor en esa casilla
            y retorna True. Si no, no cambia nada y retorna False.
        """
        guardado = False
        if self.es_posicion_valida(i, j):
            self.__datos[i][j] = valor
            guardado = True
        return guardado

    def sumar_fila(self, i):
        """
        Supone:
            0 <= i < filas.
        Garantiza:
            Retorna la suma de la fila i.
        """
        suma = 0.0
        for j in range(self.__columnas):
            suma = suma + self.__datos[i][j]
        return suma

    def sumar_columna(self, j):
        """
        Supone:
            0 <= j < columnas.
        Garantiza:
            Retorna la suma de la columna j.
        """
        suma = 0.0
        for i in range(self.__filas):
            suma = suma + self.__datos[i][j]
        return suma

    def es_cuadrada(self):
        """Garantiza: retorna True si la matriz tiene tantas filas como columnas."""
        return self.__filas == self.__columnas

    def transpuesta(self):
        """
        Garantiza:
            Retorna una Matriz NUEVA de columnas x filas, donde la
            casilla (j, i) tiene el valor de la casilla (i, j) de esta.
            Esta matriz no cambia.
        """
        resultado = Matriz(self.__columnas, self.__filas)
        for i in range(self.__filas):
            for j in range(self.__columnas):
                resultado.set(j, i, self.__datos[i][j])
        return resultado

    def sumar(self, otra):
        """
        Supone:
            otra es una Matriz del mismo tamano que esta.
        Garantiza:
            Retorna una Matriz NUEVA con la suma casilla por casilla.
            Ni esta matriz ni otra cambian.
        """
        resultado = Matriz(self.__filas, self.__columnas)
        for i in range(self.__filas):
            for j in range(self.__columnas):
                resultado.set(i, j, self.__datos[i][j] + otra.get(i, j))
        return resultado

    def __str__(self):
        texto = ""
        for i in range(self.__filas):
            for j in range(self.__columnas):
                texto = texto + f"{self.__datos[i][j]:6.1f}"
            if i < self.__filas - 1:
                texto = texto + "\n"
        return texto
````

````python
DATOS = [[3.0, 3.5, 2.5],
         [4.5, 4.0, 5.0],
         [2.0, 3.0, 3.5],
         [4.0, 4.5, 3.0]]
notas = Matriz(4, 3)
for i in range(4):
    for j in range(3):
        notas.set(i, j, DATOS[i][j])
t = notas.transpuesta()
print(t)
print("notas sigue de", notas.get_filas(), "x", notas.get_columnas())

s = notas.sumar(Matriz(4, 3, 0.5))      # media unidad mas a cada nota
print(s)

assert t.get_filas() == 3 and t.get_columnas() == 4
assert t.get(2, 1) == 5.0                # Beto, parcial 3
assert s.get(0, 0) == 3.5
assert notas.get(0, 0) == 3.0            # notas no cambio
print("Paso 4: todas las pruebas pasaron")
````

## Parte 6 — Integración: el programa que usa la clase

El programa principal **no toca la lista de listas**: crea la matriz, guarda valores con `set` y opera con métodos. Es el `main.py` de [`matriz_poo/`](https://github.com/algoritmos-UdeA/logica1-20262/tree/main/clases/14/teoria/codigo/matriz_poo).

````python
NOMBRES = ["Ana", "Beto", "Carla", "Dario"]
DATOS = [[3.0, 3.5, 2.5],
         [4.5, 4.0, 5.0],
         [2.0, 3.0, 3.5],
         [4.0, 4.5, 3.0]]
notas = Matriz(4, 3)
for i in range(4):
    for j in range(3):
        notas.set(i, j, DATOS[i][j])

print("Notas:")
print(notas)
for i in range(notas.get_filas()):
    promedio = notas.sumar_fila(i) / notas.get_columnas()
    print(f"{NOMBRES[i]}: {promedio:.2f}")
for j in range(notas.get_columnas()):
    promedio = notas.sumar_columna(j) / notas.get_filas()
    print(f"Parcial {j + 1}: {promedio:.2f}")
print("Transpuesta (una fila por parcial):")
print(notas.transpuesta())
````

### ✏️ Reto — `multiplicar`

`C = A × B` existe si las columnas de `A` son tantas como las filas de `B`. `C[i][k]` es la suma de `A[i][j] * B[j][k]` para cada `j`: hacen falta **tres ciclos**. Agregue el método `multiplicar(otra)`.

<details>
<summary>Ver solución</summary>

```python
    def multiplicar(self, otra):
        resultado = Matriz(self.__filas, otra.get_columnas())
        for i in range(self.__filas):
            for k in range(otra.get_columnas()):
                suma = 0.0
                for j in range(self.__columnas):
                    suma = suma + self.__datos[i][j] * otra.get(j, k)
                resultado.set(i, k, suma)
        return resultado
```
