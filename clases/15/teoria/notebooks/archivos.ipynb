## Parte 0 — Preparar los archivos de ejemplo

Esta celda **crea** los dos archivos que usan los ejemplos y define `mostrar_archivo`, que muestra el contenido de un archivo tal como quedó en el disco. Ejecútela primero. Su código usa lo que se explica en las partes siguientes.

````python
def crear_archivo(nombre_archivo, contenido):
    """Garantiza: CREA (o REEMPLAZA) el archivo con el texto dado."""
    archivo = open(nombre_archivo, "w")
    archivo.write(contenido)
    archivo.close()


def mostrar_archivo(nombre_archivo):
    """Garantiza: muestra el contenido del archivo, tal como esta en el disco."""
    archivo = open(nombre_archivo, "r")
    print(f"----- {nombre_archivo} -----")
    contenido = archivo.read()
    print(contenido, end="")
    if not contenido.endswith("\n"):     # el archivo no termina en salto de linea
        print()
    print("-" * (len(nombre_archivo) + 12))
    archivo.close()


crear_archivo("trabalenguas.txt",
              "Pedro Pablo Perez Pintor\n"
              "pinta puertas por poco precio\n"
              "para personas pudientes.\n")
crear_archivo("notas.csv",
              "nombre,parcial1,parcial2,parcial3\n"
              "Ana,3.0,3.5,2.5\n"
              "Beto,4.5,4.0,5.0\n"
              "Carla,2.0,3.0,3.5\n"
              "Dario,4.0,4.5,3.0\n")
mostrar_archivo("trabalenguas.txt")
mostrar_archivo("notas.csv")
````

## Parte 1 — Abrir, procesar y cerrar

Todo trabajo con archivos sigue el mismo ciclo. `open(nombre_archivo, modo)` retorna un **objeto archivo**, con atributos y métodos.

![Abrir, procesar, cerrar](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/15/teoria/images/ciclo_archivo.png)

````python
archivo = open("trabalenguas.txt", "r")
print("Nombre del archivo:", archivo.name)
print("Modo de apertura:  ", archivo.mode)
print("Cerrado?           ", archivo.closed)
archivo.close()
print("Cerrado?           ", archivo.closed)
````

| Modo | Para qué | Si el archivo existe… | Si no existe… |
|---|---|---|---|
| `"r"` | Leer | Lo abre para leer desde el inicio | `FileNotFoundError` |
| `"w"` | Escribir | **Lo vacía**: lo que tenía se pierde | Lo crea |
| `"a"` | Agregar | Escribe al final, sin borrar | Lo crea |

## Parte 2 — Escribir

`write(texto)` escribe el texto **tal cual**: no agrega el salto de línea.

````python
archivo = open("saludo.txt", "w")
archivo.write("Hola, Ana")
archivo.write("\n")
archivo.write("Hasta pronto\n")
archivo.close()

mostrar_archivo("saludo.txt")
````

> ⚠️ **Error frecuente.** Olvidar el `"\n"`. No hay error: todo queda en una sola línea.

````python
archivo = open("saludo.txt", "w")
archivo.write("Hola, Ana")
archivo.write("Hasta pronto")
archivo.close()

mostrar_archivo("saludo.txt")
````

Para escribir una lista, un elemento por línea, se recorre. Los números se convierten a texto con un f-string:

````python
notas = [3.0, 4.5, 2.0, 4.0]
archivo = open("notas.txt", "w")
for i in range(len(notas)):
    archivo.write(f"{notas[i]}\n")
archivo.close()

mostrar_archivo("notas.txt")
````

> ⚠️ **Error frecuente.** Escribir un número sin convertirlo: `write` solo recibe texto.

````python
archivo = open("notas.txt", "w")
archivo.write(4.5)
````

````python
archivo.close()     # el archivo quedo abierto por el error: se cierra
````

## Parte 4 — Leer

### Todo de una vez: `read()`

Retorna **un solo texto** (`str`), con los `"\n"` adentro. `read(n)` retorna los `n` caracteres siguientes: el archivo **recuerda hasta dónde se ha leído**.

````python
archivo = open("trabalenguas.txt", "r")
contenido = archivo.read()
archivo.close()
print(contenido)
print(type(contenido))

archivo = open("trabalenguas.txt", "r")
print(archivo.read(5))        # Pedro
print(archivo.read(7))        # " Pablo " (sigue donde quedo)
archivo.close()
````

### Línea por línea: `for linea in archivo`

Cada línea trae su `"\n"`, y `print` agrega otro: por eso aparecen líneas en blanco. `strip()` los quita.

![Cada línea termina con un salto de línea](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/15/teoria/images/lectura_lineas.png)

````python
archivo = open("trabalenguas.txt", "r")
for linea in archivo:
    print(linea)               # deja una linea en blanco entre cada una
archivo.close()

print("-----")
archivo = open("trabalenguas.txt", "r")
for linea in archivo:
    print(linea.strip())       # sin el "\n" del final
archivo.close()
````

🔍 **Para entender.** `repr` muestra el texto con sus caracteres especiales visibles:

````python
archivo = open("trabalenguas.txt", "r")
for linea in archivo:
    print(repr(linea))
archivo.close()
````

### Una línea a la vez: `readline()`

Al final del archivo retorna `""`: esa es la condición del `while`. Sirve cuando una parte se trata distinto, como el encabezado de un CSV.

````python
archivo = open("notas.csv", "r")
encabezado = archivo.readline()       # la primera linea
print("Encabezado:", encabezado.strip())

linea = archivo.readline()
while linea != "":
    print("Dato:", linea.strip())
    linea = archivo.readline()
archivo.close()
````

### Todas en una lista: `readlines()`

````python
archivo = open("trabalenguas.txt", "r")
lineas = archivo.readlines()
archivo.close()

print(lineas)
print(len(lineas), "lineas")
print(lineas[1].strip())
````

| Forma | Qué retorna | Cuándo usarla |
|---|---|---|
| `archivo.read()` | Todo el archivo, un solo texto | Archivos pequeños que se tratan como un todo |
| `for linea in archivo:` | Una línea en cada vuelta | Procesar todas las líneas (la más usada) |
| `archivo.readline()` | La línea siguiente (`""` al final) | Tratar distinto una parte: el encabezado |
| `archivo.readlines()` | Una lista con todas las líneas | Cuando se necesitan todas a la vez |

## Parte 5 — Del texto al dato

Todo lo que se lee de un archivo es **texto**. `strip()` limpia, `split(",")` parte y `float()` convierte. Esta celda no usa archivos, así que tiene enlace a Python Tutor:

![strip, split y float](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/15/teoria/images/split_linea.png)

````python
linea = "Beto,4.5,4.0,5.0\n"      # asi llega del archivo

limpia = linea.strip()             # sin el "\n"
partes = limpia.split(",")         # lista de 4 textos
print(partes)

nombre = partes[0]
notas = [float(partes[1]), float(partes[2]), float(partes[3])]
print(nombre, notas)
print(partes[1] + partes[2])       # une textos: no suma
print(notas[0] + notas[1])         # suma numeros
````

## Parte 6 — Leer y escribir un CSV

Como no se sabe de antemano cuántas líneas hay, las listas empiezan **vacías** y crecen con `append`.

![De notas.csv a la lista nombres y la matriz notas](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/15/teoria/images/csv_a_listas.png)

````python
nombres = []
notas = []                  # una fila por estudiante

archivo = open("notas.csv", "r")
archivo.readline()          # se salta el encabezado
for linea in archivo:
    partes = linea.strip().split(",")
    nombres.append(partes[0])
    fila = []
    for j in range(1, len(partes)):
        fila.append(float(partes[j]))
    notas.append(fila)
archivo.close()

print(nombres)
print(notas)
print(len(notas), "estudiantes,", len(notas[0]), "parciales")
````

Para escribir un CSV: primero el encabezado, después una línea por fila.

````python
nombres = ["Ana", "Beto", "Carla", "Dario"]
notas = [[3.0, 3.5, 2.5],
         [4.5, 4.0, 5.0],
         [2.0, 3.0, 3.5],
         [4.0, 4.5, 3.0]]

archivo = open("notas_copia.csv", "w")
archivo.write("nombre,parcial1,parcial2,parcial3\n")
for i in range(len(nombres)):
    archivo.write(f"{nombres[i]},{notas[i][0]},{notas[i][1]},{notas[i][2]}\n")
archivo.close()

mostrar_archivo("notas_copia.csv")
````

> 📏 **Excel en español** guarda los CSV con **punto y coma** (`;`) y **coma decimal** (`3,5`). Eso rompe `split(",")` y `float()`. Si va a abrir o guardar un CSV en Excel, revise el separador antes de leerlo.

## Parte 7 — `with`: el archivo se cierra solo

````python
with open("trabalenguas.txt", "r") as archivo:
    for linea in archivo:
        print(linea.strip())
    print("Dentro del with, cerrado?", archivo.closed)

print("Fuera del with, cerrado?", archivo.closed)
````

¿Por qué importa cerrar? `write` junta los datos en un **buffer** en la memoria y los pasa al disco de a bloques; `close()` vacía el buffer. Si no se cierra, lo último escrito puede no quedar en el archivo.

def contar_lineas(nombre_archivo):
    cantidad = 0
    archivo = open(nombre_archivo, "r")
    for linea in archivo:

def agregar_estudiante(nombre_archivo, nombre, n1, n2, n3):
