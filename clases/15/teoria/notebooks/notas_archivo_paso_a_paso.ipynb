## Parte 0 — Preparar

`crear_archivo` y `mostrar_archivo` son dos ayudas para el notebook: la primera crea un archivo con un texto y la segunda muestra cómo quedó en el disco. La celda también crea `notas.csv`.

````python
def crear_archivo(nombre_archivo, contenido):
    """Garantiza: CREA (o REEMPLAZA) el archivo con el texto dado."""
    archivo = open(nombre_archivo, "w")
    archivo.write(contenido)
    archivo.close()


def mostrar_archivo(nombre_archivo):
    """Garantiza: muestra el contenido del archivo, tal como esta en el disco."""
    archivo = open(nombre_archivo, "r")
    contenido = archivo.read()
    print(f"----- {nombre_archivo} -----")
    print(contenido, end="")
    if not contenido.endswith("\n"):
        print()
    print("-" * (len(nombre_archivo) + 12))
    archivo.close()


crear_archivo("notas.csv",
              "nombre,parcial1,parcial2,parcial3\n"
              "Ana,3.0,3.5,2.5\n"
              "Beto,4.5,4.0,5.0\n"
              "Carla,2.0,3.0,3.5\n"
              "Dario,4.0,4.5,3.0\n")
mostrar_archivo("notas.csv")
````

## Parte 2 — Paso 1: `leer_nombres`

Recibe el **nombre** del archivo, no el archivo abierto: lo abre, se salta el encabezado, lo recorre y lo cierra. La lista crece con `append`.

````python
def leer_nombres(nombre_archivo):
    """
    Supone:
        nombre_archivo es un CSV (encabezado + nombre,p1,p2,p3) que existe.
    Garantiza:
        Retorna una lista nueva con los nombres, en el orden del archivo.
    """
    nombres = []
    archivo = open(nombre_archivo, "r")
    archivo.readline()
    for linea in archivo:
        partes = linea.strip().split(",")
        nombres.append(partes[0])
    archivo.close()
    return nombres
````

Para probarla **no** se usa `notas.csv`: la prueba escribe su propio archivo con datos conocidos. Así no depende de los datos reales del curso.

````python
crear_archivo("prueba.csv",
              "nombre,parcial1,parcial2,parcial3\n"
              "Lisa,5.0,4.5,5.0\n"
              "Bart,2.0,3.0,1.5\n")
print(leer_nombres("prueba.csv"))

assert leer_nombres("prueba.csv") == ["Lisa", "Bart"]                    # tipico

crear_archivo("prueba.csv", "nombre,parcial1,parcial2,parcial3\n")
assert leer_nombres("prueba.csv") == []                                  # limite: solo encabezado
print("Paso 1: todas las pruebas pasaron")
````

## Parte 3 — Paso 2: `leer_notas`

Igual que `leer_nombres`, pero de cada línea toma las notas, las convierte con `float` y arma una fila. La matriz también crece con `append`.

````python
def leer_notas(nombre_archivo):
    """
    Supone:
        nombre_archivo es un CSV (encabezado + nombre,p1,p2,p3) que existe.
    Garantiza:
        Retorna una matriz nueva: una fila por estudiante, una columna por
        parcial (float).
    """
    notas = []
    archivo = open(nombre_archivo, "r")
    archivo.readline()
    for linea in archivo:
        partes = linea.strip().split(",")
        fila = []
        for j in range(1, len(partes)):
            fila.append(float(partes[j]))
        notas.append(fila)
    archivo.close()
    return notas
````

````python
crear_archivo("prueba.csv",
              "nombre,parcial1,parcial2,parcial3\n"
              "Lisa,5.0,4.5,5.0\n"
              "Bart,2.0,3.0,1.5\n")
print(leer_notas("prueba.csv"))

assert leer_notas("prueba.csv") == [[5.0, 4.5, 5.0], [2.0, 3.0, 1.5]]
assert leer_notas("notas.csv")[1] == [4.5, 4.0, 5.0]          # Beto, en el archivo real
print("Paso 2: todas las pruebas pasaron")
````

> ⚠️ **Error frecuente.** Un archivo con una **línea en blanco** al final, algo que dejan muchos editores. Ninguna de las dos funciones falla: agregan un nombre vacío y una fila vacía, **sin avisar**.

````python
crear_archivo("con_blanco.csv",
              "nombre,parcial1,parcial2,parcial3\n"
              "Lisa,5.0,4.5,5.0\n"
              "\n")
print(leer_nombres("con_blanco.csv"))
print(leer_notas("con_blanco.csv"))
````

La línea vacía, partida con `split(",")`, da `['']`: un nombre vacío y ninguna nota. El error aparece **después**, lejos de su causa, cuando otra función usa esa fila vacía (se ve en el paso 3). El *Supone* de las funciones dice "un CSV con el formato del módulo": un archivo con líneas en blanco no lo cumple.

## Parte 4 — Paso 3: calcular (lo de la clase 14) y guardar

`calcular_notas_finales`, `buscar_mayor` y `buscar_menor` son las mismas de la clase 14: **no saben nada de archivos**. Esta celda no usa archivos, así que tiene enlace a Python Tutor.

````python
def calcular_notas_finales(notas, pesos):
    """
    Supone:
        notas es una matriz de numeros con al menos una fila, y pesos es
        una lista con un peso por columna de notas; los pesos suman 1.0.
    Garantiza:
        Retorna una lista nueva con la nota final de cada fila: la suma
        de notas[i][j] * pesos[j]. No modifica notas ni pesos.
    """
    n = len(notas)
    finales = [0.0] * n
    for i in range(n):
        suma = 0.0
        for j in range(len(pesos)):
            suma = suma + notas[i][j] * pesos[j]
        finales[i] = suma
    return finales


def buscar_mayor(valores):
    """
    Supone:
        valores es una lista de numeros con al menos un elemento.
    Garantiza:
        Retorna el mayor valor de la lista. No modifica valores.
    """
    mayor = valores[0]
    for i in range(1, len(valores)):
        if valores[i] > mayor:
            mayor = valores[i]
    return mayor


def buscar_menor(valores):
    """
    Supone:
        valores es una lista de numeros con al menos un elemento.
    Garantiza:
        Retorna el menor valor de la lista. No modifica valores.
    """
    menor = valores[0]
    for i in range(1, len(valores)):
        if valores[i] < menor:
            menor = valores[i]
    return menor
````

````python
finales = calcular_notas_finales([[3.0, 3.5, 2.5], [4.5, 4.0, 5.0]], [0.3, 0.3, 0.4])
print(finales)
assert round(finales[0], 2) == 2.95 and round(finales[1], 2) == 4.55
assert buscar_mayor(finales) == finales[1]
print("Calculo: todas las pruebas pasaron")
````

`guardar_finales` escribe un CSV con el encabezado `nombre,final` y la nota con dos decimales. Su *Garantiza* dice **CREA o REEMPLAZA**.

````python
def guardar_finales(nombre_archivo, nombres, finales):
    """
    Supone:
        nombres y finales tienen el mismo tamano.
    Garantiza:
        CREA el archivo (o lo REEMPLAZA) con el encabezado "nombre,final"
        y una linea por estudiante, con la nota final redondeada a dos
        decimales. No retorna nada.
    """
    archivo = open(nombre_archivo, "w")
    archivo.write("nombre,final\n")
    for i in range(len(nombres)):
        archivo.write(f"{nombres[i]},{finales[i]:.2f}\n")
    archivo.close()
````

Una función que **escribe** un archivo se prueba **leyendo** lo que escribió, línea por línea:

````python
guardar_finales("prueba_finales.csv", ["Ana", "Beto"], [2.95, 4.55])
mostrar_archivo("prueba_finales.csv")

archivo = open("prueba_finales.csv", "r")
assert archivo.readline() == "nombre,final\n"
assert archivo.readline() == "Ana,2.95\n"
assert archivo.readline() == "Beto,4.55\n"
assert archivo.readline() == ""          # no sobran lineas
archivo.close()
print("Paso 3: todas las pruebas pasaron")
````

## Parte 5 — Integración

Es el `main.py` de la carpeta, completo: no pide nada por teclado.

````python
ENTRADA = "notas.csv"
SALIDA = "notas_finales.csv"
PESOS = [0.3, 0.3, 0.4]


def main():
    nombres = leer_nombres(ENTRADA)
    matriz = leer_notas(ENTRADA)
    finales = calcular_notas_finales(matriz, PESOS)

    print(f"Notas leidas de {ENTRADA}: {len(nombres)} estudiantes")
    for i in range(len(nombres)):
        print(f"{nombres[i]}: {finales[i]:.2f}")
    print(f"Nota final mas alta: {buscar_mayor(finales):.2f}")
    print(f"Nota final mas baja: {buscar_menor(finales):.2f}")

    guardar_finales(SALIDA, nombres, finales)
    print(f"Resultado guardado en {SALIDA}")


main()
````

````python
mostrar_archivo("notas_finales.csv")
````

🔍 **Para entender.** Para otro curso no hay que tocar el programa: basta otro archivo. Se agrega una estudiante a `notas.csv` con el modo `"a"` y se ejecuta otra vez:

````python
archivo = open("notas.csv", "a")
archivo.write("Elena,5.0,4.5,4.8\n")
archivo.close()

main()
````

## Parte 6 — ¿Qué fue incómodo?

* **El archivo se lee dos veces**, una en `leer_nombres` y otra en `leer_notas`, porque cada función retorna una sola cosa.
* **Los datos van sueltos**: `nombres` y `matriz` deben tener el mismo tamaño, y nada lo garantiza.
* **El formato está repetido**: `leer_nombres`, `leer_notas` y `guardar_finales` deben saber, cada una, cómo es el archivo.

Con objetos, el objeto que tiene los datos es el que sabe **guardarse** y **cargarse**: el formato queda en un solo lugar. Es lo que hace el notebook [La sala de cine que se guarda](https://github.com/algoritmos-UdeA/logica1-20262/blob/main/clases/15/teoria/notebooks/sala_archivo_paso_a_paso.ipynb).

def guardar_aprobados(nombre_archivo, nombres, finales):
