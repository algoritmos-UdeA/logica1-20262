## Parte 1 — Búsqueda lineal: buscar en desorden

### La idea

Una guía telefónica de 1000 páginas, ordenada por apellido. Se busca a "Pérez". La forma más simple es revisar **página por página** desde el principio: es lenta, pero funciona **aunque la guía esté en desorden**. Eso es la búsqueda lineal. (La otra estrategia, abrir por la mitad, es la Parte 7.)

![Dos formas de buscar en la guía](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/busqueda_guia.png)

En un arreglo: se recorre desde la posición 0, se compara cada elemento con el valor buscado y, al encontrarlo, se guarda su posición y se detiene el recorrido con `break`. Si se recorre todo sin encontrarlo, la posición queda en **-1**, que significa "no está": sirve porque ninguna posición válida es negativa.

![Traza de la búsqueda lineal](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/busqueda_lineal_traza.png)

### El algoritmo, con una traza

```python
X = [5.0, 4.0, 3.5, 2.87]
N = 4
buscado = 3.5
pos = -1
for i in range(N):
    print(f"i = {i}: X[i] = {X[i]}")
    if X[i] == buscado:
        pos = i
        break
if pos == -1:
    print("El valor NO se encontro")
else:
    print(f"El valor se encontro en la posicion {pos}")
```

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=X%20%3D%20%5B5.0%2C%204.0%2C%203.5%2C%202.87%5D%0AN%20%3D%204%0Abuscado%20%3D%203.5%0Apos%20%3D%20-1%0Afor%20i%20in%20range%28N%29%3A%0A%20%20%20%20print%28f%22i%20%3D%20%7Bi%7D%3A%20X%5Bi%5D%20%3D%20%7BX%5Bi%5D%7D%22%29%0A%20%20%20%20if%20X%5Bi%5D%20%3D%3D%20buscado%3A%0A%20%20%20%20%20%20%20%20pos%20%3D%20i%0A%20%20%20%20%20%20%20%20break%0Aif%20pos%20%3D%3D%20-1%3A%0A%20%20%20%20print%28%22El%20valor%20NO%20se%20encontro%22%29%0Aelse%3A%0A%20%20%20%20print%28f%22El%20valor%20se%20encontro%20en%20la%20posicion%20%7Bpos%7D%22%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

Se revisaron 3 casillas; la posición 3 no se revisó, gracias al `break`.

### Como función

Dentro de la función se usa `len(lista)`, porque la función no recibe `N`. Es del **caso 1**: retorna una posición y no modifica la lista.

```python
def buscar_lineal(lista, valor):
    """
    Busca un valor en una lista, recorriéndola desde la posición 0.

    Supone:
        lista es una lista (no necesita estar ordenada).
    Garantiza:
        Retorna la posición de la primera aparición de valor en lista,
        o -1 si valor no está. No modifica lista.
    """
    pos = -1
    for i in range(len(lista)):
        if lista[i] == valor:
            pos = i
            break
    return pos
```

### Pruebas

Las pruebas se planean antes, con casos **típicos**, **límite** y **especiales**:

| Caso | Tipo | Resultado esperado |
|---|---|---|
| 3.5 en `X` | Típico: está en el medio | 2 |
| 5.0 en `X` | Límite: es el primero | 0 |
| 2.87 en `X` | Límite: es el último | 3 |
| 7.0 en `X` | Especial: no está | -1 |
| 1 en `[1, 2, 1]` | Especial: está repetido | 0 (la primera aparición) |
| 3.5 en `[]` | Especial: lista vacía | -1 |

```python
X = [5.0, 4.0, 3.5, 2.87]
assert buscar_lineal(X, 3.5) == 2          # tipico: esta en el medio
assert buscar_lineal(X, 5.0) == 0          # limite: el primero
assert buscar_lineal(X, 2.87) == 3         # limite: el ultimo
assert buscar_lineal(X, 7.0) == -1         # especial: no esta
assert buscar_lineal([1, 2, 1], 1) == 0    # especial: repetido -> la primera
assert buscar_lineal([], 3.5) == -1        # especial: lista vacia
print("Todas las pruebas pasaron")
```

### Contar pasos: ¿qué tan lenta es?

Una forma de comparar algoritmos sin cronómetro es **contar** cuántas veces hacen su operación principal. En la búsqueda, esa operación es comparar con el valor buscado.

```python
def contar_comparaciones_lineal(lista, valor):
    comparaciones = 0
    for i in range(len(lista)):
        comparaciones = comparaciones + 1
        if lista[i] == valor:
            break
    return comparaciones


X = [5.0, 4.0, 3.5, 2.87]
print("El primero:", contar_comparaciones_lineal(X, 5.0))
print("En el medio:", contar_comparaciones_lineal(X, 3.5))
print("El ultimo:", contar_comparaciones_lineal(X, 2.87))
print("No esta:", contar_comparaciones_lineal(X, 7.0))
```

* **Mejor caso**: el valor está en la primera posición. 1 comparación.
* **Peor caso**: el valor está al final o no está. `N` comparaciones.

Si `N` se duplica, el peor caso también se duplica: con 1000 elementos, hasta 1000 comparaciones; con un millón, hasta un millón. *Grokking Algorithms* le llama a esto **tiempo lineal**: el trabajo crece al mismo ritmo que los datos.

## Parte 2 — Intercambiar dos posiciones: la pieza que se reutiliza

Ordenar es, en el fondo, **mover elementos a su lugar**. Dos de los tres métodos de esta clase lo hacen intercambiando dos posiciones. Parece trivial, pero tiene una trampa.

### Como función

`intercambiar` es del **caso 3**: modifica la lista y no tiene `return`. Se escribe una vez y la reutilizan la selección, la burbuja y el programa integrador.

```python
def intercambiar(lista, i, j):
    """
    Intercambia los elementos de las posiciones i y j de una lista.

    Supone:
        i y j son posiciones válidas de lista (0 a len(lista) - 1).
    Garantiza:
        MODIFICA lista: el valor que estaba en i queda en j, y el que
        estaba en j queda en i. No retorna nada.
    """
    temp = lista[i]
    lista[i] = lista[j]
    lista[j] = temp
```

### La posición del menor desde una posición dada

La otra pieza que usa la selección. Busca el menor **desde `inicio` hasta el final**: empieza suponiendo que el menor está en `inicio` y lo compara con los siguientes. Es del **caso 1**: retorna una posición y no modifica la lista.

```python
def posicion_menor(lista, inicio):
    """
    Busca el menor elemento desde la posición inicio hasta el final.

    Supone:
        0 <= inicio < len(lista).
    Garantiza:
        Retorna la posición del menor valor entre lista[inicio] y el
        último elemento (si se repite, la primera). No modifica lista.
    """
    pos_menor = inicio
    for j in range(inicio + 1, len(lista)):
        if lista[j] < lista[pos_menor]:
            pos_menor = j
    return pos_menor
```

## Parte 3 — Ordenamiento por selección

### La idea

De un mazo de cartas en desorden se **escoge la más baja** y se pone a la izquierda; se repite con las que quedan. En cada pasada:

1. se busca el **menor** de la parte sin ordenar;
2. se **intercambia** con la primera posición de esa parte.

La parte ordenada crece de izquierda a derecha, un elemento por pasada. Con `N` elementos bastan `N - 1` pasadas: cuando quedan `N - 1` en su lugar, el último también lo está.

[![Select-sort with Gypsy folk dance](https://img.youtube.com/vi/Ns4TPTC8whw/hqdefault.jpg)](https://www.youtube.com/watch?v=Ns4TPTC8whw)

🎬 [*Select-sort with Gypsy folk dance*](https://www.youtube.com/watch?v=Ns4TPTC8whw) (AlgoRythmics)

En el video, un bailarín recorre la fila buscando al menor y, al final de cada pasada, cambia de lugar con él: **un solo intercambio por pasada**.

![Las pasadas de la selección](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/seleccion_pasadas.png)

### El algoritmo, con una traza

Un `print` por pasada muestra cómo va quedando la lista:

```python
X = [29, 10, 14, 37, 13]
N = 5
print("inicio:", X)
for i in range(N - 1):
    # buscar la posicion del menor desde i
    pos_menor = i
    for j in range(i + 1, N):
        if X[j] < X[pos_menor]:
            pos_menor = j
    # intercambiar X[i] con el menor
    temp = X[i]
    X[i] = X[pos_menor]
    X[pos_menor] = temp
    print("i =", i, "| pos_menor =", pos_menor, "|", X)
print("final:", X)
```

* El **ciclo externo** (`i`) recorre las posiciones que se van ordenando, de 0 a `N - 2`.
* El **ciclo interno** (`j`) busca la posición del menor desde `i + 1` hasta el final.
* En la pasada `i = 2`, el menor ya estaba en su lugar (`pos_menor = 2`): se intercambia consigo mismo y la lista no cambia.

### Como función

Cada comentario del algoritmo se convierte en una llamada a función: el ciclo `j` es `posicion_menor(lista, i)` y las tres líneas del `temp` son `intercambiar(lista, i, pos_menor)`. **El nombre de la función reemplaza al comentario.**

```python
def ordenar_seleccion(lista):
    """
    Ordena una lista de menor a mayor con el método de selección.

    Supone:
        Los elementos de lista se pueden comparar con <.
    Garantiza:
        MODIFICA lista: queda ordenada de menor a mayor. No retorna nada.
    """
    for i in range(len(lista) - 1):
        pos_menor = posicion_menor(lista, i)
        intercambiar(lista, i, pos_menor)
```

## Parte 4 — Ordenamiento por burbuja

### La idea

Se recorre la lista **comparando vecinos**: si están en desorden, se intercambian. En cada pasada el mayor "sube" hasta el final, como una burbuja. Por eso cada pasada revisa un elemento menos. Y si en una pasada **no hubo intercambios**, la lista ya está ordenada y se puede parar.

[![Bubble-sort with Hungarian ("Csángó") folk dance](https://img.youtube.com/vi/lyZQPjUT5B4/hqdefault.jpg)](https://www.youtube.com/watch?v=lyZQPjUT5B4)

🎬 [*Bubble-sort with Hungarian ("Csángó") folk dance*](https://www.youtube.com/watch?v=lyZQPjUT5B4) (AlgoRythmics)

En el video, los bailarines solo se comparan con su **vecino**: si el de la izquierda tiene un número mayor, cambian de lugar.

![Las pasadas de la burbuja](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/burbuja_pasadas.png)

### El algoritmo, con una traza

```python
X = [29, 10, 14, 37, 13]
N = 5
print("inicio:", X)
for i in range(N - 1):
    hubo_intercambio = False
    # comparar vecinos e intercambiar si estan en desorden
    for j in range(N - 1 - i):
        if X[j] > X[j + 1]:
            temp = X[j]
            X[j] = X[j + 1]
            X[j + 1] = temp
            hubo_intercambio = True
    print("i =", i, "| hubo_intercambio =", hubo_intercambio, "|", X)
    # si no hubo intercambios, ya esta ordenada
    if not hubo_intercambio:
        break
print("final:", X)
```

* El **ciclo interno** (`j`) compara `X[j]` con `X[j + 1]`. Llega hasta `N - 2 - i`: hasta la penúltima casilla (para que `j + 1` no se salga), y además `i` casillas menos, porque el final ya quedó ordenado.
* `hubo_intercambio` es una **bandera**: si una pasada completa no la cambia a `True`, se sale con `break`.

### Como función

Las tres líneas del `temp` se reemplazan por `intercambiar(lista, j, j + 1)`: la misma función de la selección. Dentro de la función, `N` se reemplaza por `n = len(lista)`.

```python
def ordenar_burbuja(lista):
    """
    Ordena una lista de menor a mayor con el método de la burbuja.

    Supone:
        Los elementos de lista se pueden comparar con >.
    Garantiza:
        MODIFICA lista: queda ordenada de menor a mayor. Se detiene antes
        si en una pasada no hubo intercambios. No retorna nada.
    """
    n = len(lista)
    for i in range(n - 1):
        hubo_intercambio = False
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                intercambiar(lista, j, j + 1)
                hubo_intercambio = True
        if not hubo_intercambio:
            break
```

## Parte 5 — Ordenamiento por inserción

### La idea

Como al ordenar las cartas **en la mano**: se toma la siguiente carta y se mete en su lugar entre las que ya están ordenadas. La lista tiene dos partes: la ordenada (a la izquierda) y la que falta por revisar (a la derecha). En cada pasada, la parte ordenada crece en uno.

[![Insert-sort with Romanian folk dance](https://img.youtube.com/vi/ROalU379l3U/hqdefault.jpg)](https://www.youtube.com/watch?v=ROalU379l3U)

🎬 [*Insert-sort with Romanian folk dance*](https://www.youtube.com/watch?v=ROalU379l3U) (AlgoRythmics)

En el video, el bailarín que sale de la fila avanza hacia la izquierda mientras encuentra números mayores, que **se corren** para abrirle espacio.

![Las pasadas de la inserción](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/insercion_pasadas.png)

### El algoritmo, con una traza

```python
X = [29, 10, 14, 37, 13]
N = 5
print("inicio:", X)
for i in range(1, N):
    # guardar el actual
    actual = X[i]
    j = i - 1
    # desplazar a la derecha los mayores que el actual
    while j >= 0 and X[j] > actual:
        X[j + 1] = X[j]
        j = j - 1
    # colocar el actual en el hueco
    X[j + 1] = actual
    print("i =", i, "| actual =", actual, "| posicion", j + 1, "|", X)
print("final:", X)
```

* El **ciclo externo** empieza en 1: un solo elemento ya está "ordenado".
* El **ciclo interno** es un `while`, porque no se sabe cuántos mayores habrá que correr.
* El orden de la condición importa: `j >= 0` va **primero**. Cuando `j` llega a -1, Python ya no evalúa `X[j] > actual`, y así no se compara con `X[-1]`, que sería el último elemento.

### El desplazamiento, paso a paso

La última pasada (`i = 4`, `actual = 13`) es la que más mueve. Un `print` dentro del `while` muestra cada desplazamiento:

```python
X = [10, 14, 29, 37, 13]
i = 4
actual = X[i]
j = i - 1
while j >= 0 and X[j] > actual:
    X[j + 1] = X[j]
    j = j - 1
    print(X)
X[j + 1] = actual
print(X)
```

### Como función

Es el único de los tres métodos que **no** usa `intercambiar`.

### Por dentro trabajan distinto

| Método | Qué hace en cada pasada | Funciones de apoyo |
|---|---|---|
| Selección | Busca el menor y lo pone al inicio | `posicion_menor` e `intercambiar` |
| Burbuja | Compara vecinos y lleva el mayor al final | `intercambiar` |
| Inserción | Mete cada elemento en la parte ordenada | ninguna: desplaza |

### Contar comparaciones

Igual que con la búsqueda, se puede **contar** cuántas veces cada método compara dos elementos. Estas versiones son las mismas funciones, con un contador `cont` que aumenta en cada comparación y que se retorna al final:

```python
def comparaciones_seleccion(lista):
    cont = 0
    for i in range(len(lista) - 1):
        pos_menor = i
        for j in range(i + 1, len(lista)):
            cont = cont + 1
            if lista[j] < lista[pos_menor]:
                pos_menor = j
        temp = lista[i]
        lista[i] = lista[pos_menor]
        lista[pos_menor] = temp
    return cont


def comparaciones_burbuja(lista):
    cont = 0
    n = len(lista)
    for i in range(n - 1):
        hubo_intercambio = False
        for j in range(n - 1 - i):
            cont = cont + 1
            if lista[j] > lista[j + 1]:
                temp = lista[j]
                lista[j] = lista[j + 1]
                lista[j + 1] = temp
                hubo_intercambio = True
        if not hubo_intercambio:
            break
    return cont


def comparaciones_insercion(lista):
    cont = 0
    for i in range(1, len(lista)):
        actual = lista[i]
        j = i - 1
        while j >= 0:
            cont = cont + 1
            if lista[j] > actual:
                lista[j + 1] = lista[j]
                j = j - 1
            else:
                break
        lista[j + 1] = actual
    return cont
```

## Parte 7 — Búsqueda binaria: buscar en orden

### La idea

Volviendo a la guía telefónica: como está **ordenada**, se abre por la mitad, se mira si "Pérez" está antes o después, y **se descarta la otra mitad**. Se repite con la mitad que queda. Cada comparación descarta la mitad de lo que falta: con 1000 páginas, unos 10 pasos.

**Requisito:** la lista debe estar ordenada de menor a mayor. Por eso esta búsqueda viene **después** de ordenar.

Tres variables marcan la parte que todavía se revisa: `inicio`, `fin` y `medio = (inicio + fin) // 2`. Se compara el buscado con `X[medio]`:

* si es **igual**: encontrado;
* si `X[medio]` es **menor**: el buscado está a la derecha, `inicio = medio + 1`;
* si `X[medio]` es **mayor**: está a la izquierda, `fin = medio - 1`;
* si `inicio` pasa a `fin`: no queda nada por revisar, no está.

![Traza de la búsqueda binaria](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/busqueda_binaria_traza.png)

### El algoritmo, con una traza

```python
X = [2, 5, 10, 12, 15, 17, 20]
N = 7
buscado = 15
inicio = 0
fin = N - 1
pos = -1
while inicio <= fin:
    medio = (inicio + fin) // 2
    print("inicio =", inicio, "| fin =", fin, "| medio =", medio, "| X[medio] =", X[medio])
    if X[medio] == buscado:
        # encontrado
        pos = medio
        break
    elif X[medio] < buscado:
        # descartar la mitad izquierda
        inicio = medio + 1
    else:
        # descartar la mitad derecha
        fin = medio - 1
print("al salir: inicio =", inicio, "| fin =", fin, "| pos =", pos)
```

### Como función

El *Supone* es **nuevo y obligatorio**: la lista debe estar ordenada de menor a mayor. La función **no lo revisa**; es responsabilidad de quien la llama.

```python
def buscar_binaria(lista, valor):
    """
    Busca un valor en una lista ordenada, partiéndola a la mitad en cada paso.

    Supone:
        lista está ordenada de menor a mayor.
    Garantiza:
        Retorna una posición donde está valor, o -1 si valor no está.
        No modifica lista.
    """
    inicio = 0
    fin = len(lista) - 1
    pos = -1
    while inicio <= fin:
        medio = (inicio + fin) // 2
        if lista[medio] == valor:
            pos = medio
            break
        elif lista[medio] < valor:
            inicio = medio + 1
        else:
            fin = medio - 1
    return pos
```

### Contar pasos: lineal frente a binaria

¿Cuántas veces se puede partir una lista a la mitad hasta que quede un solo elemento? La siguiente función cuenta las vueltas del `while` en el **peor caso**: buscar un valor mayor que todos.

```python
def pasos_binaria(N):
    inicio = 0
    fin = N - 1
    pasos = 0
    while inicio <= fin:
        medio = (inicio + fin) // 2
        pasos = pasos + 1
        inicio = medio + 1      # el buscado es mayor que todos
    return pasos


for N in [10, 100, 1000, 1000000, 1000000000]:
    print(f"N = {N:13}: lineal hasta {N:13} comparaciones, binaria hasta {pasos_binaria(N):2}")
```

## Parte 8 — Integrador: administración de notas de un curso

> Un curso admite máximo `N = 10` estudiantes. De cada uno se guarda su id (texto), su nombre y su nota. Por ahora hay 4 inscritos; se pueden agregar más hasta llenar el cupo. Se necesita un programa para administrar el curso: agregar, eliminar, actualizar, buscar, listar y ordenar.

El programa completo, con su menú, está en [`teoria/codigo/curso/`](https://github.com/algoritmos-UdeA/logica1-20262/blob/main/clases/12/teoria/codigo/curso). Aquí se ve la parte que es tema de esta clase: **ordenar sin separar a cada estudiante de sus datos**.

### El arreglo parcialmente lleno

Se crean tres arreglos de tamaño `N` (el cupo), aunque al principio no haya nadie. Una variable `cantidad` dice cuántas casillas están ocupadas (de 0 a `cantidad - 1`), y es también la **próxima casilla libre**.

![El arreglo parcialmente lleno](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/arreglo_parcial.png)

Por eso las funciones reciben `cantidad` y recorren **hasta ahí**, no hasta `len()`: si no, tratarían las casillas vacías como estudiantes. Las que cambian `cantidad` la retornan.

```python
def buscar_id(ids, cantidad, id_buscado):
    """
    Busca un Id entre los estudiantes inscritos (búsqueda lineal).

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        Retorna la posición de id_buscado entre 0 y cantidad - 1,
        o -1 si no está. No modifica ids.
    """
    pos = -1
    for i in range(cantidad):
        if ids[i] == id_buscado:
            pos = i
            break
    return pos


def agregar_estudiante(ids, nombres, notas, cantidad, id_nuevo, nombre, nota):
    """
    Agrega un estudiante al final de los inscritos.

    Supone:
        ids, nombres y notas tienen el mismo tamaño (el cupo).
        0 <= cantidad <= len(ids).
    Garantiza:
        Si hay cupo y id_nuevo no está inscrito, MODIFICA ids, nombres y
        notas (guarda los datos en la posición cantidad) y retorna
        cantidad + 1. Si no, no modifica nada y retorna cantidad.
    """
    if cantidad < len(ids) and buscar_id(ids, cantidad, id_nuevo) == -1:
        ids[cantidad] = id_nuevo
        nombres[cantidad] = nombre
        notas[cantidad] = nota
        cantidad = cantidad + 1
    return cantidad


def mostrar_listado(ids, nombres, notas, cantidad):
    """
    Supone: 0 <= cantidad <= len(ids).
    Garantiza: imprime una línea por estudiante inscrito (Id, nombre y
    nota). No modifica nada. No retorna nada.
    """
    print("Id", "\t", "Nombre", "\t", "Nota")
    for i in range(cantidad):
        print(ids[i], "\t", nombres[i], "\t", notas[i])
```

**Todos quedaron con 0.0.** `ordenar_seleccion` recorre hasta `len(notas)`, es decir, las 10 casillas: los seis 0.0 de las casillas libres son los menores y se van al principio. Además, ordenó de menor a mayor, y se quería de mayor a menor.

### ⚠️ Segundo intento: hasta `cantidad`, de mayor a menor, pero solo `notas`

`posicion_mayor_nota` busca la **mayor** nota entre `inicio` y `cantidad - 1`:

```python
def posicion_mayor_nota(notas, inicio, cantidad):
    """
    Supone: 0 <= inicio < cantidad <= len(notas).
    Garantiza: retorna la posición de la mayor nota entre inicio y
    cantidad - 1 (si se repite, la primera). No modifica notas.
    """
    pos_mayor = inicio
    for j in range(inicio + 1, cantidad):
        if notas[j] > notas[pos_mayor]:
            pos_mayor = j
    return pos_mayor
```

### La solución: mover al estudiante completo

Un estudiante ocupa **la misma posición** en los tres arreglos. Para moverlo hay que moverlo en los tres: `intercambiar_estudiantes` llama tres veces a `intercambiar`, una por arreglo, con las mismas posiciones.

```python
def intercambiar_estudiantes(ids, nombres, notas, i, j):
    """
    Supone: i y j son posiciones válidas de los tres arreglos.
    Garantiza: MODIFICA ids, nombres y notas: los estudiantes de las
    posiciones i y j quedan intercambiados en los tres. No retorna nada.
    """
    intercambiar(ids, i, j)
    intercambiar(nombres, i, j)
    intercambiar(notas, i, j)
```

### Mismo método, otro criterio: por nombre

Solo cambia **qué se compara**: los nombres con `<`. Python compara textos letra por letra, como en el diccionario: `"Ana" < "Beto"` es `True`.

```python
def posicion_menor_nombre(nombres, inicio, cantidad):
    """
    Supone: 0 <= inicio < cantidad <= len(nombres). Los nombres empiezan
    con mayúscula y no llevan tilde (si no, < no sigue el orden del
    diccionario).
    Garantiza: retorna la posición del nombre que va primero en orden
    alfabético entre inicio y cantidad - 1. No modifica nombres.
    """
    pos_menor = inicio
    for j in range(inicio + 1, cantidad):
        if nombres[j] < nombres[pos_menor]:
            pos_menor = j
    return pos_menor


def ordenar_por_nombre(ids, nombres, notas, cantidad):
    """
    Ordena el curso alfabéticamente por nombre (selección).

    Supone:
        0 <= cantidad <= len(ids).
    Garantiza:
        MODIFICA ids, nombres y notas: los inscritos quedan en orden
        alfabético, y cada estudiante conserva su Id, nombre y nota.
        Las posiciones libres no cambian. No retorna nada.
    """
    for i in range(cantidad - 1):
        pos_menor = posicion_menor_nombre(nombres, i, cantidad)
        intercambiar_estudiantes(ids, nombres, notas, i, pos_menor)
```

### Pruebas

Hay algo nuevo que probar: que cada estudiante **conserva** su Id, nombre y nota, y que las casillas libres **no se tocan**. Con notas repetidas, cada nombre debe seguir con su nota.

```python
ids = ["000", "123", "987", "007", "", ""]
nombres = ["Ana", "Beto", "Carla", "Dario", "", ""]
notas = [3.0, 4.5, 2.0, 4.0, 0.0, 0.0]
ordenar_por_nota(ids, nombres, notas, 4)
assert notas == [4.5, 4.0, 3.0, 2.0, 0.0, 0.0]          # tipico, y las libres no se tocan
assert nombres == ["Beto", "Dario", "Ana", "Carla", "", ""]
assert ids == ["123", "007", "000", "987", "", ""]
ordenar_por_nombre(ids, nombres, notas, 4)
assert nombres == ["Ana", "Beto", "Carla", "Dario", "", ""]
assert notas == [3.0, 4.5, 2.0, 4.0, 0.0, 0.0]
ids = ["1", "2", "3"]
nombres = ["Ana", "Beto", "Carla"]
notas = [4.0, 3.0, 4.0]
ordenar_por_nota(ids, nombres, notas, 3)
assert notas == [4.0, 4.0, 3.0]                          # especial: notas repetidas
assert nombres == ["Ana", "Carla", "Beto"]
nombres = ["Ana", ""]
notas = [3.0, 0.0]
ordenar_por_nota(["1", ""], nombres, notas, 1)
assert nombres == ["Ana", ""]                            # limite: un solo estudiante
ordenar_por_nota(["", ""], ["", ""], notas, 0)
assert notas == [3.0, 0.0]                               # limite: curso vacio
print("Todas las pruebas pasaron")
```

## Parte 9 — Lo que fue incómodo, y lo que viene

El programa funciona, pero escribirlo dejó cuatro incomodidades:

1. **Un estudiante no existe como tal.** Son tres arreglos que comparten una posición; si se mueve uno sin los otros, los datos se mezclan sin ningún aviso (el segundo intento).
2. **Cada intercambio son tres.** `intercambiar_estudiantes` tiene que acordarse de los tres arreglos.
3. **`cantidad` va a todas partes.** Varias funciones la reciben, y las que la cambian tienen que modificar los arreglos y además retornarla.
4. **Una función de ordenar por cada criterio**, casi idénticas.

¿Qué pasa si mañana se agrega el **correo** de cada estudiante? Un arreglo más, y cambian varias funciones: agregar, eliminar, listar, intercambiar y los dos ordenamientos.

**La próxima clase (Programación Orientada a Objetos)** resuelve el mismo problema, con el mismo menú, definiendo un `Estudiante` que guarda juntos su Id, su nombre y su nota, y un `Curso` que tiene a sus estudiantes. Intercambiar dos estudiantes será un solo intercambio.

![Un estudiante es una sola cosa](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/uml_curso.png)

Y si cada estudiante tiene **varias** notas, en lugar de un arreglo por parcial se usa una tabla: la fila `i` es un estudiante y la columna `j` una evaluación. Eso es una **matriz**, `notas[i][j]`.

![Una tabla de notas](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/12/teoria/images/matriz_notas.png)

## Parte 10 — Lo que Python ya trae: `sorted()` y `.sort()`

> 🐍 **Propio de Python.** Python ya tiene ordenamiento incorporado, y ahora se sabe qué hace por dentro. `sorted(lista)` retorna una lista **nueva** ordenada, y la original no cambia. `lista.sort()` **modifica** la lista y no retorna nada: es el caso 3, como `ordenar_seleccion`. Con `reverse=True`, los dos ordenan de mayor a menor. `.sort()` tiene la forma `objeto.operación()`, que se explica en Programación Orientada a Objetos; en los ejercicios del curso se usan las funciones de esta clase.

```python
notas = [3.0, 4.5, 2.0, 4.0]
ordenada = sorted(notas)     # retorna una lista nueva
print(ordenada)
print(notas)                 # no cambia
notas.sort()                 # modifica la lista, no retorna nada
print(notas)
print(sorted(notas, reverse=True))
```

| Búsqueda lineal | Revisar uno por uno | No | — | `N` comparaciones |
| Selección | Buscar el menor y ponerlo al inicio | — | `posicion_menor`, `intercambiar` | `N (N - 1) / 2` comparaciones, siempre |
| Burbuja | Comparar vecinos; el mayor sube al final | — | `intercambiar` | `N (N - 1) / 2`; termina antes si ya está ordenada |
| Inserción | Meter cada elemento en la parte ordenada | — | ninguna: desplaza | `N (N - 1) / 2`; la que menos trabaja con listas casi ordenadas |
| Búsqueda binaria | Partir a la mitad y descartar | Sí (*Supone*) | — | cerca de log₂(`N`) comparaciones |

* Los tres métodos de ordenamiento son del **caso 3**: modifican la lista y no retornan nada. Se prueban llamándolos y revisando la lista después.
* En arreglos paralelos, un estudiante ocupa la misma posición en varios arreglos: hay que moverlos **juntos**, y recorrer solo **hasta `cantidad`**.
