## Repaso: definir e invocar una función

Una **función** es un bloque de código con nombre que realiza una tarea específica. Trabajar con una función tiene dos momentos:

**1. Definirla**, con la palabra `def`:

```python
def nombre_funcion(parametro1, parametro2):
    cuerpo de la funcion
    return resultado          # opcional
```

* Los **parámetros** son las entradas de la función. Pueden ser varios o ninguno.
* El **cuerpo** va indentado y hace el trabajo.
* `return` entrega el resultado a quien llamó la función. Si la función solo imprime, no necesita `return`.

> **Importante:** definir una función **no la ejecuta**. Solo le dice a Python cómo hacer la tarea.

**2. Invocarla** (llamarla), escribiendo su nombre con los **argumentos** entre paréntesis:

```python
variable = nombre_funcion(argumento1, argumento2)
```

Los argumentos son los valores que se envían, y se asocian a los parámetros **por posición**: el primer argumento al primer parámetro, el segundo al segundo, y así sucesivamente.

Una función se puede representar como una **caja negra**: se conocen sus entradas y su salida, sin necesidad de ver qué pasa adentro.

```
          ┌────────────┐
   c ───► │   c_to_f   │ ───► f
          └────────────┘
```

## Parte 1 — Conversión de temperatura

### Enunciado

Convertir una temperatura de grados Celsius (°C) a grados Fahrenheit (°F) con la fórmula:

$$F = \frac{9}{5}\,C + 32$$

### Paso 1: sin función

El programa lee la temperatura, aplica la fórmula y muestra el resultado:

````python
# Entrada de datos
c = float(input("Ingrese los °C: "))

# Conversion °C a °F
f = (9/5)*c + 32

# Salida
print(c,"°C =",f,"°F")
````

Ejemplo de ejecución:

```
Ingrese los °C: 10
10.0 °C = 50.0 °F
```

El programa funciona, pero ¿qué pasa si se necesita convertir **varias** temperaturas en distintas partes de un programa? Sin función, hay que **copiar la fórmula** cada vez:

### 🔍 Para entender: el código se repite

````python
c1 = 5
f1 = (9/5)*c1 + 32
print(c1, "°C =", f1, "°F")

c2 = 1
f2 = (9/5)*c2 + 32
print(c2, "°C =", f2, "°F")

c3 = 100
f3 = (9/5)*c3 + 32
print(c3, "°C =", f3, "°F")
````

La fórmula `(9/5)*c + 32` está escrita **tres veces**. Si estuviera mal (por ejemplo, `(5/9)` en lugar de `(9/5)`), habría que encontrar y corregir **cada copia**. Además, al leer `(9/5)*c1 + 32` hay que reconocer la fórmula para saber qué hace.

### Paso 2: definir la función

En clase, la fórmula se convirtió en la función `c_to_f`. Ejecute la celda y observe que **no imprime nada**: solo define la función.

````python
def c_to_f(c):
    f = (9/5)*c + 32
    return f
````

Las partes de la definición:

| Parte | En `c_to_f` | Qué es |
|---|---|---|
| Nombre | `c_to_f` | Con este nombre se invoca la función |
| Parámetro | `c` | La temperatura en °C que la función **recibe** |
| Cuerpo | `f = (9/5)*c + 32` | El cálculo |
| Retorno | `return f` | La temperatura en °F que la función **devuelve** |

### Paso 3: invocar la función

Ahora sí se usa. En clase se invocó dos veces, con datos distintos:

````python
# Invocaciones
T1 = c_to_f(5) # c = 5
T2 = c_to_f(1) # c = 1
print(f"T1 = 5°C = {T1}°F")
print(f"T2 = 1°C = {T2}°F")
````

> **Nota:** en el notebook de la clase, el comentario de la segunda invocación decía `# c = 5` y el segundo `print` decía `T1`. Aquí se corrigieron: en esa llamada `c` vale 1, y el resultado se guarda en `T2`.

¿Qué pasa en `T1 = c_to_f(5)`?

1. La ejecución **salta** a la función, y el parámetro `c` toma el valor del argumento: `c = 5`.
2. Se ejecuta el cuerpo: `f = (9/5)*5 + 32 = 41.0`.
3. `return f` **devuelve** 41.0 y la ejecución **vuelve** al punto de la llamada.
4. El valor devuelto se guarda en `T1`.

Puede ver este recorrido paso a paso en esta [simulación en Python Tutor](https://pythontutor.com/visualize.html#code=def%20c_to_f%28c%29%3A%0A%20%20%20%20f%20%3D%20%289/5%29*c%20%2B%2032%0A%20%20%20%20return%20f%0A%0A%23%20Invocaciones%0AT1%20%3D%20c_to_f%285%29%20%23%20c%20%3D%205%0AT1c%20%3D%201%0AT2%20%3D%20c_to_f%28T1c%29%20%23%20c%20%3D%205%0Aprint%28f%22T1%20%3D%205%C2%B0C%20%3D%20%7BT1%7D%C2%B0F%22%29%0Aprint%28f%22T1%20%3D%20%7BT1c%7D%C2%B0C%20%3D%20%7BT2%7D%C2%B0F%22%29%0A&curInstr=0&mode=display&origin=opt-frontend.js&py=311) (la usada en clase). Avance con el botón *Next* y observe cómo aparece el marco de `c_to_f` cuando se llama la función y desaparece cuando retorna.

### 🔍 Para entender: formas de invocar una función

El argumento puede ser un número, una variable o una expresión, y el resultado se puede guardar en una variable o usar directamente:

````python
temp = 30
print(c_to_f(100))        # argumento: un numero; el resultado se usa directamente
print(c_to_f(temp))       # argumento: una variable (c toma el valor de temp)
print(c_to_f(temp - 50))  # argumento: una expresion (se calcula antes de llamar)
````

En la tercera llamada, Python calcula primero `temp - 50`, que da `-20`, y ese valor es el que recibe `c`.

### Una segunda función: Celsius a Kelvin

Con el mismo procedimiento se definió `c_to_k`, que convierte a Kelvin con la fórmula $K = C + 273.15$. Esta vez los argumentos son **variables**:

````python
# Definicion
def c_to_k(c):
    k = c + 273.15
    return k

# Invocacion
# 1. T1c = 10
T1c = 10
T1k = c_to_k(T1c)
# 2. T2c = -20
T2c = -20 # -10 - 10
T2k = c_to_k(T2c)
# Despliegue de resultados
print(f"T1c = {T1c}°C = {T1k}°K")
print(f"T2c = {T2c}°C = {T2k:.2f}°K")
````

> **Nota:** en el notebook de la clase, los dos `print` mostraban el resultado con la unidad `°F`, aunque el valor está en Kelvin. Aquí se corrigió a `°K`, la etiqueta que usa el programa de la clase. (El Sistema Internacional escribe la unidad `K`, sin el símbolo de grado; esto se ajusta en la clase 10).

Observe que la variable que se envía se llama `T1c`, pero dentro de la función el parámetro se llama `c`. **No tienen que llamarse igual**: al invocar `c_to_k(T1c)`, el parámetro `c` recibe el **valor** de `T1c`.

### Paso 4: ¿por qué conviene la función?

| | Sin función | Con función |
|---|---|---|
| Convertir tres temperaturas | La fórmula se escribe 3 veces | La fórmula se escribe 1 vez; hay 3 llamadas |
| Si la fórmula tiene un error | Se corrige en cada copia | Se corrige en un solo lugar |
| Al leer el código | `(9/5)*c + 32`: hay que reconocer la fórmula | `c_to_f(c)`: el nombre dice qué hace |
| Usarla en otro programa | Hay que volver a escribir la fórmula | Se usa la misma función, que ya fue probada |

Estas son las ventajas de la modularización que se vieron en la teoría: **reutilización** (escribir una vez, usar muchas), **confiabilidad** (lo que ya se probó no se vuelve a escribir) y **abstracción** (quien usa `c_to_f` no necesita ver la fórmula).

## Parte 2 — Funciones con ciclos anidados

Ahora se aplica la misma idea a los ciclos anidados de la [sesión 22](../../8/sesion_magistral-22/README.md). Allí, cada programa dibujaba **una sola figura**, de un tamaño fijo. Para dibujar otra, había que copiar el ciclo completo.

Estas funciones son distintas de `c_to_f`: **reciben** datos (el tamaño, el símbolo) pero **no devuelven** ningún valor. Su trabajo es **imprimir**. Por eso no tienen `return`.

### Ejemplo 1: el rectángulo de caritas (hecho en clase)

**Sin función.** Este es el programa de la sesión 22 (aquí con `N = 3` y `M = 4`). Dibuja un solo rectángulo:

````python
carita = "^_^"

N = 3
M = 4

for i in range(N):
    # print(i, end = ": ")
    for j in range(M):
        # print(f"i = {i}, j = {j}")
        print(carita, end = "   ")
    print()
````

Para dibujar **otro** rectángulo, de otro tamaño y con otro símbolo, hay que copiar los dos ciclos y cambiar los valores:

````python
for i in range(2):
    for j in range(3):
        print(":)", end = "   ")
    print()
print()
for i in range(4):
    for j in range(4):
        print(":*", end = "   ")
    print()
````

**Con función.** En clase, los ciclos se convirtieron en la función `imprimir_caritas`. Lo que cambiaba de un rectángulo a otro (las filas, las columnas y el símbolo) se convirtió en **parámetros**:

````python
def imprimir_caritas(f, c, simbolo):
    # code
    for i in range(f):
        for j in range(c):
            print(simbolo, end = "   ")
        print()

imprimir_caritas(2,3, ":)")
print()
imprimir_caritas(4,4, ":*")
````

Los mismos dos rectángulos, ahora con **una línea cada uno**. Así se asocian los argumentos a los parámetros, por posición:

| Llamada | `f` | `c` | `simbolo` | Resultado |
|---|---|---|---|---|
| `imprimir_caritas(2, 3, ":)")` | 2 | 3 | `":)"` | 2 filas de 3 `:)` |
| `imprimir_caritas(4, 4, ":*")` | 4 | 4 | `":*"` | 4 filas de 4 `:*` |

El `print()` que hay entre las dos llamadas está **fuera** de la función: solo deja una línea en blanco para separar los dos rectángulos.

### 🔍 Para entender: ¿qué devuelve una función sin `return`?

`imprimir_caritas` no tiene `return`. Si se intenta guardar su resultado en una variable, se obtiene `None`, que en Python significa "ningún valor":

````python
resultado = imprimir_caritas(1, 2, "*")
print("La funcion devolvio:", resultado)
````

La función **sí hizo su trabajo** (imprimió `*   *`), pero no devolvió nada. Por eso a estas funciones se las llama sin guardar el resultado: `imprimir_caritas(1, 2, "*")`.

| | `c_to_f(c)` | `imprimir_caritas(f, c, simbolo)` |
|---|---|---|
| ¿Recibe datos? | Sí: la temperatura | Sí: filas, columnas y símbolo |
| ¿Devuelve un valor? | Sí, con `return` | No: imprime |
| Cómo se invoca | `T1 = c_to_f(5)` | `imprimir_caritas(2, 3, ":)")` |

### Ejemplo 2: la serie `1 2 2 3 3 3 ...`

En clase quedó **sin función**. Aquí se completa con el mismo procedimiento.

**Sin función.** El programa de la sesión 22 imprime la serie para un solo valor de `N`:

````python
N = 4
for i in range(1,N + 1):
    # print("i = " + str(i))
    for j in range(i):
        # print("   j = " + str(j))
        print(i, end = " ")
````

**Con función.** ¿Qué cambia de una serie a otra? Solo `N`. Entonces `N` se convierte en el **parámetro** `n`:

````python
def imprimir_serie(n):
    for i in range(1, n + 1):
        for j in range(i):
            print(i, end = " ")
    print()
````

Además de los dos ciclos, la función termina con `print()`. La serie se imprime en una sola línea (por el `end = " "`), y ese `print()` la **cierra**. Así, si se invoca la función varias veces, cada serie queda en su propia línea:

````python
imprimir_serie(4)
imprimir_serie(2)
imprimir_serie(6)
````

### Ejemplo 3: el triángulo

**Sin función.** La mejora del ejemplo anterior, de la sesión 22, imprime cada número en su propia fila. En clase, la versión final leía `N` con `input()`:

````python
N = int(input("Digite el numero: "))
for i in range(1,N + 1):
    # print("i = " + str(i))
    for j in range(i):
        # print("   j = " + str(j))
        print(i, end = " ")
    print()
````

Ejemplo de ejecución:

```
Digite el numero: 4
1
2 2
3 3 3
4 4 4 4
```

**Con función.** El dibujo del triángulo pasa a la función. La lectura del dato **no**: `input()` se queda en el programa principal, y la función solo recibe `n` y dibuja. Así, la misma función sirve tanto si `n` lo escribe el usuario como si viene de otra parte del programa.

````python
def imprimir_triangulo(n):
    for i in range(1, n + 1):
        for j in range(i):
            print(i, end = " ")
        print()
````

````python
imprimir_triangulo(3)
print()
imprimir_triangulo(5)
````

Y el programa principal de la versión con `input()` queda en **dos líneas**: leer `N` e invocar la función.

````python
N = int(input("Digite el numero: "))
imprimir_triangulo(N)
````

Si escribe `4`, el resultado es el mismo triángulo del ejemplo de ejecución de arriba.

### Ejemplo 4: la escalera de Mario

**Sin función.** El programa de la sesión 22 dibuja una escalera de 10 escalones:

````python
N = 10
espacios = N - 1
numerales = 1

# Filas (i)
for i in range(N):
    # Espacios
    for j in range(espacios):
        print(" ", end = "")
    # Numerales
    for j in range(numerales):
        print("#", end = "")
    print() # Nueva linea
    # Actualizacion
    espacios -= 1
    numerales += 1
````

**Con función.** El número de escalones se convierte en el parámetro `n`. Las variables `espacios` y `numerales` se crean **dentro** de la función, así que cada invocación empieza con sus valores iniciales, calculados a partir de su propio `n`:

````python
def imprimir_escalera(n):
    espacios = n - 1
    numerales = 1
    for i in range(n):
        for j in range(espacios):
            print(" ", end = "")
        for j in range(numerales):
            print("#", end = "")
        print()
        espacios -= 1
        numerales += 1
````

````python
imprimir_escalera(3)
print()
imprimir_escalera(6)
````

### ¿Por qué conviene la función? (Parte 2)

| | Sin función | Con función |
|---|---|---|
| Dibujar 3 figuras de tamaños distintos | Se copian los ciclos 3 veces y se cambian los números | Una definición y 3 llamadas de una línea |
| Si el dibujo tiene un error | Se corrige en cada copia | Se corrige dentro de la función, y todas las llamadas quedan bien |
| Al leer el programa | Hay que seguir los ciclos para saber qué dibujan | `imprimir_escalera(6)` dice qué hace |

## Parte 3 — 🔍 La misma función, hecha de dos formas

En la sesión 22, la serie se escribió de **tres formas** que dan el mismo resultado, y el rectángulo, con `for` y con `while`. Con funciones, esas formas distintas pueden quedar **por dentro**, sin que cambie la manera de usarlas.

La celda siguiente vuelve a definir `imprimir_serie`, ahora con la forma C de la sesión 22 (con un contador `num`). La invocación es exactamente la misma de antes:

````python
def imprimir_serie(n):
    num = 1
    for i in range(n):
        for j in range(i + 1):
            print(num, end = " ")
        num += 1
    print()

imprimir_serie(4)
````

Lo mismo con el rectángulo, ahora con `while`:

````python
def imprimir_caritas(f, c, simbolo):
    i = 0
    while i < f:
        j = 0
        while j < c:
            print(simbolo, end = "   ")
            j += 1
        i += 1
        print()

imprimir_caritas(2,3, ":)")
````

Las salidas son idénticas a las de la Parte 2. Quien **usa** la función solo necesita saber su nombre, qué parámetros recibe y qué hace; no necesita saber si por dentro hay un `for` o un `while`. Esto es la **caja negra**: se puede cambiar el *cómo* sin cambiar el *qué*.

> **Nota:** al ejecutar estas celdas, las versiones nuevas **reemplazan** a las anteriores, porque tienen el mismo nombre. Si vuelve a ejecutar las celdas de la Parte 2, se usan las versiones de la Parte 2.

1. **Definir** una función (`def`) no la ejecuta. Se ejecuta al **invocarla**.
2. Al invocarla, cada argumento se asocia al parámetro de la misma posición.
3. Lo que **cambia** de un uso a otro (la temperatura, el tamaño, el símbolo) se convierte en **parámetro**.
4. Una función se escribe **una vez** y se **reutiliza** cuantas veces se necesite. Si tiene un error, se corrige en un solo lugar.
5. Quien usa una función no necesita saber cómo está hecha por dentro.
