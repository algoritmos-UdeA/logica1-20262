````python
# Definicion
def c_to_k(c):
    k = c + 273.15
    return k

# Invocacion
# 1. T1c = 10 
T1c = 10 
T1k = c_to_k(T1c)
# 2. T2c = -20
T2c = -20 # -10 - 10
T2k = c_to_k(T2c)
# Despliegue de rtesultados
print(f"T1c = {T1c}°C = {T1k}°F")
print(f"T2c = {T2c}°C = {T2k:.2f}°F")
````

### Ejemplo 2

Hacer un rectangulo que imprima `NxM` caritas felices.


````python
carita = "^_^"

N = 3
M = 4

for i in range(N):
    # print(i, end = ": ")
    for j in range(M):
        # print(f"i = {i}, j = {j}")
        print(carita, end = "   ")
    print()
# print(total)
````

````python
carita = "^_^"

N = 2
M = 3

i = 0 
j = 0
total = 0
while i < N:
    # print(i, end = ": ")
    while j < M:
        # print(f"i = {i}, j = {j}")
        print(carita, end = "   ")
        j += 1
        total += 1
    i += 1 
    j = 0 # No olvidar inicia
    print()
# print(total)
````

Funcion imprimir caritas

````python
def imprimir_caritas(f, c, simbolo):
    # code
    for i in range(f):
        for j in range(c):
            print(simbolo, end = "   ")
        print()

imprimir_caritas(2,3, ":)")
print()
imprimir_caritas(4,4, ":*")
````

### Ejemplo 4

Hacer un programa que muestre la escalera de Supermario

````python
N = 10
espacios = N - 1
numerales = 1

# Filas (i)
for i in range(N):
    # Espacios
    for j in range(espacios):
        print(" ", end = "")
    # Numerales
    for j in range(numerales):
        print("#", end = "")
    print() # Nueva linea
    # Actualizacion
    espacios -= 1
    numerales += 1
````
