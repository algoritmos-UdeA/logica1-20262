# Funciones en Python: de programa a función reutilizable

En la clase 8 se resolvieron estos cinco ejercicios como programas completos: cada uno lee los datos con `input()`, hace el cálculo y muestra el resultado con `print()`. Los programas funcionan, pero el cálculo queda "amarrado" a ellos. Si otro programa necesita, por ejemplo, el factorial de un número, hay que copiar y pegar el ciclo.

En este notebook cada ejercicio se convierte en una **función**: un bloque de código con nombre que recibe datos (**parámetros**), hace **una sola tarea** y entrega un resultado (`return`). Una función así se puede **probar** por separado y **reutilizar** en otros programas. Por ejemplo, el ejercicio 3 (número primo) reutiliza la función del ejercicio 2, y el ejercicio 5 ($e^x$) reutiliza la del ejercicio 1.

## Cómo usar este notebook

- Ejecute las celdas **en orden**, de arriba hacia abajo. Una función solo existe para Python después de ejecutar la celda donde se define.
- Cada ejercicio tiene cinco partes: la versión sin funciones (de la clase 8), el análisis, la función, las pruebas con `assert` y el programa principal.
- El **ejercicio 1 está resuelto completo** y sirve de modelo. En los siguientes ejercicios usted completa el código, y las ayudas van disminuyendo.
- Las celdas de pruebas **fallan mientras la función no esté completa**. Es normal. Así se interpreta el resultado:

| Lo que aparece al ejecutar las pruebas | Qué significa |
|---|---|
| `Todas las pruebas pasaron` | La función devuelve lo esperado en todos los casos. |
| `AssertionError` | La función existe, pero en algún caso devuelve un valor distinto del esperado. El mensaje señala la línea del `assert` que falló. |
| `NameError: name '...' is not defined` | No ha ejecutado la celda donde se define la función, o el nombre está mal escrito. |
| `TypeError` que menciona `NoneType` | La función todavía no devuelve nada: tiene solo el `pass` o le falta el `return`. Sin `return`, una función devuelve `None`. |

- Al terminar, use **Kernel → Restart Kernel and Run All Cells** para comprobar que todo funciona desde cero, en orden.

## Receta: de programa a función

Para convertir un programa en una función, responda estas preguntas en orden:

| Paso | Pregunta | Se convierte en |
|---|---|---|
| 1. Entradas | ¿Qué datos lee el programa con `input()`? | Los **parámetros** de la función. |
| 2. Salida | ¿Cuál es el resultado del cálculo? | El valor que se devuelve con `return`. |
| 3. Proceso | ¿Qué instrucciones hacen el cálculo? | El **cuerpo** de la función, sin `input()` ni `print()`. |
| 4. Pruebas | ¿Qué debe devolver la función con datos conocidos? | Los `assert`: casos típicos, límite y especiales. |
| 5. Programa principal | ¿Cómo se usa la función? | Leer los datos → llamar a la función → mostrar el resultado. |

**Importante.** La función no debe tener `input()` ni `print()`. Si los tiene, deja de ser reutilizable por dos razones:

1. El programa que la use no puede darle los datos de otra forma, por ejemplo con el resultado de otro cálculo: la función siempre se detiene a preguntarle al usuario.
2. No se puede probar con `assert`, porque `assert` solo compara el valor que la función **devuelve**, no lo que muestra en pantalla.

Es la **cohesión funcional** vista en la teoría: una función, una tarea. Leer los datos y mostrar los resultados son tareas del programa principal.

## Ejercicio 1: factorial (resuelto como modelo)

Haga un programa que calcule el factorial de un número.

### Versión sin funciones (clase 8)

````python
n = int(input("Ingrese un número entero positivo: "))

fac = 1
for i in range(1, n + 1):
    fac *= i

print(f"El factorial de {n} es: {fac}")
````

### Pasos 1 y 2: entradas y salida

| | En el programa original | En la función |
|---|---|---|
| Entrada | `n = int(input(...))` | Parámetro `n` |
| Salida | `print(f"El factorial de {n} es: {fac}")` | `return fac` |

Diagrama de caja negra (solo entradas y salidas, sin los detalles internos):

```
  n  ──▶  [ factorial ]  ──▶  n!
```

Recibe un dato y devuelve un valor: es una función del **caso 3** de la teoría. El nombre `factorial` dice qué devuelve.

### Paso 3: la función

Compare con la versión original. La inicialización y el ciclo quedan **iguales**; el `input()` y el `print()` **desaparecen**, y se agrega `return fac` al final.

````python
def factorial(n):
    fac = 1
    for i in range(1, n + 1):
        fac *= i
    return fac
````

Al ejecutar la celda anterior no aparece nada en pantalla, y es lo esperado: **definir una función no la ejecuta**. Solo se ejecuta cuando se la llama.

### Paso 4: pruebas

| Tipo | Llamada | Resultado esperado | Por qué |
|---|---|---|---|
| Típico | `factorial(5)` | `120` | $5 \cdot 4 \cdot 3 \cdot 2 \cdot 1$ |
| Típico | `factorial(3)` | `6` | $3 \cdot 2 \cdot 1$ |
| Límite | `factorial(1)` | `1` | El ciclo da una sola vuelta. |
| Límite | `factorial(0)` | `1` | Por definición, $0! = 1$. El `range(1, 1)` está vacío, así que `fac` se queda en `1`. |
| Especial | `factorial(-3)` | `1` | El factorial de un negativo no existe, pero la función no lo rechaza: el `range` queda vacío y devuelve `1`. |

El último caso no es un error de la prueba: **documenta** un comportamiento de la función que hay que tener en cuenta. Por eso el programa principal (paso 5) debe validar que el número no sea negativo antes de llamarla.

````python
# Casos típicos
assert factorial(5) == 120
assert factorial(3) == 6

# Casos límite
assert factorial(1) == 1
assert factorial(0) == 1

# Caso especial: la función no rechaza los negativos
assert factorial(-3) == 1

print("Todas las pruebas pasaron")
````

### Paso 5: programa principal

El programa principal lee el dato, **valida** que tenga sentido, llama a la función y muestra el resultado. La validación va aquí y se hace con un condicional, no con `assert`: `assert` sirve para probar el programa, no para revisar lo que escribe el usuario.

````python
n = int(input("Ingrese un número entero positivo: "))

if n < 0:
    print("El número no puede ser negativo")
else:
    print(f"El factorial de {n} es: {factorial(n)}")
````

### ¿Qué se ganó?

El programa muestra lo mismo que la versión original, pero ahora `factorial()` se puede usar en **cualquier** cálculo, sin copiar el ciclo. Por ejemplo, la cantidad de formas de escoger 2 estudiantes de un grupo de 5 es $\dfrac{5!}{2! \cdot 3!}$:

````python
formas = factorial(5) // (factorial(2) * factorial(3))
print(f"Formas de escoger 2 estudiantes de 5: {formas}")
````

Sin funciones, esa línea habría necesitado **tres** ciclos iguales. La función `factorial()` se volverá a usar en el ejercicio 5.

$$e^x \approx \sum_{n=0}^{N} \frac{x^n}{n!} = 1 + x + \frac{x^2}{2!} + \frac{x^3}{3!} + \cdots + \frac{x^N}{N!}$$


### Versión sin funciones (clase 8)

````python
# Inicializacion
suma = 0 # Aproximacion de e^x

# Entradas
N = int(input("Indique el valor de N (exponente del ultimo termino): "))
x = float(input("Indique el valor de x: "))

# Proceso
for n in range(0, N + 1):
    # Calculo del factorial de n
    fact = 1
    for j in range(1, n + 1):
        fact *= j

    # Calculo del termino n-esimo
    term = (x**n)/fact

    # Actualizacion de la suma
    suma += term

# Salidas
print(f"e^{x} aproximado con N = {N} es: {suma:.6f}")
````
