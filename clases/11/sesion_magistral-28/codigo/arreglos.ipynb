## Parte 1 — ¿Por qué hace falta un arreglo? (diapositivas 8-10)

La clase 10 terminó con un problema que no se podía resolver bien:

> Leer las notas de los estudiantes de un curso y mostrar cuántas notas son mayores que el promedio.

El promedio solo se conoce **después** de leer todas las notas. Pero si cada nota se lee en la misma variable `nota`, al terminar solo queda la última: las anteriores se perdieron, y ya no se pueden comparar con el promedio.

### 📝 En el tablero: una variable por nota

Una salida es usar una variable distinta para cada nota. Con 4 estudiantes funciona (aquí las notas se escriben directamente en lugar de leerlas con `input()`):

````python
nota1 = 3.0
nota2 = 4.5
nota3 = 2.0
nota4 = 4.0
promedio = (nota1 + nota2 + nota3 + nota4) / 4
cantidad = 0
if nota1 > promedio:
    cantidad = cantidad + 1
if nota2 > promedio:
    cantidad = cantidad + 1
if nota3 > promedio:
    cantidad = cantidad + 1
if nota4 > promedio:
    cantidad = cantidad + 1
print("Notas mayores que el promedio:", cantidad)
````

Con 40 estudiantes serían 40 variables, una suma de 40 términos y 40 `if`. Y **no se puede usar un ciclo**, porque cada variable tiene un nombre distinto: no hay forma de decir "la variable número `i`". Lo mismo le pasa al señor Burns si quiere guardar la identificación de cada empleado de la planta (`id_lenny`, `id_carl`, `id_homero`…): cada empleado nuevo obliga a crear una variable nueva y a cambiar el programa.

La pregunta es: **¿existe una forma de guardar muchos valores bajo un solo nombre?** La respuesta es el arreglo.

## Parte 2 — La lista usada como arreglo (diapositivas 11-14)

Un **arreglo** es una variable que guarda muchos valores bajo un solo nombre:

* Todos los valores son del **mismo tipo** (por ejemplo, todos reales).
* Tiene un **tamaño fijo**: se decide al crearlo y no cambia.
* A cada valor se llega por su **posición** o **índice**. Las posiciones empiezan en **0**: si el arreglo tiene `N` elementos, la última posición es `N - 1`.

![El arreglo notas con sus posiciones](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/11/teoria/images/partes_arreglo.png)

En Python, un arreglo se escribe con una **lista**: los valores entre corchetes, separados por comas. Una lista puede hacer más cosas que un arreglo, pero en este curso se usa **como arreglo**, con sus reglas: tamaño fijo, un solo tipo de dato y acceso por posición.

````python
notas = [3.0, 2.5, 1.5, 4.0]
print(type(notas))
nota = 3
print(type(nota))
````

`type()` dice de qué tipo es un dato. `notas` es una **lista** (`list`): una sola variable con cuatro valores. `nota` es un entero (`int`): una variable con un solo valor.

````python
print(notas)
````

Un índice **negativo** cuenta desde el final: `notas[-1]` es el último elemento, `notas[-2]` el penúltimo, y así. Es lo mismo que `notas[N - 1]`, con `N = 4`.

> 📏 **Convención del curso.** Los índices negativos se pueden usar, por ejemplo `notas[-1]` para el último. En el pseudocódigo no existen: allí el último es `notas[N - 1]`. Pero hay que usarlos con cuidado, como muestra la celda siguiente.

### 🔍 Para entender: un índice negativo no da error

Suponga que, dentro de un recorrido, se quiere comparar cada nota con la **anterior**, `notas[i - 1]`. En la primera vuelta, `i` vale 0:

````python
i = 0
print(notas[i - 1])
````

`notas[0 - 1]` es `notas[-1]`: Python devuelve **la última** nota sin avisar, aunque la nota "anterior" a la primera no existe. El programa sigue con un dato equivocado y el error queda escondido. En cambio, un índice demasiado grande sí produce un error (se verá en la Parte 9).

`append` agrega un elemento **al final** de la lista: ahora `notas` tiene 5 elementos. Observe que se agregó `5`, un entero, a una lista de reales.

> 📏 **Convención del curso.** En este curso la lista se usa como arreglo, con **tamaño fijo**, así que no se usa `append` (ni `remove` ni otras operaciones de las listas). `notas.append(5)` tiene la forma `objeto.operación()`: en Python todos los datos son objetos, y esa forma se explica en Programación Orientada a Objetos. Además, todos los elementos deben ser del **mismo tipo**: si las notas son reales, se escribe `5.0`, no `5`.

> ⚠️ Si ejecuta esta celda dos veces, el `5` se agrega dos veces: cada ejecución cambia la lista que ya existe.

## Parte 3 — Crear un arreglo, guardar y leer valores (diapositivas 15-16)

Cuando todavía no se conocen los valores, el arreglo se crea con un valor inicial repetido en todas sus casillas. En pseudocódigo se escribe `notas = Arreglo(N)`; en Python, `notas = [0.0] * N`: el operador `*` repite el valor y forma una lista con `N` casillas.

| Qué se hace | Pseudocódigo | Python |
|---|---|---|
| Crear el arreglo con `N` casillas | `notas = Arreglo(N)` | `notas = [0.0] * N` |
| Crear e inicializar a la vez | `notas = [3.0, 4.5, 2.0, 4.0]` | `notas = [3.0, 4.5, 2.0, 4.0]` |
| Guardar en una casilla | `notas[0] = 3.0` | `notas[0] = 3.0` |

````python
# Notas de 20 estudiantes de logica 1

notas_L1 = 20*[0.0]
print(notas_L1)
````

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=%23%20Notas%20de%2020%20estudiantes%20de%20logica%201%0A%0Anotas_L1%20%3D%2020%2A%5B0.0%5D%0Aprint%28notas_L1%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

En clase se escribió `20*[0.0]`, que da lo mismo que `[0.0] * 20`: 20 casillas, todas en `0.0`, una por cada estudiante de Lógica 1.

Cada casilla se usa como una variable normal: se le asigna un valor con `=`, escribiendo su posición entre corchetes.

````python
notas_L1[0] = 5
notas_L1[19] = 3.5
print(notas_L1)
````

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=%23%20Notas%20de%2020%20estudiantes%20de%20logica%201%0A%0Anotas_L1%20%3D%2020%2A%5B0.0%5D%0Aprint%28notas_L1%29%0A%0Anotas_L1%5B0%5D%20%3D%205%0Anotas_L1%5B19%5D%20%3D%203.5%0Aprint%28notas_L1%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false) (el enlace incluye también el código de las celdas anteriores que esta celda necesita)

Con 20 casillas, las posiciones van de 0 a 19: `notas_L1[19]` es la **última** (`N - 1`). Cambiar una casilla no cambia las demás.

Observe que la primera casilla muestra `5` y no `5.0`: se guardó un entero. Para respetar la regla de un solo tipo, se escribe `notas_L1[0] = 5.0`.

````python
notas_D1 = 20*[1.0]
print(notas_D1)
````

El valor inicial puede ser cualquiera. Lo importante es que cada casilla tenga un valor desde el principio.

````python
len(notas_D1)
````

## Parte 4 — El índice puede ser una expresión (diapositiva 17)

El índice no tiene que ser un número escrito: puede ser cualquier expresión entera, como `notas[i]`, `notas[i + 1]`, `notas[N - 1]`, e incluso el valor de **otro** arreglo.

### 📝 En el tablero: ejercicio

¿Cómo queda el arreglo `B` después de las siguientes instrucciones? Estos son los dos arreglos al comenzar:

![Los arreglos A y B antes de las instrucciones](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/clases/11/teoria/images/ejercicio_indices_inicial.png)

> 🔮 **Resuélvalo en papel antes de ejecutar la celda.** Para cada instrucción, calcule primero el índice (lo que está entre corchetes) y después el valor. Por ejemplo, en `B[A[3]] = ...`, primero se busca `A[3]`, que es 0: la instrucción guarda en `B[0]`.

````python
A = [2, 5, -1, 0, 7, 10, 17, -3]
B = [0, 1, 15, -8, 5, 7, -2, 0]
B[2] = A[0] * 3
B[3] = B[3] + A[4]
B[A[3]] = A[5] // B[1]
B[A[6] % A[5]] = A[A[7] + A[5]] + B[2] * -3
print(B)
````

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=A%20%3D%20%5B2%2C%205%2C%20-1%2C%200%2C%207%2C%2010%2C%2017%2C%20-3%5D%0AB%20%3D%20%5B0%2C%201%2C%2015%2C%20-8%2C%205%2C%207%2C%20-2%2C%200%5D%0AB%5B2%5D%20%3D%20A%5B0%5D%20%2A%203%0AB%5B3%5D%20%3D%20B%5B3%5D%20%2B%20A%5B4%5D%0AB%5BA%5B3%5D%5D%20%3D%20A%5B5%5D%20%2F%2F%20B%5B1%5D%0AB%5BA%5B6%5D%20%25%20A%5B5%5D%5D%20%3D%20A%5BA%5B7%5D%20%2B%20A%5B5%5D%5D%20%2B%20B%5B2%5D%20%2A%20-3%0Aprint%28B%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

Instrucción por instrucción:

| Instrucción | Índice | Valor | Casilla que cambia |
|---|---|---|---|
| `B[2] = A[0] * 3` | 2 | `2 * 3 = 6` | `B[2] = 6` |
| `B[3] = B[3] + A[4]` | 3 | `-8 + 7 = -1` | `B[3] = -1` |
| `B[A[3]] = A[5] // B[1]` | `A[3] = 0` | `10 // 1 = 10` | `B[0] = 10` |
| `B[A[6] % A[5]] = A[A[7] + A[5]] + B[2] * -3` | `17 % 10 = 7` | `A[-3 + 10] + 6 * -3 = A[7] - 18 = -3 - 18 = -21` | `B[7] = -21` |

En la última instrucción, `B[2]` ya vale 6, porque la primera instrucción lo cambió: el orden importa.

## Parte 5 — El índice es una variable: llenar un arreglo (diapositivas 18-19)

Con índices fijos se escribe una línea por casilla: con 40 estudiantes, 40 líneas. Si el índice es una **variable** dentro de un ciclo, una sola línea sirve para cualquier `N`:

| Pseudocódigo | Python |

> "El verdadero beneficio de utilizar un arreglo está en su capacidad para usar una variable como un subíndice." (Farrell, cap. 6)

Ese es el patrón de todo lo que sigue. **Llenar** un arreglo es darle un valor a cada casilla con un ciclo, y ese valor puede venir de leer o de calcular.

## Parte 7 — Recorrer y acumular (diapositiva 21)

Dentro del recorrido se puede actualizar un **acumulador**, como en la clase 7. Aquí se suman las notas para calcular el promedio:

````python
notas = [3.0, 2.5, 1.5, 4.0]
suma = 0 
TAM = len(notas)
for i in range(TAM):
    suma += notas[i]
print(suma)
print(suma/4)
````

Esta celda vuelve a crear `notas` con los datos del comienzo de la clase. `TAM = len(notas)` guarda el tamaño, y el ciclo recorre las posiciones de 0 a `TAM - 1`. `suma += notas[i]` es lo mismo que `suma = suma + notas[i]`.

Observe la última línea: el promedio se calcula con `suma/4`, con el 4 escrito a mano, aunque el tamaño ya estaba en `TAM`. Funciona porque la lista tiene 4 notas, pero si se agrega una nota, el promedio queda mal. Es mejor dividir entre la variable del tamaño.

````python
notas = [3.0, 4.5, 2.0, 4.0]
N = 4
suma = 0
for i in range(N):
    suma = suma + notas[i]
promedio = suma / N
print(f"Promedio: {promedio:.2f}")
````

## Parte 8 — La solución del problema pendiente

Con recorrer y acumular ya se tiene todo lo necesario, así que en clase se resolvió de una vez el problema de la clase 10:

> Leer las notas de los estudiantes de un curso y mostrar cuántas notas son mayores que el promedio.

La solución tiene **dos recorridos** del mismo arreglo: el primero acumula la suma para calcular el promedio, y el segundo **cuenta** las notas mayores que el promedio. Con una sola variable `nota` eso era imposible, porque el segundo recorrido necesita volver a ver todas las notas.

### Con `for`

````python
# Solucion

notas = [3.0, 4.5, 2.0, 4.0]
N = 4
suma = 0
est_arriba_prom = 0
# Calcular promedio
for i in range(N):
    suma = suma + notas[i]
promedio = suma / N
print(f"Promedio: {promedio:.2f}")
# Arriba del promedio
for i in range(N):
    if notas[i] > promedio:
        est_arriba_prom += 1
print(f"Estudiantes arriba del promedio: {est_arriba_prom}")
````

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=%23%20Solucion%0A%0Anotas%20%3D%20%5B3.0%2C%204.5%2C%202.0%2C%204.0%5D%0AN%20%3D%204%0Asuma%20%3D%200%0Aest_arriba_prom%20%3D%200%0A%23%20Calcular%20promedio%0Afor%20i%20in%20range%28N%29%3A%0A%20%20%20%20suma%20%3D%20suma%20%2B%20notas%5Bi%5D%0Apromedio%20%3D%20suma%20%2F%20N%0Aprint%28f%22Promedio%3A%20%7Bpromedio%3A.2f%7D%22%29%0A%23%20Arriba%20del%20promedio%0Afor%20i%20in%20range%28N%29%3A%0A%20%20%20%20if%20notas%5Bi%5D%20%3E%20promedio%3A%0A%20%20%20%20%20%20%20%20est_arriba_prom%20%2B%3D%201%0Aprint%28f%22Estudiantes%20arriba%20del%20promedio%3A%20%7Best_arriba_prom%7D%22%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

> **Nota:** en la clase, la condición se escribió `notas[i] >= promedio`. El enunciado pide las notas **mayores** que el promedio, así que la condición correcta es `>`, y así quedó en esta celda y en la siguiente. Con estos datos las dos condiciones dan 2, porque ninguna nota es igual al promedio (3.375). La celda 🔍 de más abajo muestra un caso donde sí cambia el resultado.

### Con `while`

El segundo recorrido, escrito con `while`. Los tres componentes del ciclo de la clase 7 se ven por separado: la inicialización `k = 0`, la condición `k < N` y la actualización `k += 1`.

````python
# Solucion

notas = [3.0, 4.5, 2.0, 4.0]
N = 4
suma = 0
est_arriba_prom = 0
# Calcular promedio
for i in range(N):
    suma = suma + notas[i]
promedio = suma / N
print(f"Promedio: {promedio:.2f}")
# Arriba del promedio
k = 0
while k < N:
    if notas[k] > promedio:
        est_arriba_prom += 1
    k += 1
print(f"Estudiantes arriba del promedio: {est_arriba_prom}")
````

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=%23%20Solucion%0A%0Anotas%20%3D%20%5B3.0%2C%204.5%2C%202.0%2C%204.0%5D%0AN%20%3D%204%0Asuma%20%3D%200%0Aest_arriba_prom%20%3D%200%0A%23%20Calcular%20promedio%0Afor%20i%20in%20range%28N%29%3A%0A%20%20%20%20suma%20%3D%20suma%20%2B%20notas%5Bi%5D%0Apromedio%20%3D%20suma%20%2F%20N%0Aprint%28f%22Promedio%3A%20%7Bpromedio%3A.2f%7D%22%29%0A%23%20Arriba%20del%20promedio%0Ak%20%3D%200%0Awhile%20k%20%3C%20N%3A%0A%20%20%20%20if%20notas%5Bk%5D%20%3E%20promedio%3A%0A%20%20%20%20%20%20%20%20est_arriba_prom%20%2B%3D%201%0A%20%20%20%20k%20%2B%3D%201%0Aprint%28f%22Estudiantes%20arriba%20del%20promedio%3A%20%7Best_arriba_prom%7D%22%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

| Componente | `for` | `while` |
|---|---|---|
| Inicialización | la hace `range(N)` | `k = 0`, antes del ciclo |
| Condición | la hace `range(N)` | `k < N` |
| Actualización | la hace `range(N)` | `k += 1`, al final del cuerpo |

Con `while`, si se olvida `k += 1`, el ciclo nunca termina.

### 🔍 Para entender: `>` no es lo mismo que `>=`

Con estas notas, el promedio es exactamente 3.0, y dos notas son **iguales** al promedio:

````python
notas = [3.0, 3.0, 4.0, 2.0]
N = 4
suma = 0
for i in range(N):
    suma = suma + notas[i]
promedio = suma / N
mayores = 0
mayores_o_iguales = 0
for i in range(N):
    if notas[i] > promedio:
        mayores = mayores + 1
    if notas[i] >= promedio:
        mayores_o_iguales = mayores_o_iguales + 1
print(f"Promedio: {promedio:.2f}")
print("Con >: ", mayores)
print("Con >=:", mayores_o_iguales)
````

Con `>` solo cuenta el 4.0; con `>=` también cuentan los dos 3.0. Un caso de prueba con una nota **igual** al promedio es un caso **límite**: es el que distingue las dos condiciones. Con los datos de clase no aparece, por eso el error no se notaba.

### 📝 En el tablero: variantes del recorrido (diapositiva 24)

Para cualquier recorrido hay que responder tres preguntas: **desde dónde, hasta dónde y de a cuánto**. Eso define el inicio, el fin y el paso. Con `N = 6`:

| Recorrido | Pseudocódigo | Python |
|---|---|---|
| Completo | `Para (i = 0, N - 1, 1)` | `range(N)` |
| Inverso | `Para (i = N - 1, 0, -1)` | `range(N - 1, -1, -1)` |
| Desde la segunda | `Para (i = 1, N - 1, 1)` | `range(1, N)` |
| Hasta la penúltima | `Para (i = 0, N - 2, 1)` | `range(N - 1)` |
| Posiciones pares | `Para (i = 0, N - 1, 2)` | `range(0, N, 2)` |
| Posiciones impares | `Para (i = 1, N - 1, 2)` | `range(1, N, 2)` |
| Primera mitad | `Para (i = 0, N // 2 - 1, 1)` | `range(N // 2)` |
| Segunda mitad | `Para (i = N // 2, N - 1, 1)` | `range(N // 2, N)` |

### 🔍 Para entender: comprobar la tabla

`list(range(...))` muestra todos los valores que toma `i`, sin escribir el ciclo:

````python
N = 6
print("Completo:          ", list(range(N)))
print("Inverso:           ", list(range(N - 1, -1, -1)))
print("Desde la segunda:  ", list(range(1, N)))
print("Hasta la penultima:", list(range(N - 1)))
print("Posiciones pares:  ", list(range(0, N, 2)))
print("Posiciones impares:", list(range(1, N, 2)))
print("Primera mitad:     ", list(range(N // 2)))
print("Segunda mitad:     ", list(range(N // 2, N)))
````

## Resumen

| Idea | En código |
|---|---|
| Un arreglo guarda muchos valores del mismo tipo bajo un solo nombre | `notas = [3.0, 4.5, 2.0, 4.0]` |
| Se crea con un tamaño fijo y un valor inicial | `notas = [0.0] * N` |
| A cada elemento se llega por su posición, de 0 a `N - 1` | `notas[0]`, `notas[N - 1]` |
| La clave es usar el índice como variable dentro de un ciclo | `for i in range(N): ... notas[i] ...` |
| Llenar, mostrar y acumular son recorridos | `notas[i] = ...`, `print(notas[i])`, `suma = suma + notas[i]` |
| Como las notas quedan guardadas, se pueden recorrer otra vez | el problema de la clase 10, con dos recorridos |
| Un índice demasiado grande da `IndexError`; uno negativo no da error | `notas[4]` frente a `notas[-1]` |
