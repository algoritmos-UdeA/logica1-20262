```python
# Solucion

notas = [3.0, 4.5, 2.0, 4.0]
N = 4
suma = 0
est_arriba_prom = 0
# Calcular promedio
for i in range(N):
    suma = suma + notas[i]
promedio = suma / N
print(f"Promedio: {promedio:.2f}")
# Arriba del promedio
for i in range(N):
    if notas[i] >= promedio:
        est_arriba_prom += 1
print(f"Estudiantes arriba del promedio: {est_arriba_prom}")
```

```python
# Solucion

notas = [3.0, 4.5, 2.0, 4.0]
N = 4
suma = 0
est_arriba_prom = 0
# Calcular promedio
for i in range(N):
    suma = suma + notas[i]
promedio = suma / N
print(f"Promedio: {promedio:.2f}")
# Arriba del promedio
k = 0
while k < N:
    if notas[k] >= promedio:
        est_arriba_prom += 1
    k += 1
print(f"Estudiantes arriba del promedio: {est_arriba_prom}")
```
