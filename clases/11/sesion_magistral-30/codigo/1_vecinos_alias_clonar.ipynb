## 1. Comparar vecinos (diapositiva 25)

> **Problema:** un estudiante tuvo estas notas en sus cuatro evaluaciones, en orden: `[3.0, 4.5, 2.0, 4.0]`. ¿Su nota subió de una evaluación a la siguiente?

Para responder se compara cada nota con la **siguiente**: `notas[i]` con `notas[i + 1]`. Con 4 notas hay solo 3 comparaciones (la 0 con la 1, la 1 con la 2, la 2 con la 3), así que el recorrido va **hasta la penúltima** posición: `range(N - 1)`.

| Pseudocódigo | Python |
|---|---|
| `Para (i = 0, N - 2, 1) Haga` | `for i in range(N - 1):` |

```python
notas = [3.0, 4.5, 2.0, 4.0]
N = 4
for i in range(N - 1):
    if notas[i + 1] > notas[i]:
        print("Sube")
    else:
        print("Baja o se mantiene")
```

## 2. Asignar una lista no la copia (diapositiva 26)

Primero, con variables simples:

```python
a = 3.0
b = a
b = 5.0
print("a =", a)
print("b =", b)
```

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=a%20%3D%203.0%0Ab%20%3D%20a%0Ab%20%3D%205.0%0Aprint%28%22a%20%3D%22%2C%20a%29%0Aprint%28%22b%20%3D%22%2C%20b%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

`b = a` copia el **valor**: después, cambiar `b` no cambia `a`.

Ahora lo mismo con una lista. Prediga: después de cambiar `copia[0]`, ¿cuánto vale `notas[0]`?

```python
notas = [3.0, 4.5, 2.0, 4.0]
copia = notas
copia[0] = 5.0
print("notas =", notas)
print("copia =", copia)
```

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=notas%20%3D%20%5B3.0%2C%204.5%2C%202.0%2C%204.0%5D%0Acopia%20%3D%20notas%0Acopia%5B0%5D%20%3D%205.0%0Aprint%28%22notas%20%3D%22%2C%20notas%29%0Aprint%28%22copia%20%3D%22%2C%20copia%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

¡`notas` también cambió! Con listas, `copia = notas` **no** crea una lista nueva: le pone **otro nombre** a la misma lista. A ese segundo nombre se le llama **alias**. Lo que se cambie con un nombre se ve con el otro, porque hay una sola lista.

## 3. Clonar una lista (diapositivas 27-28)

**Clonar** es crear una lista nueva y copiarle los valores. Así cada nombre tiene su propia lista.

### Con un recorrido

Es la forma que funciona en cualquier lenguaje, y la del pseudocódigo:

| Pseudocódigo | Python |
|---|---|
| `copia = Arreglo(N)`<br>`Para (i = 0, N - 1, 1) Haga`<br>`    copia[i] = notas[i]`<br>`Fin_Para` | `copia = [0.0] * N`<br>`for i in range(N):`<br>`    copia[i] = notas[i]` |

```python
notas = [3.0, 4.5, 2.0, 4.0]
N = 4
copia = [0.0] * N
for i in range(N):
    copia[i] = notas[i]
copia[0] = 5.0
print("notas =", notas)
print("copia =", copia)
```

🔎 [Ver paso a paso en Python Tutor](https://pythontutor.com/visualize.html#code=notas%20%3D%20%5B3.0%2C%204.5%2C%202.0%2C%204.0%5D%0AN%20%3D%204%0Acopia%20%3D%20%5B0.0%5D%20%2A%20N%0Afor%20i%20in%20range%28N%29%3A%0A%20%20%20%20copia%5Bi%5D%20%3D%20notas%5Bi%5D%0Acopia%5B0%5D%20%3D%205.0%0Aprint%28%22notas%20%3D%22%2C%20notas%29%0Aprint%28%22copia%20%3D%22%2C%20copia%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

Ahora `notas` **no** cambió: `copia` es otra lista, creada con `[0.0] * N`. En Python Tutor se ven dos cajas distintas.

### Con `notas[:]` o `list(notas)`

Python tiene dos formas cortas de hacer lo mismo:

* `notas[:]` es la **rebanada completa**: toma todas las casillas, desde el inicio hasta el final, y forma una lista nueva.
* `list(notas)` es una función que crea una lista nueva con los mismos valores.

```python
notas = [3.0, 4.5, 2.0, 4.0]
copia1 = notas[:]
copia2 = list(notas)
copia1[0] = 5.0
copia2[1] = 1.0
print("notas  =", notas)
print("copia1 =", copia1)
print("copia2 =", copia2)
```

Las tres listas son distintas: cambiar una copia no cambia las demás.

## 4. Resumen de operaciones con arreglos (diapositiva 29)

| Operación | Pseudocódigo | Python |
|---|---|---|
| Crear con `N` casillas | `A = Arreglo(N)` | `A = [valor_inicial] * N` |
| Crear con valores | `A = [v0, v1, v2]` | `A = [v0, v1, v2]` |
| Guardar en una casilla | `A[i] = expresión` | `A[i] = expresión` |
| Leer desde el teclado | `Leer(A[i])` | `A[i] = float(input())` |
| Mostrar una casilla | `Escribir(A[i])` | `print(A[i])` |
| Recorrer completo | `Para (i = 0, N - 1, 1) Haga` | `for i in range(N):` |
| Recorrer inverso | `Para (i = N - 1, 0, -1) Haga` | `for i in range(N - 1, -1, -1):` |
| Recorrer por elemento | — | `for x in A:` |
| Comparar con el vecino | `Para (i = 0, N - 2, 1) Haga` … `A[i]` … `A[i + 1]` | `for i in range(N - 1):` … `A[i]` … `A[i + 1]` |
| Clonar | `B = Arreglo(N)` y un recorrido con `B[i] = A[i]` | `B = A[:]` o `B = list(A)` |
| Último elemento | `A[N - 1]` | `A[N - 1]` o `A[-1]` |

**Observación:** toda operación sobre un arreglo completo es un **recorrido**: un ciclo que hace algo con `A[i]`, donde `i` va desde un inicio, hasta un fin y de a un paso.
