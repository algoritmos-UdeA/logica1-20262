## Caso 1: la función recibe una lista y devuelve un valor (diapositivas 31-33)

### Definición

* La función recibe la lista completa como **un solo parámetro**: `notas`.
* **No conoce `N`**: con `len(notas)` sabe cuántos elementos tiene. Así sirve para listas de cualquier tamaño.
* Su *Garantiza* dice "No modifica notas": la función solo lee la lista.

```python
def calcular_promedio(notas):
    """
    Supone:
        notas es una lista de numeros con al menos un elemento.
    Garantiza:
        Retorna el promedio de las notas (float). No modifica notas.
    """
    suma = 0
    for i in range(len(notas)):
        suma = suma + notas[i]
    promedio = suma / len(notas)
    return promedio
```

Esta celda solo **define** la función: no imprime nada. La función se ejecuta cuando se la **invoca**.

### Invocación

Al invocar se escribe solo el **nombre** de la lista, sin corchetes:

```python
lista_notas = [3.0, 4.5, 2.0, 4.0]
prom = calcular_promedio(lista_notas)
print(f"Promedio: {prom:.2f}")
```

### Pruebas

## Caso 2: la función devuelve una lista (diapositivas 34-35)

### Definición

* La función **crea** una lista nueva, la llena y la devuelve con `return`.
* Recibe un número (`n`, cuántas notas leer) y entrega una lista: es el inverso del Caso 1.
* Aquí **sí** se usa `n` en el `range`, porque la lista todavía no existe: la función recibe el tamaño para poder crearla.

```python
def leer_notas(n):
    """
    Supone:
        n es un entero mayor que 0.
    Garantiza:
        Retorna una lista nueva con las n notas leidas, en el orden
        en que se escribieron.
    """
    notas = [0.0] * n
    for i in range(n):
        notas[i] = float(input(f"Nota del estudiante {i + 1}: "))
    return notas
```

### Invocación

El resultado se guarda en una variable. Al ejecutar la celda, escriba una nota y presione **Enter**, cuatro veces (por ejemplo: 3.0, 4.5, 2.0 y 4.0).

```python
lista_notas = leer_notas(4)
print(lista_notas)
```

`leer_notas` usa `input()`, así que **no se prueba con `assert`**: las funciones de entrada y salida se prueban ejecutando el programa.

## Caso 3: la función modifica la lista que recibe (diapositivas 36-38)

### Definición

* La función **cambia las casillas** de la lista que recibe y no devuelve nada.
* Su *Garantiza* debe decirlo claramente: **MODIFICA** la lista. Compárelo con el Caso 1, que decía "No modifica notas".

```python
NOTA_MAXIMA = 5.0

def aplicar_bonificacion(notas, puntos):
    """
    Supone:
        notas es una lista de numeros entre 0.0 y 5.0, y puntos >= 0.
    Garantiza:
        MODIFICA la lista notas: a cada nota le suma puntos, y si el
        resultado pasa de 5.0, la deja en 5.0. No retorna nada.
    """
    for i in range(len(notas)):
        notas[i] = notas[i] + puntos
        if notas[i] > NOTA_MAXIMA:
            notas[i] = NOTA_MAXIMA
```

Adentro, `notas` es un **alias** de `lista_notas`. Por eso lo que la función cambia en `notas` se ve afuera, aunque no haya `return`. El 4.5 + 0.5 llega justo a 5.0 y se queda ahí.
