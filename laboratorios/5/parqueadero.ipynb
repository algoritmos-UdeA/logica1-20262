![La matriz y sus dos numeraciones](https://raw.githubusercontent.com/algoritmos-UdeA/logica1-20262/main/laboratorios/5/images/matriz_puestos.png)

> ⚠️ Para el usuario, los pisos y los puestos se cuentan **desde 1**. En la matriz, las filas y las columnas se cuentan **desde 0**. En la fila `i`, columna `j`, va el puesto del piso `i + 1`, número `j + 1`. Y `get_puesto(piso, numero)` entrega la casilla `[piso - 1][numero - 1]`.

Una matriz se crea en dos pasos. Esta celda es un ejemplo con números; se puede ejecutar y modificar para practicar.

````python
FILAS = 2
COLUMNAS = 3

matriz = [None] * FILAS                  # una lista con una casilla por fila
for i in range(FILAS):
    matriz[i] = [0] * COLUMNAS           # cada casilla pasa a ser una fila nueva

matriz[0][2] = 7                         # fila 0, columna 2
print(matriz)

suma = 0
for i in range(FILAS):                   # por cada fila
    for j in range(COLUMNAS):            # por cada columna de esa fila
        suma = suma + matriz[i][j]
print(suma)
````

En el parqueadero la matriz se crea igual, pero en cada casilla, en lugar de un `0`, va un objeto `Puesto` **nuevo**.

### Punto de chequeo 4, segunda parte: ver el parqueadero

La celda siguiente tiene las funciones de `pantalla.py`, que **ya vienen hechas**: dibujan el mapa del parqueadero usando solo los métodos públicos de las clases. No hay que modificarla, solo ejecutarla.

````python
def texto_puesto(puesto):
    """
    Supone: puesto es un objeto Puesto.
    Garantiza: retorna "[   ]" si el puesto está libre. Si está ocupado,
    retorna la inicial del tipo de vehículo entre corchetes: "[ C ]" para
    un carro y "[ M ]" para una moto.
    """
    texto = "[   ]"
    if not puesto.esta_libre():
        inicial = puesto.get_vehiculo().get_tipo()[0].upper()
        texto = "[ " + inicial + " ]"
    return texto


def texto_encabezado(puestos_por_piso):
    """
    Supone: 1 <= puestos_por_piso <= 9.
    Garantiza: retorna la línea con el número de cada puesto (1, 2, 3...),
    alineada con los corchetes del mapa.
    """
    texto = "       "
    for numero in range(1, puestos_por_piso + 1):
        if numero > 1:
            texto = texto + " "
        texto = texto + "  " + str(numero) + "  "
    return texto.rstrip()


def texto_piso(parqueadero, piso):
    """
    Supone: parqueadero es un objeto Parqueadero y 1 <= piso <= pisos.
    Garantiza: retorna una línea con todos los puestos de ese piso, por
    ejemplo "Piso 1 [ C ] [   ] [ M ]".
    """
    texto = "Piso " + str(piso)
    for numero in range(1, parqueadero.get_puestos_por_piso() + 1):
        texto = texto + " " + texto_puesto(parqueadero.get_puesto(piso, numero))
    return texto


def texto_mapa(parqueadero):
    """
    Supone: parqueadero es un objeto Parqueadero con máximo 9 pisos y
    máximo 9 puestos por piso.
    Garantiza: retorna el mapa completo: el encabezado, una línea por piso
    (el piso más alto arriba, como en un edificio), la leyenda y cuántos
    puestos libres hay.
    """
    pisos = parqueadero.get_pisos()
    puestos_por_piso = parqueadero.get_puestos_por_piso()
    texto = texto_encabezado(puestos_por_piso) + "\n"
    for piso in range(pisos, 0, -1):
        texto = texto + texto_piso(parqueadero, piso) + "\n"
    texto = texto + "\n"
    texto = texto + "C: carro   M: moto   [   ]: libre\n"
    texto = (texto + "Libres: " + str(parqueadero.contar_libres())
             + " de " + str(pisos * puestos_por_piso))
    return texto


def texto_vehiculos(parqueadero):
    """
    Supone: parqueadero es un objeto Parqueadero.
    Garantiza: retorna una línea por cada vehículo que está en el
    parqueadero, con su placa, su tipo y dónde está. Recorre piso por piso
    y, dentro de cada piso, puesto por puesto. Si no hay ninguno, retorna
    "(no hay vehículos)".
    """
    texto = ""
    for piso in range(1, parqueadero.get_pisos() + 1):
        for numero in range(1, parqueadero.get_puestos_por_piso() + 1):
            puesto = parqueadero.get_puesto(piso, numero)
            if not puesto.esta_libre():
                vehiculo = puesto.get_vehiculo()
                if texto != "":
                    texto = texto + "\n"
                texto = (texto + vehiculo.get_placa() + "  " + vehiculo.get_tipo()
                         + "  piso " + str(piso) + ", puesto " + str(numero))
    if texto == "":
        texto = "(no hay vehículos)"
    return texto


def texto_recibo(recibo):
    """
    Supone: recibo es un objeto Recibo.
    Garantiza: retorna el recibo en cuatro líneas: el vehículo, los
    minutos, las horas que se cobran y el valor.
    """
    vehiculo = recibo.get_vehiculo()
    texto = "--- Recibo ---\n"
    texto = texto + "Vehículo: " + vehiculo.get_placa() + " (" + vehiculo.get_tipo() + ")\n"
    texto = texto + "Tiempo: " + str(recibo.get_minutos()) + " minutos\n"
    texto = texto + "Horas cobradas: " + str(recibo.calcular_horas()) + "\n"
    texto = texto + "Valor: $" + str(recibo.calcular_valor())
    return texto


def mostrar_mapa(parqueadero):
    """Garantiza: imprime el mapa del parqueadero. No retorna nada."""
    print(texto_mapa(parqueadero))


def mostrar_vehiculos(parqueadero):
    """Garantiza: imprime los vehículos del parqueadero. No retorna nada."""
    print(texto_vehiculos(parqueadero))


def mostrar_recibo(recibo):
    """Garantiza: imprime el recibo. No retorna nada."""
    print(texto_recibo(recibo))
````

````python
# texto_puesto: libre, con carro y con moto
puesto = Puesto(1, 1)
assert texto_puesto(puesto) == "[   ]"
puesto.ocupar(Vehiculo("ABC123", Vehiculo.CARRO))
assert texto_puesto(puesto) == "[ C ]"
puesto.liberar()
puesto.ocupar(Vehiculo("XYZ78A", Vehiculo.MOTO))
assert texto_puesto(puesto) == "[ M ]"

# texto_encabezado: los números quedan alineados con los corchetes
assert texto_encabezado(1) == "         1"
assert texto_encabezado(3) == "         1     2     3"

# Un parqueadero vacío de 2 pisos y 5 puestos por piso
parqueadero = Parqueadero(2, 5)
assert texto_piso(parqueadero, 1) == "Piso 1 [   ] [   ] [   ] [   ] [   ]"
assert texto_vehiculos(parqueadero) == "(no hay vehículos)"
assert texto_mapa(parqueadero) == (
    "         1     2     3     4     5\n"
    "Piso 2 [   ] [   ] [   ] [   ] [   ]\n"
    "Piso 1 [   ] [   ] [   ] [   ] [   ]\n"
    "\n"
    "C: carro   M: moto   [   ]: libre\n"
    "Libres: 10 de 10")

# El mismo parqueadero con cinco vehículos
parqueadero.get_puesto(1, 1).ocupar(Vehiculo("ABC123", Vehiculo.CARRO))
parqueadero.get_puesto(1, 2).ocupar(Vehiculo("DEF456", Vehiculo.CARRO))
parqueadero.get_puesto(1, 4).ocupar(Vehiculo("XYZ78A", Vehiculo.MOTO))
parqueadero.get_puesto(2, 2).ocupar(Vehiculo("QWE12B", Vehiculo.MOTO))
parqueadero.get_puesto(2, 5).ocupar(Vehiculo("GHI789", Vehiculo.CARRO))

assert texto_piso(parqueadero, 1) == "Piso 1 [ C ] [ C ] [   ] [ M ] [   ]"
assert texto_piso(parqueadero, 2) == "Piso 2 [   ] [ M ] [   ] [   ] [ C ]"
assert texto_mapa(parqueadero) == (
    "         1     2     3     4     5\n"
    "Piso 2 [   ] [ M ] [   ] [   ] [ C ]\n"
    "Piso 1 [ C ] [ C ] [   ] [ M ] [   ]\n"
    "\n"
    "C: carro   M: moto   [   ]: libre\n"
    "Libres: 5 de 10")
assert texto_vehiculos(parqueadero) == (
    "ABC123  carro  piso 1, puesto 1\n"
    "DEF456  carro  piso 1, puesto 2\n"
    "XYZ78A  moto  piso 1, puesto 4\n"
    "QWE12B  moto  piso 2, puesto 2\n"
    "GHI789  carro  piso 2, puesto 5")

# Al liberar un puesto, el mapa cambia
parqueadero.get_puesto(1, 2).liberar()
assert texto_piso(parqueadero, 1) == "Piso 1 [ C ] [   ] [   ] [ M ] [   ]"

# texto_recibo
recibo = Recibo(Vehiculo("ABC123", Vehiculo.CARRO), 130)
assert texto_recibo(recibo) == (
    "--- Recibo ---\n"
    "Vehículo: ABC123 (carro)\n"
    "Tiempo: 130 minutos\n"
    "Horas cobradas: 3\n"
    "Valor: $9000")

print("Punto de chequeo 4 superado: el parqueadero ya se puede ver en pantalla")
print()
mostrar_mapa(parqueadero)
print()
mostrar_vehiculos(parqueadero)
print()
mostrar_recibo(recibo)
````
